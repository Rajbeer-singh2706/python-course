## 1. The Boolean Type

A `bool` has only two values: `True` and `False`. Every operator in this notebook produces one of them.

```python
# A bool is either True or False (capital T and F)
is_student = True
has_coupon = False

print(is_student)
print(has_coupon)
print(type(is_student))
# Output: True
# Output: False
# Output: <class 'bool'>
```

## 2. Comparison Operators



A comparison looks at two values and answers `True` or `False`.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `7 > 4` | `True` |
| `<` | less than | `2 < 9` | `True` |
| `>=` | greater than or equal to | `5 >= 5` | `True` |
| `<=` | less than or equal to | `3 <= 7` | `True` |

**Remember:** `=` stores a value in a variable, `==` compares two values.

```python
# Comparing numbers
x = 10
y = 20

print("x == y :", x == y)
print("x != y :", x != y)
print("x > y  :", x > y)
print("x < y  :", x < y)
print("x >= 10:", x >= 10)
print("y <= 15:", y <= 15)
# Output: x == y : False
# Output: x != y : True
# Output: x > y  : False
# Output: x < y  : True
# Output: x >= 10: True
# Output: y <= 15: False
```

```python
# Real-life example: Swiggy free delivery needs an order of at least Rs. 399
order_total = 450
minimum_for_free_delivery = 399

print("Free delivery?", order_total >= minimum_for_free_delivery)
# Output: Free delivery? True
```

### Comparing strings

Strings can be compared with `==` and `!=`. The comparison is **case-sensitive**, so `"Mumbai"` and `"mumbai"` are different. Use `.lower()` to ignore case.

```python
city = "Mumbai"

print(city == "Mumbai")
print(city == "mumbai")
print(city.lower() == "mumbai")
print(city != "Delhi")
# Output: True
# Output: False
# Output: True
# Output: True
```

### Comparing different types

A number and a string are never equal, even if they look alike. Cast first, then compare. This matters because `input()` always gives you a string.

```python
print(5 == 5.0)      # int and float compare by value
print(5 == "5")     # number vs string: not equal
print(5 == int("5"))  # cast the string first
# Output: True
# Output: False
# Output: True

user_input = "25"            # simulating input(): always a string
age = int(user_input)        # cast before comparing with numbers
print(age > 18)
# Output: True
```

### Chained comparison

Python lets you write a range check the way you do in maths.

```python
age = 34

print(18 <= age <= 60)    # same as: 18 <= age and age <= 60
print(0 < age < 18)
# Output: True
# Output: False
```

## 3. Logical Operators: `and`, `or`, `not`



Logical operators combine several `True` / `False` answers into one.

| Operator | Meaning |
|----------|---------|
| `A and B` | `True` only when BOTH are `True` |
| `A or B` | `True` when AT LEAST ONE is `True` |
| `not A` | flips `True` to `False` and `False` to `True` |

**Truth table**

| A | B | A and B | A or B |
|---|---|---------|--------|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
print("True and True  :", True and True)
print("True and False :", True and False)
print("True or False  :", True or False)
print("False or False :", False or False)
print("not True       :", not True)
print("not False      :", not False)
# Output: True and True  : True
# Output: True and False : False
# Output: True or False  : True
# Output: False or False : False
# Output: not True       : False
# Output: not False      : True
```

```python
# Combining comparisons with and / or / not
age = 25
has_aadhaar = True
is_banned = False

can_open_account = age >= 18 and has_aadhaar
gets_discount = age < 13 or age > 65
is_active = not is_banned

print("Can open bank account:", can_open_account)
print("Gets discounted ticket:", gets_discount)
print("Account active:", is_active)
# Output: Can open bank account: True
# Output: Gets discounted ticket: False
# Output: Account active: True
```

```python
# Example: Paytm payment check - ALL three must be true
amount = 1500
balance = 5000
is_kyc_done = True

payment_allowed = amount > 0 and amount <= balance and is_kyc_done
print("Payment allowed:", payment_allowed)
# Output: Payment allowed: True

# Same check when KYC is NOT done
is_kyc_done = False
payment_allowed = amount > 0 and amount <= balance and is_kyc_done
print("Payment allowed:", payment_allowed)
# Output: Payment allowed: False
```

### Short-circuit behaviour

Python stops as soon as it knows the answer. In `A and B`, if `A` is `False`, `B` is never checked. This protects you from errors such as dividing by zero.

```python
total_orders = 0
total_spent = 5000

# total_orders != 0 is False, so the division on the right never runs
safe = total_orders != 0 and total_spent / total_orders > 500
print("High-value customer:", safe)
# Output: High-value customer: False
```

## 4. Membership Operators: `in`, `not in`



`in` checks whether a value is present inside a string, list, tuple or set. It replaces long chains of `or`.

```python
metro_cities = ["Mumbai", "Delhi", "Chennai", "Kolkata"]

print("Delhi" in metro_cities)
print("Pune" in metro_cities)
print("Pune" not in metro_cities)
# Output: True
# Output: False
# Output: True
```

```python
# Checking text inside a string
email = "priya.sharma@example.com"

print("@" in email)
print("gmail" in email)
# Output: True
# Output: False

# 'in' instead of a long 'or' chain
day = "Sunday"

print(day == "Saturday" or day == "Sunday")   # long way
print(day in ("Saturday", "Sunday"))          # short way
# Output: True
# Output: True
```

## 5. Identity Operators: `is`, `is not`


- `==` asks: do these have the **same value**?
- `is` asks: are these the **same object** in memory?

For now, remember one rule: **use `is None` and `is not None` to check for "no value".**

```python
coupon = None        # None means "no value yet"

print(coupon is None)
print(coupon is not None)
# Output: True
# Output: False

coupon = "DIWALI20"

print(coupon is None)
print(coupon is not None)
# Output: False
# Output: True
```

```python
# == versus is
list_a = [1, 2, 3]
list_b = [1, 2, 3]   # same values, separate list
list_c = list_a      # another name for the SAME list

print(list_a == list_b)   # same values
print(list_a is list_b)   # different objects
print(list_a is list_c)   # same object
# Output: True
# Output: False
# Output: True
```

## 6. Operator Precedence

When one expression has several operators, Python works in this order (highest first):

1. arithmetic: `+  -  *  /  //  %  **`
2. comparison: `==  !=  >  <  >=  <=  in  not in  is  is not`
3. `not`
4. `and`
5. `or`

`and` runs before `or`. When in doubt, add **parentheses**: they cost nothing and make your intent clear.

```python
a = True
b = False
c = False

print(a or b and c)      # b and c runs first, then a or False
print((a or b) and c)    # parentheses change the order
# Output: True
# Output: False
```

```python
# Real example: Zomato Gold member OR a big order, AND the restaurant must be open
is_gold = True
order_value = 200
restaurant_open = False

without_brackets = is_gold or order_value > 500 and restaurant_open
with_brackets = (is_gold or order_value > 500) and restaurant_open

print("Without brackets:", without_brackets)
print("With brackets   :", with_brackets)
# Output: Without brackets: True
# Output: With brackets   : False
```

## 7. Truthy and Falsy Values


Every Python value can be treated as `True` or `False`.

**Falsy:** `False`, `None`, `0`, `0.0`, `""` (empty string), `[]`, `{}`, `()`

**Truthy:** everything else (any non-zero number, any non-empty string or collection).

`bool(value)` shows how Python sees a value.

```python
print(bool(0))
print(bool(1))
print(bool(-5))
print(bool(""))
print(bool("hello"))
print(bool([]))
print(bool([1, 2]))
print(bool(None))
# Output: False
# Output: True
# Output: True
# Output: False
# Output: True
# Output: False
# Output: True
# Output: False
```

```python
# Handy: check an empty BigBasket cart without comparing to anything
cart = []
print("Cart has items:", bool(cart))
# Output: Cart has items: False

cart = ["Atta", "Milk"]
print("Cart has items:", bool(cart))
# Output: Cart has items: True
```

Meera Iyer applies for a home loan. Write ONE expression stored in `eligible` that is `True` only when ALL of these hold:

- age is between 21 and 60 inclusive (use a chained comparison)
- `monthly_income` is at least 30000
- `credit_score` is at least 700
- `city` is one of Mumbai, Delhi, Bengaluru, Pune, Chennai, Hyderabad

Expected output: `Eligible: True`

## Summary

| Group | Operators | Use it to |
|-------|-----------|-----------|
| Comparison | `==  !=  >  <  >=  <=` | compare two values |
| Chained | `18 <= age <= 60` | check a range |
| Logical | `and  or  not` | combine conditions |

**Common mistakes**
- Writing `=` when you mean `==`
- Comparing a string from `input()` with a number without casting
- Forgetting that `"Mumbai"` and `"mumbai"` are different
- Using `== None` instead of `is None`
