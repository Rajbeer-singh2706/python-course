## 🧠 Concepts

### 1. if / elif / else
The `if` statement evaluates a condition and executes a block of code only if that condition is **True**.
- `elif` (else-if) adds additional conditions to check
- `else` catches everything that didn't match
- The **colon** `:` and **indentation** (4 spaces) are mandatory

```python
if condition:
    # runs when condition is True
elif other_condition:
    # runs when first is False, this one is True
else:
    # runs when ALL above are False
```


````python
# ── Concept 1: if / elif / else ──────────────────────────────────────────

score = 72

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} → Grade: {grade}")

# Try changing score to 95, 85, 55 and re-run this cell
````

### 2. Comparison Operators
Comparison operators always return `True` or `False`:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `7 > 4` | `True` |
| `<` | Less than | `2 < 9` | `True` |
| `>=` | Greater or equal | `5 >= 5` | `True` |
| `<=` | Less or equal | `3 <= 7` | `True` |

⚠️ **Common mistake:** `==` checks equality, `=` assigns a value. Never use `=` inside an `if`.


````python
# ── Concept 2: Comparison operators ──────────────────────────────────────

x = 10
y = 20

print(f"x = {x}, y = {y}")
print(f"x == y : {x == y}")
print(f"x != y : {x != y}")
print(f"x > y  : {x > y}")
print(f"x < y  : {x < y}")
print(f"x >= 10: {x >= 10}")
print(f"y <= 20: {y <= 20}")
````

### 3. Logical Operators: and, or, not
Logical operators combine or invert boolean expressions:

| Operator | Behaviour | Example |
|----------|-----------|---------|
| `and` | Both sides must be `True` | `age >= 18 and has_id` |
| `or` | At least one side must be `True` | `is_admin or is_moderator` |
| `not` | Inverts a boolean | `not is_banned` |

**Operator precedence (high → low):** `not` → `and` → `or`  
Use parentheses `()` when in doubt to make intent explicit.


````python
# ── Concept 3: Logical operators ─────────────────────────────────────────

age = 25
has_id = True
is_vip = False

# and — both conditions must be True
if age >= 18 and has_id:
    print("Entry allowed")

# or — at least one condition must be True
if age < 13 or age > 65:
    print("Discounted ticket")
else:
    print("Full price ticket")

# not — inverts the boolean
if not is_vip:
    print("Regular queue")

# Combining all three — use parentheses for clarity
if (age >= 18 and has_id) or is_vip:
    print("Access granted")
````

### 4. Nested Conditions
An `if` statement can contain another `if` inside it — this is called nesting.

**Rule of thumb:**
- ✅ 1 level of nesting: fine
- ✅ 2 levels: acceptable
- ⚠️ 3+ levels: refactor — either use `and`/`or` or extract a function

```python
if outer_condition:
    if inner_condition:
        # runs when BOTH are True
    else:
        # outer True, inner False
else:
    # outer False (inner never checked)
```


````python
# ── Concept 4: Nested conditions ─────────────────────────────────────────

username = "admin"
password = "secret123"

# Nested version
if username == "admin":
    if password == "secret123":
        print("Welcome, admin! (nested)")
    else:
        print("Wrong password.")
else:
    print("User not found.")

print()  # blank line

# Flattened version using 'and' — often cleaner
if username == "admin" and password == "secret123":
    print("Welcome, admin! (flat)")
elif username == "admin":
    print("Wrong password.")
else:
    print("User not found.")
````

### 5. match / case Statement (Python 3.10+)
`match` provides structural pattern matching — similar to switch/case in other languages, but more powerful.

```python
match value:
    case pattern1:
        ...
    case pattern2 | pattern3:   # OR with |
        ...
    case _:                      # wildcard — matches anything
        ...
```

> 💡 **Version check:** Run `python --version` — you need 3.10 or higher. On older versions, use `if/elif` chains.


````python
# ── Concept 5: match / case ───────────────────────────────────────────────

command = "quit"

match command:
    case "start":
        print("Starting the application...")
    case "stop":
        print("Stopping the application...")
    case "quit" | "exit":       # OR — matches either value
        print("Goodbye!")
    case _:                     # wildcard default
        print(f"Unknown command: {command!r}")

print()

# Matching days of the week
day = "Saturday"

match day:
    case "Saturday" | "Sunday":
        print(f"{day}: Weekend!")
    case "Monday" | "Friday":
        print(f"{day}: Almost there!")
    case _:
        print(f"{day}: Midweek grind")
````

### 6. Truthy and Falsy Values
Every Python value has a boolean interpretation. You can use values **directly** in `if` statements.

**Falsy values** (treated as `False`):
- `False`, `None`, `0`, `0.0`
- `""` (empty string), `[]` (empty list), `{}` (empty dict), `()` (empty tuple), `set()`

**Everything else** is truthy.

```python
name = ""
if name:           # False because name is empty
    ...
if not name:       # True because name is empty
    ...
```

> ⚠️ **Identity check for None:** Use `is None` not `== None` — it's faster and the Python idiom.


````python
# ── Concept 6: Truthy / Falsy values ─────────────────────────────────────

# Empty string is falsy
name = ""
if name:
    print(f"Hello, {name}!")
else:
    print("Name is empty (falsy empty string)")

# Zero is falsy
count = 0
if count:
    print(f"Count: {count}")
else:
    print("Count is zero (falsy zero)")

# None check with 'is'
result = None
if result is None:
    print("No result yet (None)")

# Non-empty string is truthy
name = "Alice"
if name:
    print(f"Hello, {name}!")

# Demonstrate bool() conversion
print()
print("bool conversions:")
for val in [0, 1, "", "hello", None, [], [1, 2], {}, {"a": 1}]:
    print(f"  bool({val!r:12}) = {bool(val)}")
````

### Example 3 — Simple CLI Menu with match/case
A real-world pattern for handling user commands.

````python
# CLI menu simulation
def handle_command(cmd):
    """Process a user command and return the response."""
    match cmd.lower().strip():
        case "help" | "h" | "?":
            return "Commands: help, status, version, quit"
        case "status":
            return "System: Running ✅"
        case "version":
            return "App v1.0.0 — Python 3.12"
        case "quit" | "exit" | "q":
            return "Shutting down. Goodbye!"
        case "":
            return "Please enter a command. Type 'help' for options."
        case _:
            return f"Unknown command: {cmd!r}. Type 'help' for options."

# Simulate a session
for user_input in ["help", "STATUS", "unknown", "quit"]:
    response = handle_command(user_input)
    print(f"> {user_input}")
    print(f"  {response}")
    print()
````

## 🚀 Mini-Project — CLI Calculator Menu

Build a simple calculator that:
1. Accepts two numbers and an operation as variables
2. Uses `match / case` to route to the correct calculation
3. Handles edge cases: division by zero, unknown operation
4. Prints a nicely formatted result

**Operations to support:** `add`, `subtract`, `multiply`, `divide`, `power`


````python
# ── Mini-Project: CLI Calculator Menu ─────────────────────────────────────
a = 12
b = 4
operation = "divide"

# YOUR CODE HERE
# Expected output for divide: 12 ÷ 4 = 3.0
````

````python
# ── Mini-Project: CLI Calculator Menu — SOLUTION ──────────────────────────
a = 12
b = 4
operation = "divide"

match operation:
    case "add":
        result = a + b
        symbol = "+"
    case "subtract":
        result = a - b
        symbol = "-"
    case "multiply":
        result = a * b
        symbol = "×"
    case "divide":
        if b == 0:
            print("Error: Division by zero is undefined")
            result = None
            symbol = "÷"
        else:
            result = a / b
            symbol = "÷"
    case "power":
        result = a ** b
        symbol = "^"
    case _:
        print(f"Unknown operation: {operation!r}")
        result = None
        symbol = "?"

if result is not None:
    print(f"{a} {symbol} {b} = {result}")
````
