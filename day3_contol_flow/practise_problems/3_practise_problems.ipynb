### Problem 6 🟡 — Electricity Bill Tier
Calculate an electricity bill based on tiered pricing:
- First 100 units: $0.05/unit
- Next 200 units (101–300): $0.10/unit  
- Above 300 units: $0.15/unit

Print the total bill.


```python
units_used = 350

# YOUR CODE HERE
# Expected output for 350 units: Total bill: $32.50
```

```python
units_used = 350

if units_used <= 100:
    bill = units_used * 0.05
elif units_used <= 300:
    bill = (100 * 0.05) + ((units_used - 100) * 0.10)
else:
    bill = (100 * 0.05) + (200 * 0.10) + ((units_used - 300) * 0.15)

print(f"Total bill: ${bill:.2f}")
```

### Problem 11 🔴 — Triangle Validator
Given three side lengths, determine if they form a valid triangle and, if so, what type:
- **Valid triangle rule:** Each side must be less than the sum of the other two
- **Equilateral:** All three sides equal
- **Isosceles:** Exactly two sides equal
- **Scalene:** All sides different


```python
a = 5
b = 5
c = 8

# YOUR CODE HERE
# Expected output: Valid Isosceles triangle
```

```python
a = 5
b = 5
c = 8

if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        triangle_type = "Equilateral"
    elif a == b or b == c or a == c:
        triangle_type = "Isosceles"
    else:
        triangle_type = "Scalene"
    print(f"Valid {triangle_type} triangle")
else:
    print(f"Not a valid triangle: sides {a}, {b}, {c}")
```

### Problem 12 🟢 — HTTP Status Router
Use `match / case` to print a description for common HTTP status codes:
- 200 → "OK"
- 301 → "Moved Permanently"
- 400 → "Bad Request"
- 401 → "Unauthorised"
- 403 → "Forbidden"
- 404 → "Not Found"
- 500 → "Internal Server Error"
- Anything else → "Unknown status"


```python
status_code = 404

# YOUR CODE HERE  (use match / case)
# Expected output: 404 → Not Found
```

```python
status_code = 404

match status_code:
    case 200:
        description = "OK"
    case 301:
        description = "Moved Permanently"
    case 400:
        description = "Bad Request"
    case 401:
        description = "Unauthorised"
    case 403:
        description = "Forbidden"
    case 404:
        description = "Not Found"
    case 500:
        description = "Internal Server Error"
    case _:
        description = "Unknown status"

print(f"{status_code} → {description}")
```

### Problem 14 🟡 — Card Game Suit Router
Use `match / case` with `|` to group card suits into colours:
- Hearts, Diamonds → "Red suit"
- Clubs, Spades → "Black suit"
- Joker → "Wild card — no suit"
- Anything else → "Invalid suit"


```python
suit = "Hearts"

# YOUR CODE HERE  (use match / case with |)
# Expected output: Hearts → Red suit
```

```python
suit = "Hearts"

match suit:
    case "Hearts" | "Diamonds":
        colour = "Red suit"
    case "Clubs" | "Spades":
        colour = "Black suit"
    case "Joker":
        colour = "Wild card — no suit"
    case _:
        colour = "Invalid suit"

print(f"{suit} → {colour}")
```

### Problem 15 🔴 — Config Validator
Given a config dictionary, validate that required keys are present and non-empty (truthy).
Required keys: `"host"`, `"port"`, `"database"`

- Print "Config valid" if all are present and truthy
- Print each missing or empty key on its own line if not


```python
config = {
    "host": "localhost",
    "port": 5432,
    "database": "",     # deliberately empty
    "debug": True,
}

required_keys = ["host", "port", "database"]

# YOUR CODE HERE
# Expected output:
# Config error — 'database' is missing or empty
```

```python
config = {
    "host": "localhost",
    "port": 5432,
    "database": "",
    "debug": True,
}

required_keys = ["host", "port", "database"]
errors = []

for key in required_keys:
    if key not in config or not config[key]:
        errors.append(key)

if errors:
    for key in errors:
        print(f"Config error — {key!r} is missing or empty")
else:
    print("Config valid")
```

### Problem 16 🟡 — Simple ATM
Simulate an ATM transaction. Given `balance`, `pin_entered`, `correct_pin`, and `withdrawal_amount`:
1. Check if the PIN is correct (use `==`)
2. If correct, check if `withdrawal_amount` is positive and ≤ balance
3. If both pass, subtract from balance and print new balance
4. Handle each failure with a specific error message


```python
balance = 1500.00
correct_pin = "1234"
pin_entered = "1234"
withdrawal_amount = 200.00

# YOUR CODE HERE
# Expected output:
# PIN accepted.
# Dispensing $200.00. New balance: $1300.00
```

```python
balance = 1500.00
correct_pin = "1234"
pin_entered = "1234"
withdrawal_amount = 200.00

if pin_entered != correct_pin:
    print("Incorrect PIN. Transaction cancelled.")
elif withdrawal_amount <= 0:
    print("Invalid amount. Please enter a positive value.")
elif withdrawal_amount > balance:
    print(f"Insufficient funds. Your balance is ${balance:.2f}.")
else:
    balance -= withdrawal_amount
    print("PIN accepted.")
    print(f"Dispensing ${withdrawal_amount:.2f}. New balance: ${balance:.2f}")
```

### Problem 18 🔴 — Movie Rating System
Build a combined rating classifier. Given `age`, `has_adult_supervision`, and `rating`:
- "G" → All ages
- "PG" → All ages, parental guidance suggested (no block)
- "PG-13" → Must be 13+, OR under 13 with adult supervision
- "R" → Must be 17+, OR 13–16 with adult supervision
- "NC-17" → Must be 18+, NO exceptions

Print whether the person can watch the film, and a reason if not.


```python
age = 15
has_adult_supervision = True
rating = "R"

# YOUR CODE HERE
# Expected output: Can watch R — under 17 but has adult supervision
```

```python
age = 15
has_adult_supervision = True
rating = "R"

match rating:
    case "G" | "PG":
        allowed = True
        reason = f"All ages permitted for {rating}"
    case "PG-13":
        if age >= 13:
            allowed = True
            reason = f"Age {age} meets PG-13 requirement"
        elif has_adult_supervision:
            allowed = True
            reason = f"Under 13 but has adult supervision"
        else:
            allowed = False
            reason = "Must be 13+ or have adult supervision"
    case "R":
        if age >= 17:
            allowed = True
            reason = f"Age {age} meets R requirement"
        elif age >= 13 and has_adult_supervision:
            allowed = True
            reason = f"Under 17 but has adult supervision"
        else:
            allowed = False
            reason = "Must be 17+, or 13–16 with adult supervision"
    case "NC-17":
        if age >= 18:
            allowed = True
            reason = f"Age {age} meets NC-17 requirement"
        else:
            allowed = False
            reason = "Must be 18+ — no exceptions for NC-17"
    case _:
        allowed = False
        reason = f"Unknown rating: {rating!r}"

status = "Can watch" if allowed else "Cannot watch"
print(f"{status} {rating} — {reason}")
```
