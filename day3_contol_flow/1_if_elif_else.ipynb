## SECTION 1 - The simple `if` statement

````python
# ── Concept: if ──
# An `if` statement runs a block of code ONLY when its condition is True.
# If the condition is False, Python skips the block and moves on.
#
# Syntax:
#     if condition:
#         # indented block (4 spaces) - runs only when condition is True
#
# Rules you must remember:
#   1. The line ends with a colon  :
#   2. The block is indented by 4 spaces (indentation IS the syntax in Python)
#   3. The condition is anything that evaluates to True or False

print("SECTION 1 - simple if")
print("-" * 50)

temperature = 38  # degrees Celsius in Delhi

if temperature > 35:
    print("It is very hot today. Drink water!")
# Output: It is very hot today. Drink water!

# This line is NOT indented, so it runs no matter what the condition was.
print("This line always runs.")
# Output: This line always runs.

# When the condition is False, the block is skipped entirely.
temperature = 22

if temperature > 35:
    print("This will NOT print because 22 is not greater than 35.")

print("Nothing was printed by the if above.")
# Output: Nothing was printed by the if above.
````

````python
# ── Concept: multiple statements inside one if-block ──
# Every indented line belongs to the block.
balance = 5000  # Priya's Paytm wallet balance in rupees

if balance >= 1000:
    print("Wallet check passed.")
    print("You can book a train ticket on IRCTC.")
    print("Balance available: Rs.", balance)
# Output: Wallet check passed.
# Output: You can book a train ticket on IRCTC.
# Output: Balance available: Rs. 5000
````

## SECTION 2 - if ... else

````python
# ── Concept: if-else ──
# `else` gives you a "plan B". Exactly ONE of the two blocks always runs:
#   - the `if` block when the condition is True
#   - the `else` block when the condition is False
#
# Syntax:
#     if condition:
#         # runs when True
#     else:
#         # runs when False
#
# `else` has NO condition of its own, and it also ends with a colon.

print()
print("SECTION 2 - if / else")
print("-" * 50)

age = 17  # Arjun's age

if age >= 18:
    print("Arjun can vote.")
else:
    print("Arjun cannot vote yet.")
# Output: Arjun cannot vote yet.
````

````python
# ── Concept: even or odd with % (modulo) ──
# `%` gives the remainder after division. A number is even when the
# remainder of dividing by 2 is 0.
number = 27

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
# Output: 27 is odd
````

````python
# ── Concept: checking a password-like value ──
entered_pin = "4821"  # simulating input()
correct_pin = "4821"

if entered_pin == correct_pin:
    print("PIN accepted. Welcome to your Razorpay dashboard.")
else:
    print("Wrong PIN. Try again.")
# Output: PIN accepted. Welcome to your Razorpay dashboard.
````

## SECTION 3 - if ... elif ... else

````python
# ── Concept: elif ──
# `elif` means "else if". Use it when there are MORE than two outcomes.
# Python checks the conditions from top to bottom and runs the block of the
# FIRST condition that is True. After that, it skips all the remaining
# branches - even if they would also be True.
#
# Syntax:
#     if condition_1:
#         ...
#     elif condition_2:
#         ...
#     elif condition_3:
#         ...
#     else:
#         ...      # optional catch-all when nothing above matched
#
# Rules:
#   - You can have as many `elif` blocks as you need
#   - `else` is optional, but if present it must be LAST
#   - ORDER MATTERS: put the most specific / strictest condition first

print()
print("SECTION 3 - if / elif / else")
print("-" * 50)

marks = 78  # Riya's exam marks out of 100

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "Fail"

print("Marks:", marks, "-> Grade:", grade)
# Output: Marks: 78 -> Grade: B
````

````python
# ── Concept: why ORDER matters ──
# 78 is >= 70 AND >= 60 AND >= 40. Only the FIRST matching branch runs.
# If we wrote the checks in the wrong order, we would get the wrong grade.
marks = 78

if marks >= 40:  # WRONG ORDER: this is True first, so it always wins
    wrong_grade = "D"
elif marks >= 70:
    wrong_grade = "B"
else:
    wrong_grade = "Fail"

print("Wrong order gives:", wrong_grade)
# Output: Wrong order gives: D
print("Lesson: put the strictest condition first.")
# Output: Lesson: put the strictest condition first.
````

````python
# ── Concept: elif chain vs separate if statements ──
# Separate `if` statements are ALL checked independently, so more than one
# block can run. An elif chain runs at most ONE block.
score = 85

print("Separate ifs:")
if score >= 80:
    print("  distinction")
if score >= 60:
    print("  first class")
if score >= 40:
    print("  pass")
# Output: Separate ifs:
# Output:   distinction
# Output:   first class
# Output:   pass

print("elif chain:")
if score >= 80:
    print("  distinction")
elif score >= 60:
    print("  first class")
elif score >= 40:
    print("  pass")
# Output: elif chain:
# Output:   distinction
````

## SECTION 4 - One-line conditional expression (ternary)

````python
# ── Concept: value_if_true if condition else value_if_false ──
# A compact form of if-else when you only need to pick between two VALUES.
# Use it for short, simple choices. Use a normal if-else for anything longer.

print()
print("SECTION 4 - conditional expression")
print("-" * 50)

age = 20
status = "adult" if age >= 18 else "minor"
print("Status:", status)
# Output: Status: adult

items_in_cart = 1
label = "item" if items_in_cart == 1 else "items"
print(f"You have {items_in_cart} {label} in your BigBasket cart.")
# Output: You have 1 item in your BigBasket cart.
````

````python
# ── Example 1: Swiggy free delivery (simple if) ──
print("Example 1 - Swiggy free delivery")
print("-" * 50)

order_total = 450  # rupees
delivery_fee = 40

if order_total >= 399:
    delivery_fee = 0
    print("Free delivery unlocked!")

print("Delivery fee: Rs.", delivery_fee)
print("Amount to pay: Rs.", order_total + delivery_fee)
# Output: Free delivery unlocked!
# Output: Delivery fee: Rs. 0
# Output: Amount to pay: Rs. 450
````

````python
# ── Example 2: Metro ticket (if-else) ──
print()
print("Example 2 - Mumbai metro ticket")
print("-" * 50)

is_student = True
full_fare = 40

if is_student:
    fare = full_fare * 0.5  # 50 percent concession
else:
    fare = full_fare

print("Fare: Rs.", fare)
# Output: Fare: Rs. 20.0
````

````python
# ── Example 3: IRCTC age-based concession (if-elif-else) ──
print()
print("Example 3 - IRCTC concession")
print("-" * 50)

passenger_age = 63
base_fare = 1200

if passenger_age < 5:
    fare = 0
    category = "Infant (free)"
elif passenger_age < 12:
    fare = base_fare * 0.5
    category = "Child (half fare)"
elif passenger_age >= 60:
    fare = base_fare * 0.6
    category = "Senior citizen (40% off)"
else:
    fare = base_fare
    category = "Adult (full fare)"

print("Category:", category)
print("Fare: Rs.", fare)
# Output: Category: Senior citizen (40% off)
# Output: Fare: Rs. 720.0
````

````python
# ── Example 4: Tiered electricity bill (if-elif-else with calculation) ──
print()
print("Example 4 - Tiered electricity bill")
print("-" * 50)

# Rate card: first 100 units Rs.3, next 100 units Rs.5, above 200 Rs.8
units = 260  # units used by the Sharma family in Pune

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = (100 * 3) + (units - 100) * 5
else:
    bill = (100 * 3) + (100 * 5) + (units - 200) * 8

print("Units used:", units)
print("Bill: Rs.", bill)
# Output: Units used: 260
# Output: Bill: Rs. 1280
````

````python
# ── Example 5: Zomato order status message (if-elif-else with strings) ──
print()
print("Example 5 - Zomato order status")
print("-" * 50)

order_status = "out_for_delivery"  # simulating a value from the app

if order_status == "placed":
    message = "Restaurant has received your order."
elif order_status == "preparing":
    message = "Your food is being prepared."
elif order_status == "out_for_delivery":
    message = "Your delivery partner is on the way!"
elif order_status == "delivered":
    message = "Enjoy your meal!"
else:
    message = "Unknown status. Please contact support."

print(message)
# Output: Your delivery partner is on the way!
````
