## SECTION 1 - Comparison operators

````python
# ── Concept: comparison operators ──
# A comparison compares two values and ALWAYS produces a bool: True or False.
# These bools are what `if` statements use to make decisions.
#
#   ==   equal to                 5 == 5     -> True
#   !=   not equal to             5 != 3     -> True
#   >    greater than             7 > 4      -> True
#   <    less than                2 < 9      -> True
#   >=   greater than or equal    5 >= 5     -> True
#   <=   less than or equal       3 <= 7     -> True
#
# Warning: `=` ASSIGNS a value, `==` COMPARES two values.

print("SECTION 1 - comparison operators")
print("-" * 50)

x = 10
y = 20

print("x == y :", x == y)
print("x != y :", x != y)
print("x > y  :", x > y)
print("x < y  :", x < y)
print("x >= 10:", x >= 10)
print("y <= 15:", y <= 15)
# Output: x == y : False
# Output: x != y : True
# Output: x > y  : False
# Output: x < y  : True
# Output: x >= 10: True
# Output: y <= 15: False
````

````python
# ── Concept: comparing strings ──
# Strings compare by equality, and also alphabetically (dictionary order).
# Comparison is case-sensitive: "Mumbai" and "mumbai" are different.
city = "Mumbai"

print("city == 'Mumbai':", city == "Mumbai")
print("city == 'mumbai':", city == "mumbai")
print("city.lower() == 'mumbai':", city.lower() == "mumbai")
print("'Delhi' < 'Mumbai':", "Delhi" < "Mumbai")
# Output: city == 'Mumbai': True
# Output: city == 'mumbai': False
# Output: city.lower() == 'mumbai': True
# Output: 'Delhi' < 'Mumbai': True
````

````python
# ── Concept: comparing different types ──
# int and float can be compared with each other.
# A string and a number are NOT equal, and < or > between them is a TypeError.
print("5 == 5.0 :", 5 == 5.0)
print("5 == '5' :", 5 == "5")
# Output: 5 == 5.0 : True
# Output: 5 == '5' : False
````

````python
# ── Concept: floating-point surprise ──
# Floats are stored approximately, so never test them with == directly.
print("0.1 + 0.2 == 0.3 :", 0.1 + 0.2 == 0.3)
print("abs((0.1 + 0.2) - 0.3) < 1e-9 :", abs((0.1 + 0.2) - 0.3) < 1e-9)
# Output: 0.1 + 0.2 == 0.3 : False
# Output: abs((0.1 + 0.2) - 0.3) < 1e-9 : True
````

## SECTION 2 - Chained comparisons

````python
# ── Concept: a < b < c ──
# Python lets you chain comparisons like in maths.
# `18 <= age <= 60` means  `18 <= age and age <= 60`  - clean and readable.

print()
print("SECTION 2 - chained comparisons")
print("-" * 50)

age = 34

print("18 <= age <= 60 :", 18 <= age <= 60)
print("age < 18 or age > 60 :", age < 18 or age > 60)
# Output: 18 <= age <= 60 : True
# Output: age < 18 or age > 60 : False

marks = 72
if 60 <= marks < 75:
    print("Second class")
# Output: Second class
````

## SECTION 3 - Logical operators: and, or, not

````python
# ── Concept: and / or / not ──
#   A and B  -> True only if BOTH are True
#   A or B   -> True if AT LEAST ONE is True
#   not A    -> flips True to False and False to True
#
# Truth table:
#     A      B      A and B   A or B
#     True   True   True      True
#     True   False  False     True
#     False  True   False     True
#     False  False  False     False

print()
print("SECTION 3 - logical operators")
print("-" * 50)

print("True and True  :", True and True)
print("True and False :", True and False)
print("True or False  :", True or False)
print("False or False :", False or False)
print("not True       :", not True)
print("not False      :", not False)
# Output: True and True  : True
# Output: True and False : False
# Output: True or False  : True
# Output: False or False : False
# Output: not True       : False
# Output: not False      : True
````

````python
# ── Concept: using them in real conditions ──
age = 25
has_aadhaar = True
is_banned = False

if age >= 18 and has_aadhaar:
    print("Eligible to open a bank account.")
# Output: Eligible to open a bank account.

if age < 13 or age > 65:
    print("Discounted ticket")
else:
    print("Full price ticket")
# Output: Full price ticket

if not is_banned:
    print("Account is active.")
# Output: Account is active.
````

## SECTION 4 - Short-circuit evaluation

````python
# ── Concept: Python stops as soon as the answer is known ──
#   A and B -> if A is False, B is never evaluated (result must be False)
#   A or B  -> if A is True,  B is never evaluated (result must be True)
#
# This is useful for guarding against errors, for example division by zero.

print()
print("SECTION 4 - short-circuit evaluation")
print("-" * 50)

total_orders = 0
total_spent = 5000

# If total_orders is 0, the division on the right is never executed.
if total_orders != 0 and total_spent / total_orders > 500:
    print("High-value customer")
else:
    print("Not enough data (no division performed)")
# Output: Not enough data (no division performed)
````

````python
# ── Concept: and / or actually return one of their operands ──
# `or` returns the first truthy value, `and` returns the first falsy value
# (or the last value if all are truthy). `or` is handy for default values.
nickname = ""
display_name = nickname or "Guest"
print("Display name:", display_name)
# Output: Display name: Guest
````

## SECTION 5 - Membership operators: in, not in

````python
# ── Concept: in / not in ──
# `in` checks whether a value is present inside a string, list, tuple,
# set or dictionary (keys). The result is True or False.

print()
print("SECTION 5 - membership operators")
print("-" * 50)

metro_cities = ["Mumbai", "Delhi", "Chennai", "Kolkata"]

print("'Delhi' in metro_cities    :", "Delhi" in metro_cities)
print("'Pune' in metro_cities     :", "Pune" in metro_cities)
print("'Pune' not in metro_cities :", "Pune" not in metro_cities)
# Output: 'Delhi' in metro_cities    : True
# Output: 'Pune' in metro_cities     : False
# Output: 'Pune' not in metro_cities : True

# in with strings checks for a substring
email = "priya.sharma@example.com"
print("'@' in email :", "@" in email)
print("email.endswith('.com') :", email.endswith(".com"))
# Output: '@' in email : True
# Output: email.endswith('.com') : True

# `in` is a clean replacement for many `or` chains.
day = "Sunday"

if day in ("Saturday", "Sunday"):  # instead of: day == "Saturday" or day == "Sunday"
    print("Weekend!")
# Output: Weekend!
````

## SECTION 6 - Identity operators: is, is not

````python
# ── Concept: == versus is ──
#   ==  asks: do these two have the SAME VALUE?
#   is  asks: are these the SAME OBJECT in memory?
#
# The one place you should always use `is` is checking for None.

print()
print("SECTION 6 - identity operators")
print("-" * 50)

list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

print("list_a == list_b :", list_a == list_b)
print("list_a is list_b :", list_a is list_b)
print("list_a is list_c :", list_a is list_c)
# Output: list_a == list_b : True
# Output: list_a is list_b : False
# Output: list_a is list_c : True

coupon = None  # no coupon applied yet

if coupon is None:
    print("No coupon applied.")
# Output: No coupon applied.

coupon = "DIWALI20"

if coupon is not None:
    print("Coupon applied:", coupon)
# Output: Coupon applied: DIWALI20
````

## SECTION 7 - Operator precedence

````python
# ── Concept: which operator runs first? ──
# From highest to lowest priority:
#     1. arithmetic       + - * / // % **
#     2. comparison       == != > < >= <=  in  not in  is  is not
#     3. not
#     4. and
#     5. or
#
# `and` binds tighter than `or`. When in doubt, ADD PARENTHESES - they cost
# nothing and make your intent obvious to the next reader.

print()
print("SECTION 7 - operator precedence")
print("-" * 50)

a = True
b = False
c = True

print("a or b and c   :", a or b and c)
print("(a or b) and c :", (a or b) and c)
print("a or (b and c) :", a or (b and c))
# Output: a or b and c   : True
# Output: (a or b) and c : True
# Output: a or (b and c) : True

# A case where it genuinely changes the answer:
a = False
b = True
c = False

print("a or b and c   :", a or b and c)    # b and c runs first -> False; then a or False
print("(a or b) and c :", (a or b) and c)  # (a or b) -> True; then True and c
# Output: a or b and c   : False
# Output: (a or b) and c : False

a = True
b = False
c = False

print("a or b and c   :", a or b and c)    # b and c -> False; then True or False
print("(a or b) and c :", (a or b) and c)  # True and False
# Output: a or b and c   : True
# Output: (a or b) and c : False
````

## SECTION 8 - Truthy and falsy values

````python
# ── Concept: every value has a boolean meaning ──
# Falsy (treated as False):  False, None, 0, 0.0, "", [], {}, (), set()
# Truthy: everything else - any non-zero number, any non-empty string or
# collection.
#
# That means you can write  `if name:`  instead of  `if name != "":`.

print()
print("SECTION 8 - truthy and falsy")
print("-" * 50)

for value in [0, 1, -5, 0.0, "", "hello", [], [1, 2], None]:
    print(f"bool({value!r}) = {bool(value)}")
# Output: bool(0) = False
# Output: bool(1) = True
# Output: bool(-5) = True
# Output: bool(0.0) = False
# Output: bool('') = False
# Output: bool('hello') = True
# Output: bool([]) = False
# Output: bool([1, 2]) = True
# Output: bool(None) = False

cart = []  # empty BigBasket cart

if cart:
    print("Proceed to checkout")
else:
    print("Your cart is empty")
# Output: Your cart is empty

# Careful: 0 is falsy, but 0 can be a perfectly valid value.
# Use `is None` when you only want to detect "no value".
wallet_balance = 0

if not wallet_balance:
    print("Falsy check: balance is zero OR missing (ambiguous)")
if wallet_balance is not None:
    print("None check: balance is present, it just happens to be", wallet_balance)
# Output: Falsy check: balance is zero OR missing (ambiguous)
# Output: None check: balance is present, it just happens to be 0
````

````python
# ── Example 1: Leap year (and / or with parentheses) ──
print("Example 1 - leap year")
print("-" * 50)

year = 2024

# Divisible by 4, but century years must also be divisible by 400.
is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

print(year, "is a leap year:", is_leap)
# Output: 2024 is a leap year: True
````

````python
# ── Example 2: Flipkart sale eligibility (chained and / in) ──
print()
print("Example 2 - Flipkart sale eligibility")
print("-" * 50)

cart_value = 2500
is_plus_member = True
category = "electronics"
sale_categories = ["electronics", "fashion", "home"]

eligible = cart_value >= 2000 and category in sale_categories and is_plus_member

if eligible:
    discount = cart_value * 0.10
    print("Sale discount: Rs.", discount)
else:
    print("Not eligible for the sale discount.")
# Output: Sale discount: Rs. 250.0
````

````python
# ── Example 3: Paytm transaction validator (not, or, short-circuit) ──
print()
print("Example 3 - Paytm transaction validator")
print("-" * 50)

amount = 15000
balance = 12000
is_kyc_done = True
daily_limit = 20000

if amount <= 0:
    print("Invalid amount.")
elif not is_kyc_done:
    print("Complete KYC to continue.")
elif amount > balance or amount > daily_limit:
    print("Transaction declined: insufficient balance or limit exceeded.")
else:
    print("Transaction successful.")
# Output: Transaction declined: insufficient balance or limit exceeded.
````

````python
# ── Example 4: IRCTC Tatkal booking check (chained comparison + in) ──
print()
print("Example 4 - IRCTC Tatkal booking window")
print("-" * 50)

hour = 10  # 24-hour clock
travel_class = "3A"

ac_classes = ("1A", "2A", "3A", "CC")

if travel_class in ac_classes and hour == 10:
    window = "AC Tatkal window is open (10:00 AM)"
elif travel_class not in ac_classes and hour == 11:
    window = "Sleeper Tatkal window is open (11:00 AM)"
else:
    window = "Tatkal window is closed for this class right now"

print(window)
# Output: AC Tatkal window is open (10:00 AM)
````

````python
# ── Example 5: Password strength (many operators together) ──
print()
print("Example 5 - password strength checker")
print("-" * 50)

password = "Mumbai@2024"

long_enough = len(password) >= 8
has_digit = any(ch.isdigit() for ch in password)
has_upper = any(ch.isupper() for ch in password)
has_symbol = any(ch in "!@#$%^&*" for ch in password)
not_common = password.lower() not in ("password", "12345678", "qwerty123")

checks_passed = sum([long_enough, has_digit, has_upper, has_symbol, not_common])

if checks_passed == 5:
    strength = "Strong"
elif checks_passed >= 3:
    strength = "Medium"
else:
    strength = "Weak"

print("Checks passed:", checks_passed, "of 5")
print("Strength:", strength)
# Output: Checks passed: 5 of 5
# Output: Strength: Strong
````
