```python
student_name = "Rajbeer"
marks = 35
grade = ""

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "Fail"
print("Student:", student_name, "Marks:", marks, "-> Grade:", grade)
```

```python
month = "July"

match month:
    case "December" | "January" | "February":
        season = "Winter"
    case "March" | "April" | "May":
        season = "Summer"
    case "June" | "July" | "August" | "September":
        season = "Monsoon"
    case _:
        season = "Post-Monsoon"

print(f"{month} falls in the {season} season in India.")
```

```python
order_amount = 1800

match order_amount:
    case amount if amount > 2000:
        discount = 0.20 
    case amount if amount > 1000:
        discount = 0.10
    case _:
        discount = 0.0

print(f"Order Amount: {order_amount}, Discount: {discount * 100}%")
```
