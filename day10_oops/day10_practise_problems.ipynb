Define a `Circle` class:

- `area(self)` → `pi * r^2`  (import `math` and use `math.pi`)
- `circumference(self)` → `2 * pi * r`

````python
import math
# YOUR CODE HERE
````

````python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def circumference(self):
        return 2 * math.pi * self.radius

    def is_larger_than(self, other):
        return self.area() > other.area()

c1 = Circle(5)
c2 = Circle(3)
print(f'Circle 1 area          : {c1.area():.2f}')
print(f'Circle 1 circumference : {c1.circumference():.2f}')
print(f'Circle 2 area          : {c2.area():.2f}')
print(f'c1 larger than c2?     : {c1.is_larger_than(c2)}')
````

### 🔴 C4 — Classroom Tracker

Build a `Classroom` class that tracks students across all classrooms:
- **Class variables:** `total_students = 0`, `all_classrooms = []`
- **Instance variables:** `room_name`, `students` (list, starts empty)
- `enroll(self, student_name)` — adds to `self.students`, increments `total_students`, appends room to `all_classrooms` if not already there
- `roll_call(self)` — prints room name and all enrolled students
- `classmethod` (bonus): `summary()` — prints total students across all classrooms

Create two classrooms, enroll a few students, then call `roll_call()` on each and `Classroom.summary()`.


````python
# YOUR CODE HERE
````

````python
class Classroom:
    total_students = 0
    all_classrooms = []

    def __init__(self, room_name):
        self.room_name = room_name
        self.students  = []
        Classroom.all_classrooms.append(room_name)

    def enroll(self, student_name):
        self.students.append(student_name)
        Classroom.total_students += 1

    def roll_call(self):
        print(f'Room: {self.room_name}')
        for i, s in enumerate(self.students, 1):
            print(f'  {i}. {s}')

    @classmethod
    def summary(cls):
        print(f'Total students across {len(cls.all_classrooms)} rooms: {cls.total_students}')

r1 = Classroom('Room 101')
r2 = Classroom('Room 102')
r1.enroll('Priya')
r1.enroll('Arjun')
r2.enroll('Riya')
r1.roll_call()
r2.roll_call()
Classroom.summary()
````

### 🟢 D1 — Stopwatch

Define a `Stopwatch` class:
- `__init__(self)` — sets `seconds = 0`
- `tick(self, n=1)` — adds `n` to `seconds`
- `reset(self)` — sets `seconds` back to `0`
- `display(self)` — prints time as `MM:SS` (e.g. `01:45`)

Simulate: tick 125 seconds, display, reset, display again.


````python
# YOUR CODE HERE
````

````python
class Stopwatch:
    def __init__(self):
        self.seconds = 0

    def tick(self, n=1):
        self.seconds += n

    def reset(self):
        self.seconds = 0

    def display(self):
        m, s = divmod(self.seconds, 60)
        print(f'{m:02d}:{s:02d}')

sw = Stopwatch()
sw.tick(125)
sw.display()   # 02:05
sw.reset()
sw.display()   # 00:00
````

- `show(self)` — prints all pending tasks numbered from 1
- `count(self)` → returns number of remaining tasks

Test with Priya's todo list: add 4 tasks, complete one, show remaining.


````python
# YOUR CODE HERE
````

````python
class TodoList:
    def __init__(self, owner):
        self.owner = owner
        self.tasks = []

    def add(self, task):
        self.tasks.append(task)

    def complete(self, task):
        if task in self.tasks:
            self.tasks.remove(task)
            print(f'Completed: {task}')
        else:
            print('Task not found')

    def show(self):
        print(f"{self.owner}'s Tasks ({self.count()} remaining):")
        for i, t in enumerate(self.tasks, 1):
            print(f'  {i}. {t}')

    def count(self):
        return len(self.tasks)

todo = TodoList('Priya')
todo.add('Buy groceries')
todo.add('Study Python Day 10')
todo.add('Call Arjun')
todo.add('Submit assignment')
todo.complete('Call Arjun')
todo.show()
````

Define a `Card` class:

- `face_name(self)` → returns `'Ace'` for 1, `'Jack'` for 11, `'Queen'` for 12, `'King'` for 13, else the number as string

Test with cards: Ace of Spades, 7 of Hearts, Queen of Diamonds.


````python
# YOUR CODE HERE
````

````python
class Card:
    def __init__(self, suit, value):
        self.suit  = suit
        self.value = value

    def face_name(self):
        names = {1: 'Ace', 11: 'Jack', 12: 'Queen', 13: 'King'}
        return names.get(self.value, str(self.value))

    def display(self):
        print(f'{self.face_name()} of {self.suit}')

    def is_face_card(self):
        return self.value >= 11

c1 = Card('Spades',   1)
c2 = Card('Hearts',   7)
c3 = Card('Diamonds', 12)
for c in [c1, c2, c3]:
    c.display()
    print(f'  Face card: {c.is_face_card()}')
````

### 🔴 D4 — Inventory Manager

Define an `Inventory` class for a small shop:
- `__init__(self, shop_name)` — `items` = empty dict `{name: (price, quantity)}`
- `add_stock(self, name, price, quantity)` — add new item or increase existing stock
- `sell(self, name, quantity)` — decrease stock; print error if insufficient or not found
- `total_value(self)` → sum of `price * quantity` for all items
- `low_stock_alert(self, threshold=5)` → prints all items where quantity <= threshold

Simulate a day of stock management for 'Sharma General Store'.


````python
# YOUR CODE HERE
````

````python
class Inventory:
    def __init__(self, shop_name):
        self.shop_name = shop_name
        self.items     = {}

    def add_stock(self, name, price, quantity):
        if name in self.items:
            old_p, old_q = self.items[name]
            self.items[name] = (old_p, old_q + quantity)
        else:
            self.items[name] = (price, quantity)
        print(f'Stock added: {name} x{quantity}')

    def sell(self, name, quantity):
        if name not in self.items:
            print(f'Item not found: {name}')
            return
        price, stock = self.items[name]
        if quantity > stock:
            print(f'Insufficient stock for {name}: only {stock} left')
        else:
            self.items[name] = (price, stock - quantity)
            print(f'Sold {quantity}x {name}')

    def total_value(self):
        return sum(p * q for p, q in self.items.values())

    def low_stock_alert(self, threshold=5):
        low = [(n, q) for n, (p, q) in self.items.items() if q <= threshold]
        if low:
            print('-- Low Stock Alert --')
            for name, qty in low:
                print(f'  {name}: {qty} units')
        else:
            print('All stock levels OK')

inv = Inventory('Sharma General Store')
inv.add_stock('Rice (5kg)', 350, 20)
inv.add_stock('Dal (1kg)',  120, 8)
inv.add_stock('Oil (1L)',   180, 4)
inv.sell('Rice (5kg)', 17)
inv.sell('Oil (1L)', 10)
print(f'Total inventory value: Rs {inv.total_value()}')
inv.low_stock_alert()
````

### 🟡 E1 — Student Grade Book

Define a `Student` class:
- `__init__(self, name, roll_number)` — `scores` = empty dict `{subject: score}`
- `add_score(self, subject, score)` — add subject score (0–100)
- `average(self)` → mean of all scores; return `0` if no scores
- `grade(self)` → based on average: A(>=90), B(>=75), C(>=60), D(>=45), F(<45)
- `report(self)` → print a formatted report card

Test with Priya who scored: Maths 88, Science 92, English 76, History 65.


````python
# YOUR CODE HERE
````

````python
class Student:
    def __init__(self, name, roll_number):
        self.name        = name
        self.roll_number = roll_number
        self.scores      = {}

    def add_score(self, subject, score):
        self.scores[subject] = score

    def average(self):
        if not self.scores:
            return 0
        return sum(self.scores.values()) / len(self.scores)

    def grade(self):
        avg = self.average()
        if avg >= 90: return 'A'
        if avg >= 75: return 'B'
        if avg >= 60: return 'C'
        if avg >= 45: return 'D'
        return 'F'

    def report(self):
        print(f'Report Card: {self.name} (Roll: {self.roll_number})')
        print('-' * 35)
        for subj, score in self.scores.items():
            print(f'  {subj:<12}: {score}')
        print('-' * 35)
        print(f'  Average   : {self.average():.1f}')
        print(f'  Grade     : {self.grade()}')

priya = Student('Priya Sharma', 'CS2024-01')
priya.add_score('Maths',   88)
priya.add_score('Science', 92)
priya.add_score('English', 76)
priya.add_score('History', 65)
priya.report()
````

### 🟡 E2 — ATM Machine

Build an `ATM` class that simulates a basic ATM session:
- `__init__(self, owner, balance, pin)` — stores all three
- `_authenticate(self, entered_pin)` → returns `True` if pin matches (prefix `_` = internal method)
- `check_balance(self, pin)` — if auth OK: print balance; else: print 'Wrong PIN'
- `withdraw(self, amount, pin)` — auth, then check funds, then deduct
- `deposit(self, amount)` — no auth needed, just add and confirm

Simulate: Arjun deposits Rs 5000, checks balance, withdraws Rs 2000 (correct PIN), then tries to withdraw Rs 8000 (insufficient).


````python
# Simulate PIN as a pre-assigned variable instead of input()
correct_pin = '1234'
wrong_pin   = '0000'
# YOUR CODE HERE
````

````python
correct_pin = '1234'
wrong_pin   = '0000'

class ATM:
    def __init__(self, owner, balance, pin):
        self.owner   = owner
        self.balance = balance
        self._pin    = pin

    def _authenticate(self, entered_pin):
        return entered_pin == self._pin

    def check_balance(self, pin):
        if self._authenticate(pin):
            print(f'Balance: Rs {self.balance:,}')
        else:
            print('Wrong PIN. Access denied.')

    def withdraw(self, amount, pin):
        if not self._authenticate(pin):
            print('Wrong PIN.')
            return
        if amount > self.balance:
            print(f'Insufficient funds. Balance: Rs {self.balance:,}')
        else:
            self.balance -= amount
            print(f'Rs {amount:,} dispensed. New balance: Rs {self.balance:,}')

    def deposit(self, amount):
        self.balance += amount
        print(f'Rs {amount:,} deposited. New balance: Rs {self.balance:,}')

atm = ATM('Arjun Mehta', 10_000, correct_pin)
atm.deposit(5_000)
atm.check_balance(correct_pin)
atm.withdraw(2_000, correct_pin)
atm.withdraw(8_000, correct_pin)
atm.check_balance(wrong_pin)
````

### 🔴 E3 — Fleet Manager

Design two classes that work together:

**`Car` class:**
- `__init__(self, reg_number, make, model, fuel_level=100)`
- `drive(self, km)` — consumes 8L per 100km; reduce fuel; raise `ValueError` if not enough fuel
- `refuel(self, litres)` — add fuel, cap at 100L
- `status(self)` → prints reg, make/model, fuel level

**`Fleet` class:**
- `__init__(self, company)` — `cars` = empty list
- `add_car(self, car)` — appends a Car object
- `fleet_status(self)` — calls `status()` on every car
- `cars_needing_fuel(self, threshold=20)` → returns list of Car objects with fuel <= threshold

Create a fleet of 3 cars, drive them varying distances, then report which need refuelling.


````python
# YOUR CODE HERE
````

````python
class Car:
    CONSUMPTION = 8   # litres per 100 km

    def __init__(self, reg_number, make, model, fuel_level=100):
        self.reg_number = reg_number
        self.make       = make
        self.model      = model
        self.fuel_level = fuel_level

    def drive(self, km):
        needed = km * Car.CONSUMPTION / 100
        if needed > self.fuel_level:
            raise ValueError(f'{self.reg_number}: not enough fuel for {km} km')
        self.fuel_level -= needed
        print(f'{self.reg_number} drove {km} km. Fuel left: {self.fuel_level:.1f} L')

    def refuel(self, litres):
        self.fuel_level = min(100, self.fuel_level + litres)
        print(f'{self.reg_number} refuelled. Fuel: {self.fuel_level:.1f} L')

    def status(self):
        print(f'{self.reg_number} | {self.make} {self.model} | Fuel: {self.fuel_level:.1f} L')


class Fleet:
    def __init__(self, company):
        self.company = company
        self.cars    = []

    def add_car(self, car):
        self.cars.append(car)

    def fleet_status(self):
        print(f'-- {self.company} Fleet Status --')
        for car in self.cars:
            car.status()

    def cars_needing_fuel(self, threshold=20):
        return [c for c in self.cars if c.fuel_level <= threshold]


c1 = Car('MH01AB1234', 'Maruti', 'Swift')
c2 = Car('MH02CD5678', 'Hyundai', 'i20')
c3 = Car('MH03EF9012', 'Tata', 'Tiago')

fleet = Fleet('Mumbai Cabs Pvt Ltd')
fleet.add_car(c1)
fleet.add_car(c2)
fleet.add_car(c3)

c1.drive(600)   # uses 48L -> 52L left
c2.drive(900)   # uses 72L -> 28L left
c3.drive(950)   # uses 76L -> 24L left

fleet.fleet_status()

low = fleet.cars_needing_fuel(30)
print(f'Cars needing fuel (<= 30L): {[c.reg_number for c in low]}')
````

````python
class Scoreboard:
    high_score = 0

    def __init__(self, player_name):
        self.player_name          = player_name
        self.score                = 0
        self.questions_attempted  = 0
        self._correct_count       = 0

    def answer(self, correct):
        self.questions_attempted += 1
        if correct:
            self.score += 10
            self._correct_count += 1
        else:
            self.score -= 5

    def percentage(self):
        if self.questions_attempted == 0:
            return 0.0
        return round(self._correct_count / self.questions_attempted * 100, 1)

    def finish(self):
        print(f'Player    : {self.player_name}')
        print(f'Score     : {self.score}')
        print(f'Accuracy  : {self.percentage()}%')
        if self.score > Scoreboard.high_score:
            Scoreboard.high_score = self.score
            print('New high score!')
        print()

# Priya: 4 correct, 1 wrong -> score = 35
priya = Scoreboard('Priya')
for correct in [True, True, False, True, True]:
    priya.answer(correct)
priya.finish()

# Arjun: 5 correct -> score = 50
arjun = Scoreboard('Arjun')
for correct in [True, True, True, True, True]:
    arjun.answer(correct)
arjun.finish()

print(f'All-time high score: {Scoreboard.high_score}')
````
