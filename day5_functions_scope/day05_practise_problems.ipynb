````python
# SOLUTION
def collatz(n):
    """Return the full Collatz sequence from n to 1."""
    sequence = [n]
    while n != 1:
        n = n // 2 if n % 2 == 0 else n * 3 + 1
        sequence.append(n)
    return sequence

def collatz_length(n):
    """Return the number of steps in the Collatz sequence from n."""
    return len(collatz(n))

print(collatz(6))           # [6, 3, 10, 5, 16, 8, 4, 2, 1]
print(collatz(1))           # [1]
print(collatz_length(27))   # 112

longest_n = max(range(1, 21), key=collatz_length)
print(f"Longest sequence 1-20: starts at {longest_n} ({collatz_length(longest_n)} steps)")
# Longest: starts at 18 (21 steps)
````

````python
# SOLUTION
def html_tag(tag, content, **attributes):
    """Build an HTML element string."""
    # Build attribute string
    attr_parts = []
    for key, value in attributes.items():
        # class_ → class  (Python keyword workaround)
        attr_name = key.rstrip("_")
        attr_parts.append(f'{attr_name}="{value}"')
    attr_str = (" " + " ".join(attr_parts)) if attr_parts else ""
    return f"<{tag}{attr_str}>{content}</{tag}>"

print(html_tag("a", "Click here", href="https://python.org", target="_blank"))
print(html_tag("p", "Hello world", class_="intro", id="first"))
print(html_tag("img", "", src="photo.jpg", alt="My photo", width="200"))
````

````python
# SOLUTION
def register(**functions):
    """
    Create a function registry.

    Returns:
        A callable run(name, *args) that dispatches to registered functions.
    """
    def run(name, *args):
        if name not in functions:
            return f"Error: '{name}' not registered"
        return functions[name](*args)
    return run

registry = register(
    double=lambda x: x * 2,
    square=lambda x: x ** 2,
    greet=lambda name: f"Hello, {name}!",
)

print(registry("double", 5))       # 10
print(registry("square", 4))       # 16
print(registry("greet", "Raj"))    # Hello, Raj!
print(registry("unknown", 1))      # Error: 'unknown' not registered
````

````python
# SOLUTION
def pipeline(*functions):
    """
    Return a function that applies each function in sequence.

    Each function's return value becomes the next function's input.
    """
    def execute(value):
        result = value
        for func in functions:
            result = func(result)
        return result
    return execute

process = pipeline(
    str.strip,
    str.lower,
    lambda s: s.replace(" ", "_"),
    lambda s: s + ".txt",
)

print(process("  Hello World  "))         # hello_world.txt
print(process("  Python Course  "))       # python_course.txt
print(process("  Functions and Scope  ")) # functions_and_scope.txt

number_process = pipeline(
    lambda x: x * 2,
    lambda x: x + 10,
    lambda x: x ** 2,
)
print(number_process(5))   # 400  →  (5*2=10, 10+10=20, 20^2=400)
````

````python
# SOLUTION
def caesar_cipher(text, shift=3, mode="encode"):
    """Encode or decode text using the Caesar cipher."""
    if mode == "decode":
        shift = -shift
    result = []
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            shifted = (ord(char) - base + shift) % 26 + base
            result.append(chr(shifted))
        else:
            result.append(char)
    return "".join(result)

def crack_caesar(ciphertext):
    """Try all 25 shifts and return the most likely plaintext."""
    common_words = {"the", "and", "for", "are", "but", "not", "you",
                    "all", "can", "had", "her", "was", "one", "our",
                    "out", "day", "get", "has", "him", "his", "how",
                    "its", "may", "new", "now", "old", "see", "two",
                    "way", "who", "boy", "did", "man", "end", "why"}
    best_score, best_text = -1, ""
    for shift in range(1, 26):
        attempt = caesar_cipher(ciphertext, shift, mode="decode")
        words = attempt.lower().split()
        score = sum(1 for w in words if w.strip(".,!?") in common_words)
        if score > best_score:
            best_score, best_text = score, attempt
    return best_text

encoded = caesar_cipher("Hello, World!", shift=3)
print("Encoded:", encoded)                              # Khoor, Zruog!

decoded = caesar_cipher(encoded, shift=3, mode="decode")
print("Decoded:", decoded)                              # Hello, World!

secret = caesar_cipher("Functions and scope are the foundation of Python!", shift=13)
print("Secret:", secret)
print("Cracked:", crack_caesar(secret))
# Functions and scope are the foundation of Python!
````
