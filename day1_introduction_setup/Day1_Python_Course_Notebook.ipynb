## 🧠 Concepts

### Concept 1 — What is Python?

Python is a **high-level, interpreted programming language** created by Guido van Rossum in 1991.

| Property | What it means |
|----------|---------------|
| **High-level** | You write human-readable code; Python handles memory, types, etc. |
| **Interpreted** | Code runs line-by-line — no compilation step needed |
| **Dynamically typed** | You don't declare variable types — Python figures them out |
| **General purpose** | Web, data science, automation, AI, scripting — all of it |

**Why Python 3.12 specifically?**
- Improved error messages (easier debugging)
- `match` statement (pattern matching — Day 3)
- Faster execution than earlier 3.x versions
- `@override` decorator support (Day 17)


````python
# ── Concept 1: Python basics ──────────────────────────

# Python uses print() to display output
print('Hello from Python 3!')

# Check which version is running
import sys
print(f'Python version: {sys.version}')

# Python reads like English
name = 'World'
if name == 'World':
    print(f'Hello, {name}!')
````

### Concept 3 — Virtual Environments

A **virtual environment** is an isolated Python installation for a single project.

```
Your Computer
├── System Python 3.12
├── project-a/
│   └── .venv/  ← Project A's private Python + packages
└── project-b/
    └── .venv/  ← Project B's private Python + packages
```

**Why this matters:**
- Project A needs `requests==2.28` — Project B needs `requests==2.31`
- Without environments, they conflict. With environments, no problem.
- Professional Python development **always** uses virtual environments.

**With `uv` (what we use):**
```bash
uv init          # initialise project
uv venv          # create .venv environment
uv add pandas    # install a package into the environment
```


````python
# ── Concept 3: Check if we're in a virtual environment ──

import sys
import os

# Check if a virtual environment is active
in_venv = sys.prefix != sys.base_prefix
print(f'Running inside a virtual environment: {in_venv}')
print(f'Python executable: {sys.executable}')
print(f'Python prefix: {sys.prefix}')

# If in_venv is False, activate your .venv before opening this notebook!
# Command: source .venv/bin/activate (Mac/Linux) or .venv\Scripts\activate (Windows)
````

### Concept 4 — Your First Python Function

A **function** is a reusable block of code with a name. You define it once, call it many times.

```python
def function_name(parameter):   # def = define
    # indented body (4 spaces)
    return result               # return sends a value back
```

**Key parts:**
| Part | Example | Meaning |
|------|---------|--------|
| `def` | `def greet` | Keyword that starts a function definition |
| Name | `greet` | What you'll call to use it |
| Parameter | `(name: str)` | Input the function receives |
| Type hint | `: str` / `-> str` | Optional — says what types go in/out |
| `return` | `return message` | Sends a value back to the caller |

> **Deep dive:** Functions are covered fully on **Day 5**. Today we're just getting comfortable with the syntax.


````python
# ── Concept 4: Defining and calling a function ──────────

# Define the function
def greet(name: str) -> str:
    """Return a greeting string for the given name."""
    message = f'Hello, {name}! Welcome to Python 3.12.'
    return message

# Call the function
result = greet('World')
print(result)

# Call it again with a different argument
print(greet('Python Learner'))
print(greet('Day 1'))
````

### Concept 5 — f-strings (Formatted String Literals)

An **f-string** lets you embed variables and expressions directly inside a string.

```python
name = 'Alice'
age  = 30

# Old way (don't use this)
print('Hello ' + name + ', you are ' + str(age))

# f-string way (use this)
print(f'Hello {name}, you are {age}')
```

The `f` prefix before the quote activates f-string mode. Anything in `{}` is evaluated as Python.

> **Deep dive:** String formatting is covered fully on **Day 7**.


````python
# ── Concept 5: f-strings in action ─────────────────────

name = 'Python'
version = 3.12
year = 2024

# Basic f-string
print(f'Welcome to {name} {version}!')

# Expressions inside {}
print(f'2 + 2 = {2 + 2}')
print(f'Uppercase: {name.upper()}')

# Formatting numbers
pi = 3.14159265
print(f'Pi to 2 decimal places: {pi:.2f}')
print(f'Pi to 4 decimal places: {pi:.4f}')

# Multi-line f-string
summary = (
    f'Language : {name}\n'
    f'Version  : {version}\n'
    f'Year     : {year}'
)
print(summary)
````

### Concept 6 — The `__name__` Guard

You'll see this at the bottom of almost every Python script:

```python
if __name__ == '__main__':
    main()
```

**What it does:** Python sets `__name__` to `'__main__'` when you **run** a file directly.  
When a file is **imported** by another file, `__name__` is set to the module name instead.

```
Run directly:  python hello.py   → __name__ == '__main__'  → code runs ✓
Imported by:   import hello      → __name__ == 'hello'     → code skipped ✓
```

This means your script can be both **runnable** and **importable** without side effects.


````python
# ── Concept 6: __name__ guard ───────────────────────────

# In a Jupyter notebook, __name__ is NOT '__main__'
# (because notebooks are interactive, not run as scripts)
print(f'__name__ in this notebook: {__name__!r}')

# Demonstration of how the guard would work in a .py file:
def simulate_script_run(name_value):
    """Simulate what happens when __name__ has different values."""
    if name_value == '__main__':
        print('Script is being run directly → executing main code')
    else:
        print(f'Script imported as module "{name_value}" → skipping main code')

simulate_script_run('__main__')   # As if run directly
simulate_script_run('hello')      # As if imported
````

#### Example 3 — Building a More Useful Greeting Function

Let's extend the `greet` function step by step to show how functions can grow.


````python
# ── Example 3: Extending a function ─────────────────────

# Version 1: Simple
def greet_v1(name):
    return f'Hello, {name}!'

# Version 2: With a default argument (preview of Day 5)
def greet_v2(name, language='English'):
    greetings = {
        'English': 'Hello',
        'Hindi':   'Namaste',
        'Spanish': 'Hola',
        'French':  'Bonjour',
    }
    word = greetings.get(language, 'Hello')  # fallback to 'Hello'
    return f'{word}, {name}!'

# Version 3: With multiple return values (preview of Day 5)
def greet_v3(name, language='English'):
    message = greet_v2(name, language)
    char_count = len(message)
    return message, char_count

# Test all three versions
print('--- Version 1 ---')
print(greet_v1('Alice'))

print('\n--- Version 2 ---')
print(greet_v2('Alice'))
print(greet_v2('Ravi', 'Hindi'))
print(greet_v2('Carlos', 'Spanish'))

print('\n--- Version 3 ---')
msg, count = greet_v3('Alice', 'French')
print(f'Message : {msg}')
print(f'Length  : {count} characters')
````

#### Example 4 — Running a System Check Script

This is a practical example: a script that checks if the environment is set up correctly. You could run this at the start of every session as a quick sanity check.


````python
# ── Example 4: Environment health check ─────────────────

import sys
import os

def check_environment():
    """Run a series of environment checks and report results."""
    checks = []

    # Check 1: Python version
    v = sys.version_info
    ok = v.major == 3 and v.minor >= 12
    checks.append(('Python 3.12+', ok, f'{v.major}.{v.minor}.{v.micro}'))

    # Check 2: Virtual environment
    in_venv = sys.prefix != sys.base_prefix
    checks.append(('Virtual environment active', in_venv, sys.prefix if in_venv else 'NOT active'))

    # Check 3: sys module available
    checks.append(('sys module', True, 'available'))

    # Check 4: os module available
    checks.append(('os module', True, 'available'))

    # Print report
    print('=' * 50)
    print('  Python Environment Health Check')
    print('=' * 50)
    for name, passed, detail in checks:
        icon = '✅' if passed else '❌'
        print(f'{icon}  {name:<35} {detail}')
    print('=' * 50)
    all_ok = all(p for _, p, _ in checks)
    if all_ok:
        print('  All checks passed — you are good to go! 🚀')
    else:
        print('  Some checks failed — review the setup steps.')
    print('=' * 50)

check_environment()
````

#### Example 5 — f-strings: Everything You Can Do

A comprehensive look at f-string features you'll use throughout the course.


````python
# ── Example 5: f-string feature tour ────────────────────

# Basic variable insertion
language = 'Python'
version  = 3.12
print(f'{language} {version}')

# Expressions inside {}
a, b = 10, 3
print(f'{a} + {b} = {a + b}')
print(f'{a} × {b} = {a * b}')
print(f'{a} / {b} = {a / b:.2f}')   # .2f = 2 decimal places

# String methods inside {}
name = 'alice'
print(f'Name: {name.title()}')        # Alice
print(f'Upper: {name.upper()}')       # ALICE
print(f'Length: {len(name)} chars')   # 5

# Padding and alignment
print(f'|{"left":<10}|')              # left-align in 10 chars
print(f'|{"right":>10}|')             # right-align
print(f'|{"center":^10}|')            # center

# Debug mode (Python 3.8+): variable name + value
x = 42
print(f'{x = }')                      # x = 42
````
