# The PaisaTracker codebase



Now we help Priya. We build the menu in **three small stages**. Each stage adds one thing to the last.

| Stage | What it adds |
|-------|--------------|
| 1 | The menu loop (no data yet) |
| 2 | Real expenses: add, view, total |
| 3 | Safe input and a limit warning (the finished v0 menu) |

We use one list called `typed` to pretend the user is typing, in the same order as a real session.

### **Stage 1: The menu loop skeleton**


A `running` flag controls the loop. Choosing **4** sets `running = False`, and the loop ends.

````python
typed = ["2", "9", "1", "4"]            # pretend the user chooses: 2, 9, 1, 4

running = True
while running:
    print()
    print("1) Add expense   2) View expenses   3) Show total   4) Exit")
    choice = typed.pop(0)
    print("You chose:", choice)

    if choice == "1":
        print("(adding an expense - coming in stage 2)")
    elif choice == "2":
        print("(showing expenses - coming in stage 2)")
    elif choice == "3":
        print("(showing total - coming in stage 2)")
    elif choice == "4":
        running = False
        print("Goodbye!")
    else:
        print("Invalid choice. Please type 1, 2, 3 or 4.")
````

**What to notice:** choice `9` printed an error but the menu came back. Only `4` ends the program.

### **Stage 2: Real expenses**

We keep expenses in a **list of dicts** (Day 4). 
* Option 1(Menu Item): asks for name, amount and category. 
* Option 2(Menu Item): uses a `for` loop to show them. 
* Option 3(Menu Item): adds them up

````python
expenses = []
typed = ["1", "chai", "20", "food",
         "1", "auto-rickshaw", "85", "travel",
         "2", "3", "4"]

running = True
while running:
    print()
    print("1) Add expense   2) View expenses   3) Show total   4) Exit")
    choice = typed.pop(0)
    print("You chose:", choice)

    if choice == "1":
        name = typed.pop(0)                   # real program: input("Name: ")
        amount = int(typed.pop(0))            # real program: int(input("Amount: "))
        category = typed.pop(0)               # real program: input("Category: ")
        expenses.append({"name": name, "amount": amount, "category": category})
        print(f"Added: {name} - ₹{amount} ({category})")
    elif choice == "2":
        if len(expenses) == 0:
            print("No expenses yet.")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                number += 1
    elif choice == "3":
        total = 0
        for expense in expenses:
            total += expense["amount"]
        print(f"Total spent: ₹{total}")
    elif choice == "4":
        running = False
        print("Goodbye!")
    else:
        print("Invalid choice. Please type 1, 2, 3 or 4.")
````

### **Stage 3: Safe input and a limit warning (finished v0 menu)**


Two upgrades:
* A **retry loop** (Concept 7) inside option 1: keep asking for the amount until it is valid.
* A **limit warning** (Day 3): warn when an expense is above ₹500.

This is a **while loop inside a while loop**. The inner loop finishes completely every time option 1 runs.

````python
limit = 500
expenses = []
typed = ["1", "chai", "20", "food",
         "1", "groceries", "abc", "640", "food",       # 'abc' will be rejected
         "2", "3", "9", "4"]

print("Welcome to PaisaTracker v0")
running = True
while running:
    print()
    print("1) Add expense   2) View expenses   3) Show total   4) Exit")
    choice = typed.pop(0)
    print("You chose:", choice)

    if choice == "1":
        name = typed.pop(0)
        amount_text = typed.pop(0)
        while not (amount_text.isdigit() and int(amount_text) > 0):    # retry loop
            print(f"  '{amount_text}' is not a valid amount. Try again.")
            amount_text = typed.pop(0)
        amount = int(amount_text)
        category = typed.pop(0)
        expenses.append({"name": name, "amount": amount, "category": category})
        print(f"Added: {name} - ₹{amount} ({category})")
        if amount > limit:
            print(f"  Warning: ₹{amount} is above your limit of ₹{limit}")
    elif choice == "2":
        if len(expenses) == 0:
            print("No expenses yet.")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                number += 1
    elif choice == "3":
        total = 0
        for expense in expenses:
            total += expense["amount"]
        print(f"Total spent: ₹{total}")
    elif choice == "4":
        running = False
        print("Goodbye! (Your data is lost for now. On Day 8 we save it to a file.)")
    else:
        print("Invalid choice. Please type 1, 2, 3 or 4.")
````

### **Read it line by line**


| Part of the code | Concept used |
|------------------|--------------|
| `running = True` ... `while running:` | flag + menu loop (Concepts 6 and 12 ideas) |
| `typed.pop(0)` | simulated typing (Concept 5) |
| `while not (amount_text.isdigit() ...)` | retry until valid (Concept 7) |
| `elif choice == "4": running = False` | the way out of the loop |
| `if amount > limit:` | Day 3 control flow |
| `for expense in expenses:` | Day 5 for loop |
| `expenses.append({...})` | Day 4 list of dicts |
