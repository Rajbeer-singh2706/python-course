```python
## 1. Calculation functions : total, category totals, biggest expense
def calculate_total(expenses):
    total=0
    for expense in expenses:
        total += expense['amount']
    return total 

def category_totals(expenses):
    totals = {}
    for expense in expenses:
        category = expense['category']
        if category in totals:
            totals[category] += expense['amount']
        else:
            totals[category] = expense['amount']
    return totals

def biggest_expense(expenses):
    if len(expenses) == 0:
        return None
    biggest = expenses[0] #  # {"name": "auto-rickshaw", "amount": 85, "category": "travel"}
    for expense in expenses: 
        if expense['amount'] > biggest['amount']: 
            biggest = expense
    return biggest
```

```python
# 2. Action functions      : validate an amount, add an expense, show expenses
LIMIT = 500

def is_valid_amount(text):
    return text.isdigit() and int(text)> 0

def add_expense(expenses, name, amount, category):
    expense = {
        "name": name,
        "amount": amount, 
        "category": category
    }
    expenses.append(expense)
    print(f"Added: {name} - ₹{amount} ({category})")
    if amount > LIMIT:
        print(f"  Warning: ₹{amount} is above your limit of ₹{LIMIT}")

def show_expenses(expenses):
    if len(expenses) == 0:
        print("No Expenses yet.")
        return 
    number = 1 
    for expense in expenses:
        print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
        number +=1 
```

```python
# 3. Menu functions        : show the menu and run it.
```

```python
def ask(prompt):
    return input(prompt)

def show_menu():
    print()
    print("1.Add Expense 2. View expense 3. Show Total 4. Category Totals 5.exit")

def read_amount():
    amount = ask("Amount: ")

    ## RETRY LOOP
    while not is_valid_amount(amount):
        print("INvalid amount, please type a positive whoe numner ")
        amount = ask("Amount: ")

    # typecasting and returning   
    return int(amount)
```

```python
def run_menu(expenses):
    running = True
    while running:
        show_menu()
        choice = ask("enter the choice: ")
        print(f"You choose : {choice}")

        if choice == "1":
            print("############ Adding an Expense  #########") 
            name = ask("Name: ")
            amount = read_amount()
            category = ask("Category: ")
            add_expense(expenses, name, amount, category)

        elif choice == "2":
            print("########## Viewing Expenses #############") 
            show_expenses(expenses)
        elif choice == "3":
            print(f"Total Spent : ₹{calculate_total(expenses)}")
        elif choice == "4":
            for category , amount in category_totals(expenses).items():
                print(f"  {category}: ₹{amount}")
        elif choice == "5":
            running = False
            print("Goodbye! (Your data is lost for now. On Day 8 we save it to a file.)")
        else:
            print("Invalid Choice, please type 1 , 2 , 3 or 4.")
```

```python
print("===== PaisaTracker =====")
expenses = []
run_menu(expenses)
```
