# The PaisaTracker codebase



Now we help Priya. We build the menu in **three small stages**. Each stage adds one thing to the last.

| Stage | What it adds |
|-------|--------------|
| 1 | The menu loop (no data yet) |
| 2 | Real expenses: add, view, total |
| 3 | Safe input and a limit warning (the finished v0 menu) |

We use one list called `typed` to pretend the user is typing, in the same order as a real session.

### **Stage 1: The menu loop skeleton**


A `running` flag controls the loop. Choosing **4** sets `running = False`, and the loop ends.

```python
## Stage 1 : Menu Skeleton 
print("===== PaisaTracker =====")
running = True
while running:
    print("\n1. ADD Expense\n2. View expense\n3. Show Total \n4. Exit")
    choice = input("enter the choice: ")
    print(f"YOu choose : {choice}")

    if choice == "1":
        print("Adding an Expense - DEVELOPMENT IN STAGE 2 ") 
    elif choice == "2":
        print("Viewing Expenses - DEVELOPMENT IN STAGE 2 ")  
    elif choice == "3":
        print("TOTAL Expense - DEVELOPMENT IN STAGE 2 ")   
    elif choice == "4":
        running = False 
        print("GOOD BYE")
    else:
        print("Invalid Choice, please type 1 , 2 , 3 or 4.")
```

### **Stage 2: Real expenses**

We keep expenses in a **list of dicts** (Day 4). 
* Option 1(Menu Item): asks for name, amount and category. 
* Option 2(Menu Item): uses a `for` loop to show them. 
* Option 3(Menu Item): adds them up

```python
## Stage 1 : Menu Skeleton 
print("===== PaisaTracker =====")

expenses = []

running = True
while running:
    print("\n1.ADD Expense \n2. View expense \n3. Show Total \n4. Exit")
    choice = input("enter the choice: ")
    print(f"You choose : {choice}")

    if choice == "1":
        print("############ Adding an Expense  #########") 
        name = input("Enter Expense name: ")
        amount = int(input("Enter expense amount: "))
        category = input("Enter Category name: ")
        expenses.append({"name": name , "amount": amount, "category": category})
        print(f"Added: {name} - ₹{amount} ({category})")
    elif choice == "2":
        print("########## Viewing Expenses #############")  
        if len(expenses) == 0:
            print("no expenses yet")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                number+=1
    elif choice == "3":
        print("#####  TOTAL Expense #############") 
        total = 0 
        for expense in expenses:
            total+= expense['amount']  
        print(f"Total Spent : ₹{total}")
    elif choice == "4":
        running = False 
        print("GOOD BYE")
    else:
        print("Invalid Choice, please type 1 , 2 , 3 or 4.")
```

### **Stage 3: Safe input and a limit warning (finished v0 menu)**


Two upgrades:
* A **retry loop** (Concept 7) inside option 1: keep asking for the amount until it is valid.
* A **limit warning** (Day 3): warn when an expense is above ₹500.

This is a **while loop inside a while loop**. The inner loop finishes completely every time option 1 runs.

```python
#amount_text = "abc"
amount_text = "-50"

print(int(amount_text) > 0)
```

```python
## Stage 1 : Menu Skeleton 
print("===== PaisaTracker =====")
EXP_LIMIT = 500
expenses = []

running = True
while running:
    print("\n1.ADD Expense \n2. View expense \n3. Show Total \n4. Exit")
    choice = input("enter the choice: ")
    print(f"You choose : {choice}")

    if choice == "1":
        print("############ Adding an Expense  #########") 
        name = input("Enter Expense name: ")
        amount_text = input("Enter expense amount: ")

        ### RETRY LOOP
        while not (amount_text.isdigit() and int(amount_text) > 0):
            print(f"{amount_text} is not valid , please try again.")
            amount_text = input("Enter expense amount: ")

        amount = int(amount_text)
        category = input("Enter Category name: ")
        expenses.append({"name": name , "amount": amount, "category": category})
        print(f"Added: {name} - ₹{amount} ({category})")
        if amount > EXP_LIMIT:
            print(f"  Warning: ₹{amount} is above your limit of ₹{EXP_LIMIT}")
    elif choice == "2":
        print("########## Viewing Expenses #############")  
        if len(expenses) == 0:
            print("no expenses yet")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                number+=1
    elif choice == "3":
        print("#####  TOTAL Expense #############") 
        total = 0 
        for expense in expenses:
            total+= expense['amount']  
        print(f"Total Spent : ₹{total}")
    elif choice == "4":
        running = False 
        print("GOOD BYE")
    else:
        print("Invalid Choice, please type 1 , 2 , 3 or 4.")
```
