```python
# 3. `return` vs `print`
```

```python
def add_print(a, b):
    print(a + b )

def add_return(a, b):
    return a + b

x = add_print(2,3) 
y = add_return(2,3)

print(x)
print(y)  # 5
print("y doubled:", y * 2) # 10
print(x * 2)  
```

```python
# 4. `return` ends the function (early return and `None`)
```

```python
def biggest(amounts):
    if len(amounts) == 0:
        return None 
    
    largest = amounts[0]

    for amount in amounts:
        if amount > largest:
            largest = amount 

    return largest
```

```python
amounts = []
print(biggest(amounts))
```

```python
amounts = [120, 45, 300, 20, 50]

print(biggest(amounts))
```
