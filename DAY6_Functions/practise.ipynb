```python
LIMIT = 500
```

```python
def show_menu():
    print()
    print(" 1.ADD Expense 2. View expense 3. Show Total  4.Category Totals  5.Exit")

def ask(prompt):
    return input(prompt)
```

```python
def run_menu(expenses):
    running = True
    while running:
        show_menu()
        choice = ask("Enter the choice: ")
        print(f"You choose : {choice}")

        if choice == "1":
            print("############ Adding an Expense  #########") 
            
            name = ask("Enter Expense name: ")
            amount_text = ask("Enter expense amount: ")

            ### RETRY LOOP
            while not (amount_text.isdigit() and int(amount_text) > 0):
                print(f"{amount_text} is not valid , please try again.")
                amount_text = ask("Enter expense amount: ")

            amount = int(amount_text)
            category = ask("Enter Category name: ")
            expenses.append({"name": name , "amount": amount, "category": category})
            print(f"Added: {name} - ₹{amount} ({category})")
            if amount > LIMIT:
                print(f"  Warning: ₹{amount} is above your limit of ₹{LIMIT}")
        elif choice == "2":
            print("########## Viewing Expenses #############")  
            if len(expenses) == 0:
                print("no expenses yet")
            else:
                number = 1
                for expense in expenses:
                    print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                    number+=1
        elif choice == "3":
            print("#####  TOTAL Expense #############") 
            total = 0 
            for expense in expenses:
                total+= expense['amount']  
            print(f"Total Spent : ₹{total}")
        elif choice == "4":
            running = False 
            print("GOOD BYE")
        else:
            print("Invalid Choice, please type 1 , 2 , 3 or 4.")
```

```python
print("===== Welcome to PaisaTracker =====")
expenses = []
run_menu(expenses)
```
