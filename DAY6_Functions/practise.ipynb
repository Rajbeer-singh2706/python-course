```python
def calculate_total(expenses):
    total = 0
    for expense in expenses:
        total += expense["amount"]
    return total
```

```python
data1 = [ 
    {
        "amount": 10
    },
    {
        "amount": 20
    }
]

if calculate_total(data1)  == 30:
    print("Test case passed")
else:
    print("test case failed")
```
