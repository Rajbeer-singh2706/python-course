```python
expenses = [
    {"name": "chai","amount": 20 },
    {"name": "groceries", "amount": 640},
    {"name": "metro", "amount": 30},
]

# place 1: the report
total = 0
for expense in expenses:
    total += expense["amount"]
print("Report total:", total)
```

```python
def calculate_total(expenses):
    total = 0 
    for expense in expenses:
        total += expense['amount']
    return total


calculate_total(expenses)
```
