### 🟡 E1 — Compound Interest Calculator

Given: principal = £1000, annual rate = 5%, years = 10, compounded monthly.

Formula: `A = P × (1 + r/n)^(n×t)`  where `n = 12`

Print:
```
Principal  : £1,000.00
Rate       : 5.0% per year
Period     : 10 years (120 months)
Final value: £1,647.01

```


````python
# YOUR CODE HERE
principal = 1000.0
rate      = 0.05
years     = 10
n         = 12   # compounded monthly
````

````python
# ── SOLUTION E1 ──
principal = 1000.0
rate      = 0.05
years     = 10
n         = 12

amount   = principal * (1 + rate / n) ** (n * years)
interest = amount - principal

print(f"Principal  : £{principal:,.2f}")
print(f"Rate       : {rate:.1%} per year")
print(f"Period     : {years} years ({n * years} months)")
print(f"Final value: £{amount:,.2f}")
print(f"Interest   : £{interest:,.2f}")
````

### 🟡 E2 — Caesar Cipher (Single Character)

The Caesar cipher shifts each letter by a fixed number.
Using `ord()` and `chr()` (built-in functions that convert between characters and ASCII codes):

1. Take `char = 'H'` and `shift = 3`
2. Print the ASCII code of `char`
3. Compute the shifted character
4. Print: `H (72) → K (75)`

Bonus: make it wrap around (Z → C for shift 3).


````python
# YOUR CODE HERE
char  = 'H'
shift = 3
# Hint: ord('A') = 65, ord('Z') = 90
# Wrap formula: chr((ord(char) - 65 + shift) % 26 + 65)
````

````python
# ── SOLUTION E2 ──
char  = 'H'
shift = 3

original_code = ord(char)
shifted_char  = chr((original_code - 65 + shift) % 26 + 65)
shifted_code  = ord(shifted_char)

print(f"Original : {char} ({original_code})")
print(f"Shifted  : {shifted_char} ({shifted_code})")
print(f"{char} ({original_code}) → {shifted_char} ({shifted_code})")

# Bonus — wrap-around test
for c in ['X', 'Y', 'Z', 'A']:
    enc = chr((ord(c) - 65 + shift) % 26 + 65)
    print(f"{c} → {enc}")
````
