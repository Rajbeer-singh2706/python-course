## 🧠 Concepts

### 2.1 Variables

A **variable** is a name that refers to a value stored in memory. In Python you don't need to
declare a type — just assign a value and Python figures it out automatically.

**Naming rules:**
- Can contain letters, digits, and underscores `_`
- Cannot start with a digit
- Case-sensitive (`name` ≠ `Name`)
- Cannot be a reserved keyword (`if`, `for`, `class`, etc.)
- Convention: use `snake_case` (PEP 8 style guide)


````python
# ── Variable assignment ──
name       = "Alice"    # str
age        = 28         # int
height     = 1.68       # float
is_student = True       # bool

print(name, age, height, is_student)

# Reassignment — variables can change value (and even type)
age = 29
print("Updated age:", age)
````

### 2.2 The Four Primitive Data Types

| Type | Example Values | Use Case |
|------|---------------|----------|
| `int` | `0`, `42`, `-7`, `1_000_000` | Counting, indexing, whole-number math |
| `float` | `3.14`, `-0.5`, `2e8` | Measurements, ratios, currency |
| `str` | `'hello'`, `"world"` | Text, names, messages |
| `bool` | `True`, `False` | Flags, conditions, switches |

> 💡 `True` and `False` are also integers in Python: `True == 1` and `False == 0`.


````python
# ── The four primitive types ──
my_int   = 42
my_float = 3.14
my_str   = 'hello'
my_bool  = True

print("int:  ", my_int,   type(my_int))
print("float:", my_float, type(my_float))
print("str:  ", my_str,   type(my_str))
print("bool: ", my_bool,  type(my_bool))

# Booleans behave as integers
print("\nTrue + True =", True + True)   # 2
print("True * 10   =", True * 10)     # 10
````

### 2.3 `type()` — Inspecting Types at Runtime

`type(value)` returns the class of any object. This is your go-to debugging tool
when you're unsure what type a variable holds.


````python
# ── type() on various values ──
values = [42, 3.14, 'hello', True, None, [1, 2], {'a': 1}]

for v in values:
    print(f"{str(v):<12}  →  {type(v)}")
````

### 2.4 `print()` & String Formatting

Python offers three ways to embed values inside strings. **f-strings** are the modern standard:

```python
# f-string (preferred)
f"{expression}"   # evaluate expression and insert result
f"{value:.2f}"    # format float to 2 decimal places
f"{text:^20}"     # centre text in a 20-character field
```


````python
# ── String formatting methods ──
name = "Bob"
age  = 34
pi   = 3.14159

# Method 1 — concatenation (requires explicit str cast)
print("Name: " + name + ", Age: " + str(age))

# Method 2 — .format()
print("Name: {}, Age: {}".format(name, age))

# Method 3 — f-string (preferred)
print(f"Name: {name}, Age: {age}")
print(f"In 10 years: {age + 10}")         # expressions allowed
print(f"Pi to 2dp:   {pi:.2f}")
print(f"{'centred':^30}")                  # centred in 30 chars
print(f"{1_000_000:,}")                    # thousand separator
````

````python
# ── print() keyword arguments ──
print('a', 'b', 'c', sep=' | ')     # custom separator
print('Loading', end='...')          # no newline at end
print('Done')                        # continues on same line
````

### 2.6 Type Casting

| Function | Converts to | Raises on bad input |
|----------|------------|---------------------|
| `int(x)` | Integer | `ValueError` if non-numeric string |
| `float(x)` | Float | `ValueError` if non-numeric string |
| `str(x)` | String | Never |
| `bool(x)` | Boolean | Never (uses truthiness rules) |

**Falsy values:** `0`, `0.0`, `''`, `None`, `[]`, `{}`, `()` → `False`  
Everything else → `True`


````python
# ── Type casting examples ──

# str → number
print(int("42"))         # 42
print(float("3.14"))    # 3.14

# number → str
print(str(100))          # '100'
print(str(3.14))         # '3.14'

# int ↔ float (int truncates, does NOT round)
print(int(3.9))          # 3  ← not 4!
print(float(7))          # 7.0

# Truthiness
falsy = [0, 0.0, '', None, [], {}]
for v in falsy:
    print(f"bool({str(v):<6}) = {bool(v)}")
````

````python
# ── ValueError demo — what happens on bad cast ──
try:
    result = int("hello")
except ValueError as e:
    print(f"ValueError caught: {e}")

try:
    result = int("3.14")   # can't go str → int directly via a decimal string
except ValueError as e:
    print(f"ValueError caught: {e}")
    print("Fix: use int(float('3.14')) =", int(float("3.14")))
````

### Example 3 — Arithmetic with Mixed Types

Demonstrate that mixing int and float produces a float, and that dividing two ints
with `/` always produces a float.


````python
# ── Arithmetic and type promotion ──
a = 7       # int
b = 2       # int

print(f"{a} + {b}   = {a + b}   ({type(a + b).__name__})")   # int
print(f"{a} - {b}   = {a - b}   ({type(a - b).__name__})")   # int
print(f"{a} * {b}   = {a * b}   ({type(a * b).__name__})")   # int
print(f"{a} / {b}   = {a / b}   ({type(a / b).__name__})")   # float ALWAYS
print(f"{a} // {b}  = {a // b}  ({type(a // b).__name__})") # int (floor div)
print(f"{a} % {b}   = {a % b}   ({type(a % b).__name__})")   # int (remainder)
print(f"{a} ** {b}  = {a ** b}  ({type(a ** b).__name__})") # int (power)

# int + float → float
c = a + 0.5
print(f"\n{a} + 0.5 = {c}  ({type(c).__name__})")
````

### Example 4 — f-String Format Specifiers

Format floats, align text, and add thousand separators.


````python
# ── f-string format specifiers ──
pi      = 3.14159265
price   = 1234567.89
label   = "score"

print(f"{pi:.2f}")             # 3.14      — 2 decimal places
print(f"{pi:10.4f}")          #     3.1416 — width 10, 4 decimals
print(f"{price:,.2f}")        # 1,234,567.89 — thousand sep
print(f"{label:>15}")         #          score — right-align
print(f"{label:<15}|")        # score          | — left-align
print(f"{label:^15}|")        #      score     | — centre
print(f"{42:05d}")            # 00042 — zero-padded int
````

### Example 5 — Multiple Assignment & Swapping

Python supports tuple unpacking for multi-variable assignment and variable swapping.


````python
# ── Multiple assignment ──
x, y, z = 1, 2, 3
print(f"x={x}, y={y}, z={z}")

# Swap two variables without a temp variable
a, b = 10, 20
print(f"Before swap: a={a}, b={b}")
a, b = b, a
print(f"After swap:  a={a}, b={b}")

# Assign the same value to multiple variables
p = q = r = 0
print(f"p={p}, q={q}, r={r}")
````

## 📋 Summary & Next Steps


### What we covered today

| Concept | Key Takeaway |
|---------|-------------|
| Variables | Names bound to values. No type declaration needed. |
| `int` / `float` | Whole numbers vs decimal numbers. `/` always returns `float`. |
| `str` | Text in quotes. f-strings are the preferred formatting method. |
| `bool` | `True` / `False`. Falsy: `0`, `''`, `None`, `[]`, `{}`. |
| `type()` | Inspect any value's type at runtime. |
| `print()` | Display output. Use `sep=` and `end=` for control. |
| `input()` | Always returns `str`. Cast immediately. |
| Type casting | `int()`, `float()`, `str()`, `bool()` — `ValueError` on bad input. |

### 🔑 Key Takeaways

1. Python is **dynamically typed** — types are inferred, not declared.
2. `input()` **always returns a string** — cast before doing arithmetic.
3. `int(3.9)` → `3` — truncates, does **not** round.
4. f-strings accept **any expression** inside `{}`, not just variable names.
5. Falsy values: `0`, `0.0`, `''`, `None`, `[]`, `{}` — everything else is truthy.
