## **🔍 Step 0 — Before We Begin: Check What's Already Installed**

````python
#Let's see what's already on your system before installing anything new.
#Run the cell below — it's fine if some things are missing, that's what the rest of this notebook fixes.
# ── Step 0: Pre-flight check ─────────────────────────────
import subprocess, sys, os

def run(cmd):
    """Run a shell command and return (stdout, returncode)."""
    try:
        r = subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=15)
        return r.stdout.strip() or r.stderr.strip(), r.returncode
    except Exception as e:
        return str(e), 1

checks = [
    ('python3.12 --version', 'Python 3.12'),
    ('python3 --version',    'Python 3.x'),
    ('python --version',     'Python (any)'),
    ('uv --version',         'uv'),
    ('code --version',       'VS Code'),
    ('git --version',        'Git'),
]

print('Pre-flight system check')
print('=' * 45)
for cmd, label in checks:
    out, rc = run(cmd)
    icon = '✅' if rc == 0 else '❌'
    val  = out.split('\n')[0] if rc == 0 else 'not found'
    print(f'{icon}  {label:<20} {val}')
print('=' * 45)
print('Missing items will be installed in the steps below.')
````

````python
# ── Step 1 verification: Python 3.12 ────────────────────
import sys, subprocess

def check_python():
    major = sys.version_info.major
    minor = sys.version_info.minor
    patch = sys.version_info.micro

    print('Python Version Check')
    print('─' * 40)
    print(f'  Detected  : Python {major}.{minor}.{patch}')
    print(f'  Required  : Python 3.12+')
    print(f'  Executable: {sys.executable}')
    print(f'  Platform  : {sys.platform}')
    print('─' * 40)

    if major == 3 and minor >= 12:
        print('✅  Python 3.12+ confirmed — Step 1 complete!')
    elif major == 3 and minor >= 8:
        print(f'⚠️  Python {major}.{minor} found — works for now but please upgrade to 3.12')
        print('    Some course features (match statement, etc.) require 3.12+')
    else:
        print(f'❌  Python {major}.{minor} is too old — please install Python 3.12')
        print('    Follow the install instructions in the cell above.')

check_python()
````

## **📦 Step 2 — Install uv (Package Manager)**



### What is uv and why do we use it?

| Tool | Old way | With uv |
|------|---------|--------|
| Install packages | `pip install` | `uv add` |
| Create environments | `python -m venv .venv` | `uv venv` |
| Lock dependencies | `pip freeze > requirements.txt` | automatic `uv.lock` |
| Speed | baseline | **10–100× faster** |

uv is written in **Rust** and replaces pip, virtualenv, pip-tools, and pipx all at once.

### Install instructions

**macOS / Linux:**
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh

# Then restart terminal OR run:
source ~/.zshrc    # macOS
source ~/.bashrc   # Linux
```

**Windows (PowerShell as Administrator):**
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
# Close and reopen PowerShell after
```

> ⚠️ **After installing uv, restart your terminal completely before running the cell below.**


````python
# ── Step 2 verification: uv ──────────────────────────────
import subprocess

def check_uv():
    result = subprocess.run('uv --version', shell=True, capture_output=True, text=True)

    print('uv Package Manager Check')
    print('─' * 40)

    if result.returncode == 0:
        version = result.stdout.strip()
        print(f'  Detected : {version}')
        print('─' * 40)
        print('✅  uv found — Step 2 complete!')
    else:
        print('  Detected : not found')
        print('─' * 40)
        print('❌  uv not found')
        print('    Fix: install using the commands above, then restart terminal.')
        print('    If on Windows, make sure you ran PowerShell as Administrator.')

check_uv()
````

````python
# ── Step 2b: uv feature overview ─────────────────────────
import subprocess

# Show available uv commands (useful reference)
result = subprocess.run('uv --help', shell=True, capture_output=True, text=True)

if result.returncode == 0:
    # Print just the commands section
    lines = result.stdout.split('\n')
    print('Key uv commands available to you:')
    print('─' * 40)
    for line in lines:
        # Show command lines (indented, non-empty)
        stripped = line.strip()
        if stripped and not stripped.startswith('uv') and len(stripped) > 2:
            print(f'  {line.rstrip()}')
        if 'venv' in line or 'add' in line or 'init' in line or 'run' in line:
            pass  # already shown above
else:
    print('uv not installed yet — complete Step 2 install first.')
````

## **🔌 Step 4 — Install VS Code Extensions**



### The 4 essential extensions

| Extension | ID | What it does |
|-----------|----|--------------|
| **Python** | `ms-python.python` | Core support: debugger, test runner, IntelliSense |
| **Pylance** | `ms-python.vscode-pylance` | Fast type-aware autocomplete and error detection |
| **Ruff** | `charliermarsh.ruff` | Linter + formatter — formats on every save |
| **GitLens** | `eamodio.gitlens` | Git blame, history, branch management |

### How to install
**Option A — GUI (recommended):**
1. Press `Cmd+Shift+X` (Mac) or `Ctrl+Shift+X` (Windows/Linux)
2. Search each extension name and click **Install**

**Option B — Terminal (faster):**
```bash
code --install-extension ms-python.python
code --install-extension ms-python.vscode-pylance
code --install-extension charliermarsh.ruff
code --install-extension eamodio.gitlens
```


````python
# ── Step 4 verification: VS Code extensions ──────────────
import subprocess

REQUIRED = {
    'ms-python.python':            'Python',
    'ms-python.vscode-pylance':    'Pylance',
    'charliermarsh.ruff':          'Ruff',
    'eamodio.gitlens':             'GitLens',
}

def check_extensions():
    result = subprocess.run('code --list-extensions', shell=True, capture_output=True, text=True)

    print('VS Code Extensions Check')
    print('─' * 45)

    if result.returncode != 0:
        print('❌  Cannot check extensions — VS Code not found in PATH')
        print('    Install VS Code first (Step 3), then re-run this cell.')
        return

    installed = [e.lower() for e in result.stdout.strip().split('\n')]
    all_ok = True

    for ext_id, name in REQUIRED.items():
        found = ext_id.lower() in installed
        icon  = '✅' if found else '❌'
        status = 'installed' if found else 'MISSING'
        print(f'  {icon}  {name:<12}  {ext_id}  →  {status}')
        if not found:
            all_ok = False

    print('─' * 45)
    if all_ok:
        print('✅  All 4 extensions installed — Step 4 complete!')
    else:
        print('❌  Some extensions missing. Install them and re-run this cell.')
        print('    Quick install command:')
        for ext_id, name in REQUIRED.items():
            if ext_id.lower() not in installed:
                print(f'    code --install-extension {ext_id}')

check_extensions()
````

## **🧪 Step 5 — Create Your Virtual Environment**



### What is a virtual environment?

Think of it as a **clean room** just for your project:

```
Without venv:                  With venv:
─────────────────────          ─────────────────────────────────
System Python                  System Python
  └── requests 2.28            ├── python-course/
  └── numpy 1.23                   └── .venv/
  └── flask 2.0                        └── requests 2.31  ← isolated!
                                         └── numpy 1.26
```

**One system, many projects, zero conflicts.**

### Setup commands (run these in your terminal, NOT here)

```bash
# 1. Create your course folder
mkdir python-course
cd python-course

# 2. Initialise uv project
uv init

# 3. Create the virtual environment
uv venv

# 4. Activate it
source .venv/bin/activate    # macOS / Linux
.venv\Scripts\Activate.ps1   # Windows PowerShell

# 5. Open VS Code in the folder
code .
```

> ✅ You'll know it worked when you see `(.venv)` at the start of your terminal prompt
> and VS Code's status bar shows **Python 3.12.x (.venv)**


````python
# ── Step 5 verification: Virtual environment ─────────────
import sys, os

def check_venv():
    in_venv   = sys.prefix != sys.base_prefix
    venv_path = sys.prefix if in_venv else None
    venv_name = os.path.basename(venv_path) if venv_path else None

    print('Virtual Environment Check')
    print('─' * 45)
    print(f'  Active    : {in_venv}')
    print(f'  Name      : {venv_name or "N/A"}')
    print(f'  Path      : {venv_path or "N/A"}')
    print(f'  Executable: {sys.executable}')
    print('─' * 45)

    if in_venv and venv_name == '.venv':
        print('✅  Inside .venv — Step 5 complete!')
    elif in_venv:
        print(f'✅  Inside a virtual environment ({venv_name})')
        print('    (Expected .venv — but any venv is fine for now)')
    else:
        print('❌  No virtual environment active')
        print('    Fix: run the commands above in your terminal,')
        print('         then restart Jupyter from inside the activated venv.')

check_venv()
````

## **🚀 Step 6 — Hello World: Your First Python Script**


Now we verify the **entire stack** works end-to-end by writing and running real Python code.

### What we'll do
1. Write the Hello World script **right here in the notebook**
2. Understand every single line of it
3. Write it to a `.py` file and run it from the terminal
4. Confirm the output matches

> This is the exact script from the video session.


````python
# ── Step 6a: Hello World — run it here first ─────────────

def greet(name: str) -> str:
    """
    Return a personalised greeting string.

    Args:
        name (str): The name to greet.

    Returns:
        str: A greeting message.
    """
    return f'Hello, {name}! Welcome to Python 3.12.'


# __name__ guard — only runs when executed as a script, not when imported
if __name__ == '__main__':
    message = greet('World')
    print(message)

# In a notebook we call functions directly:
print(greet('World'))
print(greet('Python Learner'))
````

````python
# ── Step 6b: Line-by-line breakdown ──────────────────────

# Line 1: def — define a function named 'greet'
# name: str  — type hint: the parameter 'name' should be a string
# -> str     — type hint: this function returns a string
# These hints are optional but good practice (used everywhere from Day 18 on)

def greet_explained(name: str) -> str:
    # f-string: the f prefix lets you embed {variables} in a string
    return f'Hello, {name}!'

# Calling the function:
result = greet_explained('Alice')  # 'Alice' is the ARGUMENT
print(result)                       # prints: Hello, Alice!

# What 'return' does:
# Without return:                  With return:
# greet_explained('Alice') → None  greet_explained('Alice') → 'Hello, Alice!'

def no_return(name):
    message = f'Hello, {name}!'    # message is created but never sent back
    # (no return statement)

print(greet_explained('Bob'))    # → Hello, Bob!
print(no_return('Bob'))          # → None   (nothing returned!)
````

````python
# ── Step 6c: Write hello.py to disk and run it ───────────
import subprocess, os

# Write the script to a file
script = '''# hello.py — your first Python script

def greet(name: str) -> str:
    return f"Hello, {name}! Welcome to Python 3.12."

if __name__ == "__main__":
    message = greet("World")
    print(message)
'''

filepath = 'hello.py'
with open(filepath, 'w') as f:
    f.write(script)

print(f'✅  Written to: {os.path.abspath(filepath)}')
print()

# Run it using the current Python interpreter
result = subprocess.run(
    [sys.executable, filepath],
    capture_output=True, text=True
)

print('Running hello.py...')
print('─' * 40)
print(result.stdout.strip())
print('─' * 40)

expected = 'Hello, World! Welcome to Python 3.12.'
if result.stdout.strip() == expected:
    print('✅  Output matches — Step 6 complete!')
else:
    print('❌  Output mismatch')
    print(f'   Expected : {expected}')
    print(f'   Got      : {result.stdout.strip()}')
````

````python
# ── Step 6d: Inspect hello.py we just created ────────────

print('Contents of hello.py:')
print('─' * 40)
with open('hello.py') as f:
    for i, line in enumerate(f, 1):
        print(f'  {i:2}  {line}', end='')
print('\n' + '─' * 40)
print('You can open this file in VS Code with: code hello.py')
````

## **🏁 Final Check — Complete Environment Report**


Run this cell last. It summarises the entire setup in one place.
**All items should show ✅ before you move to Day 2.**


````python
# ── FINAL: Complete environment health report ────────────
import sys, os, subprocess

def run_check(cmd):
    try:
        r = subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=10)
        return r.stdout.strip().split('\n')[0], r.returncode == 0
    except:
        return 'error', False

print('╔══════════════════════════════════════════════════╗')
print('║     Day 1 — Complete Environment Report         ║')
print('╠══════════════════════════════════════════════════╣')

results = []

# 1. Python version
v = sys.version_info
ok = v.major == 3 and v.minor >= 12
val = f'Python {v.major}.{v.minor}.{v.micro}'
results.append(('Python 3.12+', ok, val))

# 2. uv
uv_ver, uv_ok = run_check('uv --version')
results.append(('uv installed', uv_ok, uv_ver if uv_ok else 'not found'))

# 3. VS Code
code_ver, code_ok = run_check('code --version')
results.append(('VS Code installed', code_ok, f'v{code_ver}' if code_ok else 'not found'))

# 4. VS Code extensions
ext_out, ext_ok = run_check('code --list-extensions')
required_exts = ['ms-python.python', 'ms-python.vscode-pylance', 'charliermarsh.ruff', 'eamodio.gitlens']
if ext_ok:
    ext_list = ext_out.lower()
    all_exts = all(e.lower() in ext_list for e in required_exts)
    results.append(('VS Code extensions (4)', all_exts, 'all installed' if all_exts else 'some missing'))
else:
    results.append(('VS Code extensions (4)', False, 'cannot check'))

# 5. Virtual environment
in_venv = sys.prefix != sys.base_prefix
venv_name = os.path.basename(sys.prefix) if in_venv else 'none'
results.append(('Virtual env active', in_venv, venv_name if in_venv else 'not activated'))

# 6. hello.py exists and runs
script_exists = os.path.exists('hello.py')
if script_exists:
    run_out, run_ok = run_check(f'{sys.executable} hello.py')
    hello_ok = run_ok and 'Hello, World' in run_out
    results.append(('hello.py runs correctly', hello_ok, run_out if hello_ok else 'error running script'))
else:
    results.append(('hello.py runs correctly', False, 'file not found — run Step 6c first'))

# Print results
for label, ok, detail in results:
    icon = '✅' if ok else '❌'
    print(f'║  {icon}  {label:<28} {detail[:16]:<16}  ║')

print('╠══════════════════════════════════════════════════╣')
all_ok = all(ok for _, ok, _ in results)
if all_ok:
    print('║  🎉  All checks passed!  Ready for Day 2.        ║')
else:
    failed = sum(1 for _, ok, _ in results if not ok)
    print(f'║  ⚠️   {failed} check(s) failed — review steps above.     ║')
print('╚══════════════════════════════════════════════════╝')
````
