````python
# SOLUTION
prices = {
    'milk': 25.0, 'bread': 35.0, 'butter': 110.0,
    'eggs': 80.0, 'sugar': 45.0,
}

def checkout(cart, prices):
    total = 0
    for item in cart:
        if item in prices:
            total += prices[item]
        else:
            print(f"Warning: '{item}' not found in price list")
    return total

total = checkout(['milk', 'bread', 'butter', 'jam'], prices)
print(f'Total: ₹{total}')   # ₹170.0 (jam not found)
````

````python
# SOLUTION
def caesar_encrypt(text, shift):
    def shift_char(ch):
        if ch.isupper():
            return chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
        elif ch.islower():
            return chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
        return ch
    return ''.join(shift_char(ch) for ch in text)

def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)

msg = 'Hello, World! 123'
encrypted = caesar_encrypt(msg, 3)
print('Encrypted:', encrypted)
print('Decrypted:', caesar_decrypt(encrypted, 3))
````

````python
# SOLUTION
old_stock = {'apples': 50, 'bananas': 30, 'carrots': 20, 'dates': 15}
new_stock = {'apples': 45, 'bananas': 30, 'elderberry': 10, 'figs': 8}

def inventory_diff(old, new):
    old_keys = set(old)
    new_keys = set(new)
    added   = {k: new[k] for k in new_keys - old_keys}
    removed = {k: old[k] for k in old_keys - new_keys}
    changed = {k: (old[k], new[k])
               for k in old_keys & new_keys
               if old[k] != new[k]}
    return {'added': added, 'removed': removed, 'changed': changed}

diff = inventory_diff(old_stock, new_stock)
print('Added:  ', diff['added'])
print('Removed:', diff['removed'])
print('Changed:', diff['changed'])
````

````python
# SOLUTION
employees = [
    {'name': 'Priya',   'dept': 'Engineering', 'salary': 85000, 'yrs': 3},
    {'name': 'Arjun',   'dept': 'Marketing',   'salary': 62000, 'yrs': 5},
    {'name': 'Lakshmi', 'dept': 'Engineering', 'salary': 95000, 'yrs': 7},
    {'name': 'Rahul',   'dept': 'HR',           'salary': 55000, 'yrs': 2},
    {'name': 'Sunita',  'dept': 'Engineering', 'salary': 78000, 'yrs': 4},
    {'name': 'Vikram',  'dept': 'Marketing',   'salary': 70000, 'yrs': 6},
]

def filter_rows(sheet, column, value):
    return [row for row in sheet if row[column] == value]

def col_values(sheet, column):
    return [row[column] for row in sheet]

def col_stats(sheet, column):
    vals = col_values(sheet, column)
    return {'min': min(vals), 'max': max(vals), 'mean': sum(vals)/len(vals)}

def sort_by(sheet, column, reverse=False):
    return sorted(sheet, key=lambda row: row[column], reverse=reverse)

def add_column(sheet, col_name, func):
    for row in sheet:
        row[col_name] = func(row)

eng = filter_rows(employees, 'dept', 'Engineering')
print('Engineering:', [e['name'] for e in eng])
print('Salary stats:', col_stats(employees, 'salary'))
top_earners = sort_by(employees, 'salary', reverse=True)
print('Top earner:', top_earners[0]['name'])
add_column(employees, 'bonus', lambda row: int(row['salary'] * 0.10))
print('Priya bonus:', employees[0]['bonus'])
````
