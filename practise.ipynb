## Control Flow 
 * If 
 * If / Else 
 * If / elif
## Operators 
 * Comparision Operators :  == != > < >= <
 * Logical Operators : and / or / not
 * Nested Conditions

```python
month = "July"

match month:
    case "December" | "January" | "February":
        season = "Winter"
    case "March" | "April" | "May":
        season = "Summer"
    case "June" | "July" | "August" | "September":
        season = "Monsoon"
    case _:
        season = "Post-Monsoon"

print(f"{month} falls in the {season} season in India.")
```
