## 🧠 Concepts

### 2.1 Variables

A **variable** is a name that refers to a value stored in memory. In Python you don't need to
declare a type — just assign a value and Python figures it out automatically.

**Naming rules:**
- Can contain letters, digits, and underscores `_`
- Cannot start with a digit
- Case-sensitive (`name` ≠ `Name`)
- Cannot be a reserved keyword (`if`, `for`, `class`, etc.)
- Convention: use `snake_case` (PEP 8 style guide)


```python
name = 'rajbeer'
age = 30
height = 1.68
is_student = True
```

```python
print(age, name, height, is_student)
```

```python
age = 29 
print(age)
```

```python
type(is_student)
```

### 2.3 `type()` — Inspecting Types at Runtime

`type(value)` returns the class of any object. This is your go-to debugging tool
when you're unsure what type a variable holds.

```python
# ════════════════════════════════════════════════════════════════
# MINI-CHALLENGE
# ════════════════════════════════════════════════════════════════
# Create four variables describing a chai stall in Delhi:
#   stall_name (str), 
#   cups_sold_today (int),
#   price_per_cup (float), 
# is_open_now (bool)

# Then print each variable's type using type().
```

```python
age = 10 
print(age)
print(type(age))
```

```python
age = "20"
print(age)        # 20 
print(type(age))  # str 
```
