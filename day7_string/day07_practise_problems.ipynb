````python
# Solution
import re

def encode(s):
    if not s:
        return ''
    result = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result.append(s[i-1] + str(count))
            count = 1
    result.append(s[-1] + str(count))
    return ''.join(result)

def decode(s):
    pairs = re.findall(r'([A-Za-z])(\d+)', s)
    return ''.join(char * int(count) for char, count in pairs)

original = 'AAABBBCCDDDDEA'
encoded = encode(original)
decoded = decode(encoded)
print(f'Original : {original}')
print(f'Encoded  : {encoded}')
print(f'Decoded  : {decoded}')
print(f'Match    : {original == decoded}')
````

````python
# Solution
def caesar(text, shift):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            shifted = (ord(ch) - base + shift) % 26 + base
            result.append(chr(shifted))
        else:
            result.append(ch)
    return ''.join(result)

msg = 'Python is great!'
encoded = caesar(msg, 13)
decoded = caesar(encoded, -13)
print(f'Original : {msg}')
print(f'Encoded  : {encoded}')
print(f'Decoded  : {decoded}')
````

````python
# Solution
import re

def parse_config(config_text):
    result = {}
    current_section = None
    for line in config_text.splitlines():
        line = line.strip()
        if not line or line.startswith('#'):
            continue
        section_match = re.match(r'^\[(.+)\]$', line)
        if section_match:
            current_section = section_match.group(1)
            result[current_section] = {}
            continue
        kv_match = re.match(r'^([^=]+)=(.+)$', line)
        if kv_match and current_section:
            key = kv_match.group(1).strip()
            val = kv_match.group(2).strip()
            result[current_section][key] = val
    return result

config_text = '''
# App configuration
[database]
host = localhost
port = 5432

[server]
host = 0.0.0.0
port = 8000
debug = true
'''
config = parse_config(config_text)
for section, values in config.items():
    print(f'[{section}]')
    for k, v in values.items():
        print(f'  {k} = {v}')
````
