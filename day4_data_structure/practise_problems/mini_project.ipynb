```python
# SOLUTION
contacts = {}

def add_contact(contacts, name, phone, email, tags):
    contacts[name] = {
        'phone': phone,
        'email': email,
        'tags':  set(tags)
    }

def remove_contact(contacts, name):
    if name in contacts:
        del contacts[name]
        print(f'{name} removed.')
    else:
        print(f"Contact '{name}' not found.")

def search_by_tag(contacts, tag):
    return [name for name, info in contacts.items() if tag in info['tags']]

def display_all(contacts):
    print(f'\n{'='*45}')
    print(f"{'CONTACT BOOK':^45}")
    print(f'{'='*45}')
    for name, info in contacts.items():
        print(f"  {name}")
        print(f"    Phone: {info['phone']}")
        print(f"    Email: {info['email']}")
        print(f"    Tags:  {', '.join(sorted(info['tags']))}")
    print(f'{'='*45}\n')

add_contact(contacts, 'Priya',   '+91-9000000001', 'priya@example.com',   ['friend','work'])
add_contact(contacts, 'Arjun',   '+91-9000000002', 'arjun@example.com',   ['family'])
add_contact(contacts, 'Lakshmi', '+91-9000000003', 'lakshmi@example.com', ['friend'])
add_contact(contacts, 'Rahul',   '+91-9000000004', 'rahul@example.com',   ['work','college'])

display_all(contacts)
print('Friends:', search_by_tag(contacts, 'friend'))
print('Work contacts:', search_by_tag(contacts, 'work'))
remove_contact(contacts, 'Arjun')
remove_contact(contacts, 'Unknown')
print(f'Contacts after removal: {list(contacts.keys())}')
```
