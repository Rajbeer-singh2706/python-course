````python
## slicing 
````

Slicing extracts a **sub-sequence** from any ordered type (list, tuple, string):

```python
sequence[start:stop:step]
```

- `start` — index to begin (inclusive, default 0)
- `stop`  — index to end (exclusive, default end)
- `step`  — stride (default 1; use -1 to reverse)

All three are optional. Negative indices count from the end.


````python
nums
````

````python
print(nums[0])
print(nums[1])
print(nums[2])
````
