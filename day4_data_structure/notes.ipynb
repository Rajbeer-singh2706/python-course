```python
students = {
    'S001': {'name': 'Arjun',   'grade': 'A', 'scores': [90, 85, 92]},
    'S002': {'name': 'Lakshmi', 'grade': 'B', 'scores': [75, 80, 78]},
    'S003': {'name': 'Rahul',   'grade': 'A', 'scores': [88, 91, 95]},
}
students
```

```python
print(students['S001']['scores'])
```

```python
for student_id, info in students.items():
    avg = sum(info['scores']) / len(info['scores'])
    print(student_id , "-", info['name'], info['grade'], info['scores'], f"Avg: {avg:.1f}" )
```
