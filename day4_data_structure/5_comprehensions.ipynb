## 1. What is a comprehension?

A **comprehension** builds a new collection from an existing iterable in **one readable line**. It replaces the common "create an empty collection, loop, add to it" pattern.

| Type | Brackets | Pattern |
|------|----------|---------|
| List | `[ ]` | `[expression for item in iterable]` |
| Set | `{ }` | `{expression for item in iterable}` |
| Dict | `{ : }` | `{key: value for item in iterable}` |

Optional parts: a filter at the end (`if condition`), or a conditional expression at the front (`a if condition else b`).

````python
# The long way: loop + append
squares_loop = []
for n in range(1, 6):
    squares_loop.append(n ** 2)

# The comprehension way: same result, one line
squares = [n ** 2 for n in range(1, 6)]

print(squares_loop)
print(squares)
print("Same result:", squares_loop == squares)
````

### How to read one

```
[ n ** 2   for n in range(1, 6)   if n % 2 == 1 ]
  ^^^^^^     ^^^^^^^^^^^^^^^^^^^    ^^^^^^^^^^^^
  what goes   where items come      keep only items
  into the    from (the loop)       passing this test
  new list    (optional)
```

Read it as: *"give me `n ** 2` for every `n` in `range(1, 6)` where `n` is odd."*

## 2. List comprehensions

### 2.1 Transforming every item

````python
names = ["priya sharma", "arjun verma", "riya deshmukh"]

print([name.title() for name in names])        # change each item
print([len(name) for name in names])           # compute something from each item
print([name.split()[0] for name in names])     # first names
````

````python
# Prices in rupees, adding 18% GST
prices_rs = [120, 560, 45, 899]
with_gst = [round(price * 1.18, 2) for price in prices_rs]
print(with_gst)

# Celsius to Fahrenheit
celsius = [25, 30, 22, 35]
print([(c * 9 / 5) + 32 for c in celsius])

# Any iterable works: strings, ranges, tuples, dict keys...
print([ch.upper() for ch in "mumbai"])
print([int(digit) for digit in "2026"])
````

### 2.2 Filtering with `if`

Put the condition at the **end**. Items that fail the test are left out.

````python
marks = [45, 78, 92, 34, 88, 61, 39]

passed = [m for m in marks if m >= 40]
print("Passed:", passed)

evens = [n for n in range(1, 21) if n % 2 == 0]
print("Evens:", evens)

cities = ["Mumbai", "Delhi", "Ahmedabad", "Pune", "Agra", "Chennai"]
a_cities = [c for c in cities if c.startswith("A")]
print("Cities starting with A:", a_cities)
````

````python
# Filter AND transform together
expensive_in_dollars = [round(p / 83, 2) for p in [120, 560, 45, 899, 3000] if p > 500]
print(expensive_in_dollars)

# Several conditions: use `and` / `or` (clearer than stacking ifs)
fizzbuzz_numbers = [n for n in range(1, 51) if n % 3 == 0 and n % 5 == 0]
print("Divisible by 3 and 5:", fizzbuzz_numbers)

weekend_or_holiday = [d for d in ["Mon", "Sat", "Sun", "Wed"] if d in ("Sat", "Sun")]
print(weekend_or_holiday)
````

### 2.3 Choosing between two values: `if / else` at the front

When you want to **keep every item but change it depending on a condition**, the `if/else` goes at the **front**, before `for`.

| Goal | Shape | Items kept |
|------|-------|-----------|
| Leave some items out | `[x for x in items if condition]` | Only those passing |
| Pick one of two results | `[a if condition else b for x in items]` | All of them |

````python
exam_scores = [55, 32, 78, 40, 25]

# Every score gets a label (note: if/else comes BEFORE the for)
results = ["Pass" if score >= 40 else "Fail" for score in exam_scores]
print(results)

# Replace invalid values (negative) with 0
readings = [12, -5, 30, -1, 8]
print([r if r >= 0 else 0 for r in readings])

# A chain of conditions is possible, but keep it short
temps = [41, 33, 27, 19]
labels = ["Very hot" if t >= 40 else "Warm" if t >= 25 else "Cool" for t in temps]
print(labels)
````

### 2.4 Working with two sequences: `zip` and `enumerate`

````python
students = ["Priya", "Arjun", "Riya"]
scores = [88, 74, 91]

# zip pairs up items from two lists
print([f"{name}: {score}" for name, score in zip(students, scores)])

# enumerate gives a counter (start=1 for human-friendly numbering)
menu = ["masala dosa", "idli", "vada"]
print([f"{i}. {dish}" for i, dish in enumerate(menu, start=1)])

# Only the toppers, using zip with a filter
print([name for name, score in zip(students, scores) if score >= 80])
````

### 2.5 Two loops in one comprehension

Write the `for` clauses in the **same order** you would write nested loops.

````python
colours = ["Red", "Blue", "Black"]
sizes = ["S", "M"]

# Equivalent nested loops:
#   for colour in colours:
#       for size in sizes:
#           ...
combos = [(colour, size) for colour in colours for size in sizes]
print(combos)
print("Total variants:", len(combos))

# With a condition: pairs where the first number is smaller
pairs = [(i, j) for i in range(1, 4) for j in range(1, 4) if i < j]
print(pairs)
````

### 2.6 Lists of lists (nested data)

**Flattening:** loop over the outer list first, then the inner one.

**Building a grid:** put a comprehension *inside* a comprehension.

````python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

# Flatten: outer loop first, inner loop second (reads like nested for loops)
flat = [value for row in matrix for value in row]
print("Flattened:", flat)

# Flatten with a filter
print("Only even:", [value for row in matrix for value in row if value % 2 == 0])

# Build a multiplication table (a list of lists)
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]
for row in table:
    print(row)
````

````python
# Transpose: rows become columns
transposed = [[row[col] for row in matrix] for col in range(len(matrix[0]))]
print(transposed)

# Row-wise calculation: average marks per student
marks_matrix = [[88, 92, 79], [45, 38, 52], [91, 85, 94]]
print([round(sum(row) / len(row), 1) for row in marks_matrix])

# Diagonal
print([matrix[i][i] for i in range(len(matrix))])
````

### 2.7 Calling functions inside a comprehension

The expression can be any Python expression, including your own function.

````python
def gst_amount(price):
    return round(price * 0.18, 2)

bill_items = [250, 99, 1200]
print([gst_amount(p) for p in bill_items])

words = "mumbai is a busy city and mumbai never sleeps".split()
print([len(w) for w in words])
print([w.capitalize() for w in words if len(w) > 3])
````

## 3. Set comprehensions

Same idea with curly braces `{ }`. The result is a **set**, so duplicates disappear and order is not guaranteed. Use one when you care about the **unique** results.

````python
# List vs set: same loop, different brackets
numbers = [1, 2, 2, 3, 3, 3, 4]
print("List:", [n * 10 for n in numbers])
print("Set: ", sorted({n * 10 for n in numbers}))
````

````python
# Unique first letters
cities = ["Mumbai", "Delhi", "Manali", "Dehradun", "Pune", "Mysuru"]
print(sorted({city[0] for city in cities}))

# Unique word lengths
sentence = "to be or not to be that is the question"
print(sorted({len(word) for word in sentence.split()}))

# Squares: negative and positive numbers give the same square, so duplicates collapse
print(sorted({n ** 2 for n in range(-3, 4)}))
````

````python
# Unique email domains
emails = ["priya@gmail.com", "arjun@yahoo.com", "riya@gmail.com",
          "sanjay@outlook.com", "meera@yahoo.com"]
domains = {email.split("@")[1] for email in emails}
print(sorted(domains))

# Case-insensitive uniqueness
visitors = ["Priya", "priya", "ARJUN", "Arjun", "Riya"]
print(sorted({name.lower() for name in visitors}))

# Filter inside a set comprehension: vowels used in a sentence
text = "Mumbai never sleeps"
print(sorted({ch for ch in text.lower() if ch in "aeiou"}))
````

````python
# A set comprehension with a membership test acts like an intersection
python_batch = {"Priya", "Arjun", "Riya", "Sanjay"}
sql_batch = {"Riya", "Sanjay", "Kavita"}

both = {name for name in python_batch if name in sql_batch}
print(sorted(both))
print(both == python_batch & sql_batch)   # same as the & operator
````

## 4. Dictionary comprehensions

Write `key: value` before the `for`. Each loop pass creates one key-value pair.

### 4.1 Building a dictionary from a sequence

````python
# Number -> square
print({n: n ** 2 for n in range(1, 6)})

# Name -> length of name
names = ["Priya", "Arjun", "Riya Deshmukh"]
print({name: len(name) for name in names})

# Start every counter at zero
subjects = ["Maths", "Science", "English"]
print({subject: 0 for subject in subjects})

# Number the items from 1
print({i: dish for i, dish in enumerate(["idli", "dosa", "vada"], start=1)})
````

### 4.2 Combining two lists with `zip`

````python
cities = ["Mumbai", "Delhi", "Chennai", "Bengaluru"]
airport_codes = ["BOM", "DEL", "MAA", "BLR"]

print({city: code for city, code in zip(cities, airport_codes)})

# dict(zip(...)) does the same job when no change is needed
print(dict(zip(cities, airport_codes)))
````

### 4.3 Reshaping an existing dictionary with `.items()`

Loop over `.items()` to get each key and value, then build a new dictionary from them.

````python
tea_stall = {"tea": 15, "coffee": 25, "samosa": 20, "vada pav": 18}

# Change values
print("Price hike: ", {item: price + 5 for item, price in tea_stall.items()})
print("10% off:    ", {item: round(price * 0.9, 1) for item, price in tea_stall.items()})

# Change keys
print("Upper keys: ", {item.upper(): price for item, price in tea_stall.items()})

# Filter entries
print("Rs.20 and up:", {item: price for item, price in tea_stall.items() if price >= 20})

# Change values depending on a condition (if/else goes in the VALUE part)
print({item: ("premium" if price >= 25 else "regular") for item, price in tea_stall.items()})
````

### 4.4 Swapping keys and values

````python
city_code = {"Mumbai": "BOM", "Delhi": "DEL", "Chennai": "MAA"}
code_city = {code: city for city, code in city_code.items()}
print(code_city)

# Warning: values must be unique, or later keys overwrite earlier ones
grades = {"Priya": "A", "Arjun": "B", "Riya": "A"}
print({grade: name for name, grade in grades.items()})   # Priya is lost!
````

### 4.5 Dictionaries whose values are lists or dictionaries

````python
marks = {"Priya": [88, 92, 79], "Arjun": [45, 38, 52], "Riya": [91, 85, 94]}

averages = {name: round(sum(m) / len(m), 1) for name, m in marks.items()}
print("Averages:", averages)

print("Highest scores:", {name: max(m) for name, m in marks.items()})

# Keep only students averaging 80+
print("Toppers:", {name: avg for name, avg in averages.items() if avg >= 80})
````

````python
# A dictionary comprehension inside another one: a blank mark sheet
students = ["Priya", "Arjun"]
subjects = ["Maths", "Science"]
mark_sheet = {student: {subject: None for subject in subjects} for student in students}
print(mark_sheet)

# Counting with count() (fine for small data)
words = "to be or not to be".split()
word_counts = {word: words.count(word) for word in set(words)}
print(dict(sorted(word_counts.items())))
````

## 5. One dataset, three comprehensions

The same data, built into a list, a set and a dictionary.

````python
exam = {"Priya": 91, "Sanjay": 74, "Riya": 88, "Arjun": 67, "Meera": 74}

def grade(score):
    return "A" if score >= 85 else "B" if score >= 70 else "C"

as_list = [score for score in exam.values() if score >= 70]
as_set = {score for score in exam.values()}
as_dict = {name: grade(score) for name, score in exam.items()}

print("List (scores 70+, duplicates kept):", as_list)
print("Set (distinct scores):             ", sorted(as_set))
print("Dict (name -> grade):              ", as_dict)
````

| You want | Use | Result keeps duplicates? | Has order? |
|----------|-----|--------------------------|-----------|
| A sequence of results | List `[ ]` | Yes | Yes |
| Only the distinct results | Set `{ }` | No | No |
| A lookup from a name to a value | Dict `{k: v}` | Keys are unique | Yes (insertion order) |

## 6. A quick look at generator expressions

Use **round brackets** and you get a **generator**: it produces items one at a time instead of building a whole list. It is ideal when you only need to feed the values into `sum()`, `max()`, `any()` or `all()`.

There is **no "tuple comprehension"**: `(x for x in ...)` is a generator. Use `tuple(...)` around it if you need a tuple.

````python
print(sum(n ** 2 for n in range(1, 6)))              # no brackets needed inside sum()
print(max(len(name) for name in ["Priya", "Arjun", "Riya Deshmukh"]))

marks = [45, 78, 92, 34]
print("Anyone failed?", any(m < 40 for m in marks))
print("Everyone passed?", all(m >= 40 for m in marks))

gen = (n * 2 for n in range(3))
print(type(gen))
print(list(gen))
print(tuple(n * 2 for n in range(3)))
````

## 7. When NOT to use a comprehension

Use a comprehension to **build a collection**. Switch to a normal loop when:

- you need **side effects** (printing, writing files, updating other variables)
- the logic needs several steps or several `if/elif` branches
- it no longer fits on one or two readable lines

Rule of thumb: if you have to stop and decode it, write the loop.

````python
scores = [92, 45, 78, 33, 61]

# Hard to read: nested conditions squeezed into one line
labels_hard = ["Distinction" if s >= 85 else "First" if s >= 60 else "Pass" if s >= 40 else "Fail" for s in scores]

# Easier to read: a small function, then a clean comprehension
def label(score):
    if score >= 85:
        return "Distinction"
    elif score >= 60:
        return "First"
    elif score >= 40:
        return "Pass"
    return "Fail"

labels_clear = [label(s) for s in scores]
print(labels_hard == labels_clear)
print(labels_clear)
````

````python
# Not a good use: a comprehension just for printing (it builds a useless list of None)
# [print(s) for s in scores]      <- avoid

# A loop says what it means
for s in scores[:2]:
    print("Score:", s)
````

## Guided examples

### Example 1: Shopping cart with GST

````python
cart = {"notebook": 60, "pen": 10, "school bag": 450, "geometry box": 120}

with_gst = {item: round(price * 1.18, 2) for item, price in cart.items()}
print(with_gst)
print("Total payable: Rs.", round(sum(with_gst.values()), 2))
print("Items under Rs.100:", [item for item, price in cart.items() if price < 100])
````

### Example 2: Student results report

````python
results = {
    "Priya":  [88, 92, 79],
    "Arjun":  [45, 38, 52],
    "Riya":   [91, 85, 94],
    "Sanjay": [30, 35, 41],
}

averages = {name: sum(m) / len(m) for name, m in results.items()}
status = {name: ("Pass" if avg >= 40 else "Fail") for name, avg in averages.items()}
failed = [name for name, s in status.items() if s == "Fail"]
all_marks = [mark for marks in results.values() for mark in marks]

print("Averages:", {name: round(avg, 1) for name, avg in averages.items()})
print("Status:  ", status)
print("Failed:  ", failed)
print("Highest single mark:", max(all_marks))
````

### Example 3: Cleaning messy names

````python
raw_names = ["  priya SHARMA ", "ARJUN   verma", "riya  deshmukh "]

# split() with no arguments removes extra spaces; join() puts single spaces back
clean_names = [" ".join(name.split()).title() for name in raw_names]
print(clean_names)

initials = [".".join(part[0] for part in name.split()) + "." for name in clean_names]
print(initials)
````

### Example 4: Validating phone numbers

````python
numbers = ["9876543210", "98765", "9123456780", "abcdefghij", "8000000001"]

valid = [n for n in numbers if len(n) == 10 and n.isdigit()]
print("Valid:", valid)

masked = {n: n[:2] + "XXXXXX" + n[-2:] for n in valid}
print(masked)
print("Rejected:", len(numbers) - len(valid))
````

### Example 5: Food orders (list, set and dict together)

````python
orders = {
    "Priya": ["tea", "samosa"],
    "Arjun": ["coffee"],
    "Riya":  ["tea", "vada pav", "samosa"],
}

all_items = [item for items in orders.values() for item in items]          # list (flatten)
unique_items = {item for item in all_items}                                # set
popularity = {item: all_items.count(item) for item in unique_items}        # dict

print("All items ordered: ", all_items)
print("Distinct items:    ", sorted(unique_items))
print("Popularity:        ", dict(sorted(popularity.items())))
print("Items per customer:", {name: len(items) for name, items in orders.items()})
````

### Example 6: Attendance shortage (the 75% rule)

````python
classes_held = 20
attended = {"Priya": 19, "Arjun": 14, "Riya": 18, "Sanjay": 12, "Meera": 15}

percentage = {name: round(count / classes_held * 100) for name, count in attended.items()}
shortage = {name: pct for name, pct in percentage.items() if pct < 75}

print("Percentages:", percentage)
print("Below 75%:  ", shortage)
print("Safe:       ", sorted(name for name in percentage if name not in shortage))
````

## Common mistakes

| Mistake | What happens | Fix |
|---------|--------------|-----|
| `[x if x > 0 for x in nums]` | `SyntaxError`: `if` without `else` in front | Use `[x for x in nums if x > 0]` to filter, or add `else` |
| `[x for x in nums else 0]` | `SyntaxError` | `else` belongs with the front `if`: `[x if cond else 0 for x in nums]` |
| Using a colon when you wanted a set (or no colon when you wanted a dict) | `{k: v for ...}` builds a dict, `{x for ...}` builds a set | Check for the colon before the `for` |
| Expecting order from a set comprehension | Order is not guaranteed | Wrap in `sorted()` |
| Swapping dict keys and values with repeated values | Later keys overwrite earlier ones | Make sure values are unique first |
| Using `(x for x in ...)` and expecting a tuple | You get a generator | Use `tuple(...)` |
| Printing inside a comprehension | Builds a list of `None` | Use a normal loop |
| Nesting three or four levels deep | Unreadable | Break it into steps or use loops |

## Summary

### Cheat sheet

| Task | Code |
|------|------|
| Transform every item | `[f(x) for x in items]` |
| Keep only some items | `[x for x in items if cond]` |
| Choose between two values | `[a if cond else b for x in items]` |
| Two loops | `[(a, b) for a in xs for b in ys]` |
| Flatten a list of lists | `[v for row in rows for v in row]` |
| Build a grid | `[[... for j in ...] for i in ...]` |
| Unique results | `{f(x) for x in items}` |
| Build a dictionary | `{k: v for k, v in pairs}` |
| Reshape a dictionary | `{k: f(v) for k, v in d.items()}` |
| Filter a dictionary | `{k: v for k, v in d.items() if cond}` |
| Two lists to a dictionary | `{k: v for k, v in zip(keys, values)}` |
| Swap keys and values | `{v: k for k, v in d.items()}` |
| Lazy sum/max/any/all | `sum(f(x) for x in items)` |

### Key ideas

- A comprehension **builds a new collection**; it does not change the original.
- A filter `if` goes at the **end**. An `if/else` that picks a value goes at the **front**.
- `[ ]` gives a list, `{ }` gives a set, `{k: v}` gives a dict, `( )` gives a generator.
- Multiple `for` clauses read in the same order as nested loops.
- If it is hard to read, use a loop or a helper function.
