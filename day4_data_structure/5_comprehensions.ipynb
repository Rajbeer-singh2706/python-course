### What is a Set?

A **set** is an **unordered** collection of **unique** elements.

Best uses:
- Removing duplicates from a list
- Fast membership testing (O(1) vs O(n) for lists)
- Mathematical set operations (union, intersection…)

> ⚠️ **Gotcha:** `{}` creates an **empty dict**, not an empty set!  
> Use `set()` for an empty set.


## SECTION A — Creating Sets

* A set stores UNIQUE items only, with NO guaranteed order.
* Adding a duplicate simply has no effect.

````python
empty_set = set()   # NOT {}  — that creates an empty dict!
print(type(empty_set))
````

````python
mumbai_visitors = {"Priya", "Arjun", "Priya", "Riya"}
print("Mumbai visitors (duplicates removed):", mumbai_visitors)
print("Type:", type(mumbai_visitors))
````

````python
# ── Creating sets ────────────────────────────────────────────────────
tags    = {'python', 'web', 'data', 'python'}  # duplicate dropped!
print(tags)           # order not guaranteed

# ── Add and remove ────────────────────────────────────────────────────
tags.add('api')
tags.add('python')          # no duplicate added
tags.discard('web')         # safe remove — no error if missing
print(tags)
````

````python
# Converting a list to a set is a quick way to de-duplicate.
repeated_cities = ["Mumbai", "Delhi", "Mumbai", "Pune", "Delhi"]
unique_cities = set(repeated_cities)
print("Unique cities:", unique_cities)
# Output: Unique cities: {'Mumbai', 'Delhi', 'Pune'}
````

````python
# ── Remove duplicates from a list ────────────────────────────────────
nums = [1, 2, 2, 3, 3, 3, 4, 5, 5]
unique = list(set(nums))
print(sorted(unique))  # [1, 2, 3, 4, 5]
````

## SECTION B — Set Operations

````python
# ── Set operations ───────────────────────────────────────────────────
backend_skills  = {'python', 'sql', 'docker', 'git'}
frontend_skills = {'javascript', 'css', 'git', 'docker'}

print('Union (all skills):')
print(backend_skills | frontend_skills)

print('\nIntersection (shared skills):')
print(backend_skills & frontend_skills)

print('\nDifference (backend only):')
print(backend_skills - frontend_skills)

print('\nSymmetric difference (not shared):')
print(backend_skills ^ frontend_skills)

# ── Membership (O(1) — much faster than lists for large collections) ──
allowed_users = {'alice', 'bob', 'charlie', 'diana'}
print('\nalice allowed?', 'alice' in allowed_users)    # True
print('mallory allowed?', 'mallory' in allowed_users) # False
````

````python
# Adding, removing, and membership testing
mumbai_visitors.add("Kavita")
mumbai_visitors.discard("Arjun")     # no error even if item is absent
print("Updated set:", mumbai_visitors)
print("Is 'Priya' in the set?", "Priya" in mumbai_visitors)
# Output:
# Updated set: {'Priya', 'Riya', 'Kavita'}   (order may vary)
# Is 'Priya' in the set? True
````

## SECTION C — List Comprehensions

### List Comprehensions

A **list comprehension** builds a new list in a single, readable expression:

```python
[expression  for item in iterable]
[expression  for item in iterable  if condition]
```

Compare:
```python
# Old-style loop
squares = []
for x in range(10):
    squares.append(x**2)

# Comprehension — same result, one line
squares = [x**2 for x in range(10)]
```


````python
# ── List comprehension examples ──────────────────────────────────────
squares = [x**2 for x in range(10)]
print('Squares:', squares)

evens = [x for x in range(20) if x % 2 == 0]
print('Evens:', evens)

upper = [word.upper() for word in ['hello', 'world', 'python']]
print('Upper:', upper)

# Filtering a list of dicts
people = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob',   'age': 17},
    {'name': 'Carol', 'age': 25},
    {'name': 'Dave',  'age': 16},
]
adults = [p['name'] for p in people if p['age'] >= 18]
print('Adults:', adults)
````

````python
# A comprehension is a compact way to build a new collection from
# an existing iterable, optionally filtering with an `if` clause.
# The long way, with a regular loop:
squares_loop = []
for n in range(1, 6):
    squares_loop.append(n ** 2)
print("Squares (loop):", squares_loop)

# The comprehension way -- same result, one line:
squares = [n ** 2 for n in range(1, 6)]
print("Squares (comprehension):", squares)

# Comprehensions can filter with an `if` clause.
prices_in_rupees = [120, 560, 45, 899, 300, 610]
expensive_items = [price for price in prices_in_rupees if price > 500]
print("Prices above Rs.500:", expensive_items)
# Output: Prices above Rs.500: [560, 899, 610]
````

## SECTION D — Dict and Set Comprehensions

### Dict and Set Comprehensions

The same idea works for dicts and sets:

```python
{key_expr: val_expr  for item in iterable}   # dict
{expr  for item in iterable}                 # set
```


````python

price_in_rupees = {"tea": 15, "coffee": 25, "samosa": 20}
doubled_prices = {item: price * 2 for item, price in price_in_rupees.items()}
print("Doubled prices:", doubled_prices)
# Output: Doubled prices: {'tea': 30, 'coffee': 50, 'samosa': 40}

even_set = {n for n in range(10) if n % 2 == 0}
print("Even numbers set:", even_set)
# Output: Even numbers set: {0, 2, 4, 6, 8}
````

````python
# ── Dict comprehension ───────────────────────────────────────────────
words = ['apple', 'banana', 'cherry', 'date']
word_lengths = {word: len(word) for word in words}
print(word_lengths)

# Invert a dict (swap keys and values)
original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}
print(inverted)   # {1: 'a', 2: 'b', 3: 'c'}

# ── Set comprehension ────────────────────────────────────────────────
nums = [1, 1, 2, 3, 3, 4, 5, 5, 5]
unique_squares = {x**2 for x in nums}
print(sorted(unique_squares))  # [1, 4, 9, 16, 25]

# ── Conditional expression in comprehension ───────────────────────────
# 'FizzBuzz lite'
labels = ['Fizz' if x % 3 == 0 else str(x) for x in range(1, 11)]
print(labels)
````

## GUIDED EXAMPLES

````python
# Example 1: Finding common subjects between two students using sets
priya_subjects = {"Math", "Physics", "Chemistry"}
arjun_subjects = {"Physics", "Biology", "Math"}
common_subjects = priya_subjects & arjun_subjects
print(f"Example 1 -- Common subjects: {common_subjects}")

# Example 2: List comprehension to convert Celsius to Fahrenheit
celsius_temps = [25, 30, 22, 35]
fahrenheit_temps = [(c * 9 / 5) + 32 for c in celsius_temps]
print(f"Example 2 -- Fahrenheit temps: {fahrenheit_temps}")

# Example 3: Dict comprehension to build a name-length lookup
names = ["Priya", "Arjun", "Riya Deshmukh", "Sanjay Verma"]
name_lengths = {name: len(name) for name in names}
print(f"Example 3 -- Name lengths: {name_lengths}")

# Example 4: Set comprehension to find unique first letters of cities
cities = ["Mumbai", "Delhi", "Manali", "Dehradun", "Pune"]
unique_first_letters = {city[0] for city in cities}
print(f"Example 4 -- Unique first letters: {unique_first_letters}")
````

### What We Covered

| Topic | Key Takeaway |
|-------|--------------|
| **Lists** | Ordered, mutable; use for sequences that change |
| **Tuples** | Ordered, immutable; use for fixed data, multi-return |
| **Dicts** | Key→value; use `.get()` for safe access |
| **Sets** | Unique items; great for deduplication & membership tests |
| **Slicing** | `seq[start:stop:step]`; `[::-1]` reverses any sequence |
| **Comprehensions** | `[expr for x in iterable if cond]` — concise & Pythonic |

### Key Principles

- **Mutability matters** — tuples protect data from accidental changes
- **Sets beat lists** for membership tests when the collection is large (O(1) vs O(n))
- **`.get()` is your friend** — always use it on dicts when the key might not exist
- **Comprehensions > loops** when you're just building a new collection
