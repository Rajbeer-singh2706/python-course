````python
# ── Solution ──────────────────────────────────────────────
import logging, time, random, functools
logging.basicConfig(level=logging.INFO, format="%(levelname)-8s %(message)s")
logger = logging.getLogger("retry")

def retry(max_attempts, delay=0):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            last_exc = None
            for attempt in range(1, max_attempts + 1):
                try:
                    result = func(*args, **kwargs)
                    logger.info("%s succeeded on attempt %d", func.__name__, attempt)
                    return result
                except Exception as e:
                    last_exc = e
                    logger.warning("%s attempt %d/%d failed: %s",
                                   func.__name__, attempt, max_attempts, e)
                    if delay and attempt < max_attempts:
                        time.sleep(delay)
            raise last_exc
        return wrapper
    return decorator

random.seed(42)

@retry(max_attempts=5, delay=0)
def fetch_exchange_rate(currency):
    if random.random() < 0.7:
        raise ConnectionError(f"Timeout fetching rate for {currency}")
    return round(random.uniform(80, 90), 2)

try:
    rate = fetch_exchange_rate("USD")
    print(f"Exchange rate: {rate}")
except ConnectionError as e:
    print(f"All attempts failed: {e}")
````

### Problem 18 🔴 — Full Error-Handling Pipeline
Build a `StudentReportPipeline` that:
1. Loads student data from JSON (list of `{"name": str, "scores": [int, ...]}`)
2. Validates each student: name must be a non-empty string, scores must be a non-empty list of 0–100 ints
3. Computes average score for each valid student
4. Saves a summary JSON: `[{"name": ..., "average": ..., "grade": ...}]`
5. Logs every step — `DEBUG` for detail, `INFO` for milestones, `WARNING` for skips, `ERROR` for failures
6. Uses custom exceptions: `PipelineError`, `ValidationError(PipelineError)`, `ExportError(PipelineError)`


````python
import json, logging, tempfile, os
logging.basicConfig(level=logging.DEBUG, format="%(levelname)-8s %(message)s")
logger = logging.getLogger("pipeline")

# YOUR CODE HERE

# Test data
test_data = [
    {"name": "Priya Sharma",  "scores": [88, 92, 85, 90]},
    {"name": "Arjun Patel",   "scores": [72, 68, 75]},
    {"name": "",              "scores": [80]},           # invalid name
    {"name": "Kavya",        "scores": [105, 88]},       # invalid score
    {"name": "Rohan Mehta",  "scores": []},              # empty scores
    {"name": "Sneha Joshi",  "scores": [95, 98, 92, 97]},
]

# Save test input
src = tempfile.mktemp(suffix=".json")
dst = tempfile.mktemp(suffix=".json")
with open(src, "w") as f:
    json.dump(test_data, f)

# pipeline = StudentReportPipeline()
# pipeline.run(src, dst)

# if os.path.exists(dst):
#     with open(dst) as f:
#         print(json.dumps(json.load(f), indent=2))

for p in [src, dst]:
    if os.path.exists(p): os.unlink(p)
````

````python
# ── Solution ──────────────────────────────────────────────
import json, logging, tempfile, os
logging.basicConfig(level=logging.DEBUG, format="%(levelname)-8s %(message)s")
logger = logging.getLogger("pipeline")

class PipelineError(Exception): pass
class ValidationError(PipelineError):
    def __init__(self, name, reason): super().__init__(f"'{name}': {reason}")
class ExportError(PipelineError): pass

GRADE = [(90,'A'),(80,'B'),(70,'C'),(60,'D'),(0,'F')]

class StudentReportPipeline:
    def _load(self, path):
        try:
            with open(path, encoding="utf-8") as f:
                data = json.load(f)
            logger.info("Loaded %d records from %s", len(data), path)
            return data
        except FileNotFoundError as e:
            raise PipelineError(f"Input file not found: {path}") from e
        except json.JSONDecodeError as e:
            raise PipelineError(f"Invalid JSON: {e}") from e

    def _validate(self, record):
        name = record.get("name", "")
        if not isinstance(name, str) or not name.strip():
            raise ValidationError(name or "<empty>", "name must be a non-empty string")
        scores = record.get("scores", [])
        if not scores:
            raise ValidationError(name, "scores list is empty")
        for s in scores:
            if not isinstance(s, (int, float)) or not 0 <= s <= 100:
                raise ValidationError(name, f"invalid score {s}")
        return name.strip(), [float(s) for s in scores]

    def _grade(self, avg):
        for threshold, letter in GRADE:
            if avg >= threshold: return letter

    def _save(self, summary, path):
        try:
            with open(path, "w", encoding="utf-8") as f:
                json.dump(summary, f, indent=2)
            logger.info("Summary saved to %s", path)
        except OSError as e:
            raise ExportError(f"Cannot write {path}: {e}") from e

    def run(self, input_path, output_path):
        records = self._load(input_path)
        summary = []
        for rec in records:
            try:
                name, scores = self._validate(rec)
                avg = sum(scores) / len(scores)
                grade = self._grade(avg)
                logger.debug("%s -> avg=%.1f grade=%s", name, avg, grade)
                summary.append({"name": name, "average": round(avg, 1), "grade": grade})
            except ValidationError as e:
                logger.warning("Skipping record: %s", e)
        self._save(summary, output_path)
        return summary

test_data = [
    {"name": "Priya Sharma",  "scores": [88, 92, 85, 90]},
    {"name": "Arjun Patel",   "scores": [72, 68, 75]},
    {"name": "",              "scores": [80]},
    {"name": "Kavya",        "scores": [105, 88]},
    {"name": "Rohan Mehta",  "scores": []},
    {"name": "Sneha Joshi",  "scores": [95, 98, 92, 97]},
]
src = tempfile.mktemp(suffix=".json")
dst = tempfile.mktemp(suffix=".json")
with open(src, "w") as f:
    json.dump(test_data, f)

result = StudentReportPipeline().run(src, dst)
print("\nFinal Summary:")
print(json.dumps(result, indent=2))

for p in [src, dst]:
    if os.path.exists(p): os.unlink(p)
````
