## 1. The story: three lines for a tiny job

Priya works in Mumbai. At month end she looks at her expenses and asks, *"What are my amounts, just the numbers?"*

Here is her data. Each expense is a dictionary (Day 4):

````python
expenses = [
    {"name": "Chai", "amount": 20, "category": "Food"},
    {"name": "Auto", "amount": 120, "category": "Travel"},
    {"name": "Groceries", "amount": 850, "category": "Food"},
    {"name": "Rent", "amount": 9000, "category": "Home"},
    {"name": "Movie", "amount": 350, "category": "Fun"},
    {"name": "Metro", "amount": 60, "category": "Travel"},
]
print(len(expenses), "expenses loaded")
````

With a `for` loop, she writes **three lines** just to collect the amounts:

````python
amounts = []
for item in expenses:
    amounts.append(item["amount"])
print(amounts)
````

Now the same job in **one line**:

````python
amounts = [item["amount"] for item in expenses]
print(amounts)
````

Same answer. This one-line style is called a **list comprehension**.

**Real-life picture:** imagine telling the *munim ji* at the kirana shop, "Write down every item's price in a new list."
Without a comprehension you dictate three steps. With one, you say **one sentence**.

> A comprehension is **not** a new kind of loop. It is a **short way to write the "make a new list" loop** you already know.

---
## 2. The syntax, part by part

```text
new_list = [ expression   for item in old_list ]
             ^^^^^^^^^^   ^^^^^^^^^^^^^^^^^^^^
             what goes    the same for line
             into the     you already know
             new list
```

| Part | Meaning |
|------|---------|
| `[ ]` | Square brackets: the result is a **new list** |
| `expression` | What to put in the new list, for each item (it is the thing you used to `.append()`) |
| `for item in old_list` | The loop line, **without** the colon |

**Read it aloud:** "Give me `expression`, **for each** `item` **in** `old_list`."

### Turn any loop into a comprehension in 3 steps

| Step | Loop version | Comprehension |
|------|--------------|---------------|
| 1 | `result = []` | write `[` `]` |
| 2 | `result.append(item * 2)` | put `item * 2` **first** inside the brackets |
| 3 | `for item in numbers:` | add `for item in numbers` after it |

```text
result = []                         result = [ item * 2  for item in numbers ]
for item in numbers:         -->
    result.append(item * 2)
```

## 3. Three very simple examples (traced)

### Example A: double every number

````python
numbers = [10, 20, 30]
doubled = [n * 2 for n in numbers]
print(doubled)
````

The comprehension works **round by round**, just like a loop:

| Round | `n` | `n * 2` goes into the new list | New list so far |
|-------|-----|-------------------------------|-----------------|
| 1 | 10 | 20 | `[20]` |
| 2 | 20 | 40 | `[20, 40]` |
| 3 | 30 | 60 | `[20, 40, 60]` |

### Example B: squares of 1 to 5 (with `range`)

````python
squares = [n * n for n in range(1, 6)]
print(squares)
````

### Example C: make names look neat

````python
names = ["priya", "arjun", "riya"]
neat = [name.title() for name in names]
print(neat)
print(names)   # the old list did not change
````

Notice the last line: the **old list is untouched**. A comprehension always builds a **new** one.

## Concept 3: Keep only some items (filter with `if`)

Add an `if` at the **end** to keep only the items that pass the test.

```text
[ expression  for item in old_list  if condition ]
```

````python
amounts = [20, 120, 850, 9000, 350, 60]
big = [a for a in amounts if a > 100]
small = [a for a in amounts if a <= 100]
print("Big:", big)
print("Small:", small)
````

**What to notice**

- The `if` at the **end** works like a gate: items that fail are **left out**
- The new list can be **shorter** than the old one

## Concept 4: Choose between two values (`if` ... `else` in front)

When every item must give an answer, but the answer depends on a test, put `if ... else` **before** `for`.

```text
[ value_if_true  if condition  else  value_if_false   for item in old_list ]
```

````python
amounts = [20, 120, 850, 9000]
tags = ["High" if a > 500 else "Low" for a in amounts]
print(tags)
````

**What to notice**

| Goal | Where the `if` goes | Size of new list |
|------|--------------------|------------------|
| **Leave some out** | at the **end** (`... for x in y if test`) | can be shorter |
| **Pick one of two values** | in **front** with `else` | same as the old list |

The most common mistake is mixing these two places up.

## Concept 5: Comprehensions on text

A string is also something you can loop over, letter by letter. Names are a common source.

````python
names = ["Priya", "Arjun", "Riya"]
first_letters = [n[0] for n in names]
lengths = [len(n) for n in names]
shouting = [n.upper() for n in names]
print(first_letters)
print(lengths)
print(shouting)
````

- Any method works inside: `.upper()`, `.lower()`, `.title()`

## Concept 7: Comprehension + `sum()`, `max()`, `min()`, `len()`

A comprehension builds a list. Then a built-in function can total or count it. This is how you get **totals in one line**.

````python
expenses = [
    {"name": "Chai", "amount": 20, "category": "Food"},
    {"name": "Groceries", "amount": 850, "category": "Food"},
    {"name": "Rent", "amount": 9000, "category": "Home"},
]
total = sum([e["amount"] for e in expenses])
food_total = sum([e["amount"] for e in expenses if e["category"] == "Food"])
biggest = max([e["amount"] for e in expenses])
count_big = len([e for e in expenses if e["amount"] > 100])
print("Total:", total)
print("Food total:", food_total)
print("Biggest:", biggest)
print("Big expenses:", count_big)
````

**What to notice**

- `sum`, `max`, `min` and `len` take the new list and give **one number**
- `len([...])` is a neat way to **count** how many items pass a test

## Concept 8: Dictionary comprehension

Use curly brackets with `key: value` to build a **new dictionary**.

```text
{ key : value  for item in old_list }
```

````python
expenses = [
    {"name": "Chai", "amount": 20, "category": "Food"},
    {"name": "Auto", "amount": 120, "category": "Travel"},
    {"name": "Movie", "amount": 350, "category": "Fun"},
]
price_of = {e["name"]: e["amount"] for e in expenses}
print(price_of)

# Looping over an existing dictionary with .items()
doubled = {name: price * 2 for name, price in price_of.items()}
over_100 = {name: price for name, price in price_of.items() if price > 100}
print(doubled)
print(over_100)
````

**What to notice**

- The colon `:` separates the key from the value
- `.items()` gives `name, price` pairs (you used it in `for` loops)
- The `if` at the end filters, just like for lists

## Concept 9: Set comprehension (unique values)

Curly brackets **without** a colon build a **set**. A set keeps each value **only once** (Day 4).

````python
expenses = [
    {"name": "Chai", "amount": 20, "category": "Food"},
    {"name": "Auto", "amount": 120, "category": "Travel"},
    {"name": "Groceries", "amount": 850, "category": "Food"},
    {"name": "Metro", "amount": 60, "category": "Travel"},
]
categories = {e["category"] for e in expenses}
print(sorted(categories))
````

**What to notice**

- `Food` and `Travel` each appear twice in the data but only **once** in the set
- A set has **no fixed order**, so we wrap it in `sorted()` to print it in a steady order

## Concept 11: A comprehension makes a **new** list

The original list is never changed.

````python
amounts = [20, 120, 850]
bigger = [a + 100 for a in amounts]
print("New list:", bigger)
print("Old list:", amounts)
````

**What to notice**

- If you want the old list changed, you must **assign the answer back**: `amounts = [a + 100 for a in amounts]`

## Concept 12: When to use a loop instead

Comprehensions are great for **one small job**. Use a normal `for` loop when:

| Situation | Use |
|-----------|-----|
| You want to **print** something for each item | normal `for` loop |
| You need **several steps** per item | normal `for` loop |
| The comprehension does not fit **comfortably in one line** | normal `for` loop |
| You need `break` or `continue` | normal `for` loop |

### Do not do this

````python
# Bad habit: using a comprehension just to print
result = [print("Spent", a) for a in [20, 120]]
print(result)   # a useless list of None
````

The `print` worked, but the list `[None, None]` is junk. For printing, use a plain loop:

````python
for a in [20, 120]:
    print("Spent", a)
````

### Common mistakes

| Mistake | Wrong | Right |
|---------|-------|-------|
| Colon at the end of the `for` | `[a for a in x:]` | `[a for a in x]` |
| `if/else` at the end | `[a for a in x if a > 5 else 0]` | `[a if a > 5 else 0 for a in x]` |
| Round brackets by accident | `(a for a in x)` makes something different (you meet it on Day 13) | use `[ ]` |
| Forgetting to save the result | `[a * 2 for a in x]` | `doubled = [a * 2 for a in x]` |

## Recap

| Kind | Brackets | Shape | Result |
|------|----------|-------|--------|
| List comprehension | `[ ]` | `[expr for x in data]` | new list |
| With filter | `[ ]` | `[expr for x in data if test]` | shorter list |
| With choice | `[ ]` | `[a if test else b for x in data]` | same length |
| Dict comprehension | `{ : }` | `{k: v for x in data}` | new dict |
| Set comprehension | `{ }` | `{expr for x in data}` | unique values |

**Rule of thumb:** write the loop first. If it is just "make a new list/dict/set", turn it into a comprehension.

# Part C: PaisaTracker with comprehensions

We build in **three small stages**: skeleton, real data, then the safe finished version.

### Stage 1: skeleton (just the amounts)

````python
# Typed answers (simulating what Priya would enter)
typed = ["20", "120", "850", "9000", "350", "60"]

# Text to numbers, in one line
amounts = [int(t) for t in typed]
print(amounts)
print("Total:", sum(amounts))
````

Real typing later would be `typed.append(input("Amount: "))`. Here we simulate it so the notebook runs by itself.

### Stage 2: real expense data

````python
expenses = [
    {"name": "Chai", "amount": 20, "category": "Food"},
    {"name": "Auto", "amount": 120, "category": "Travel"},
    {"name": "Groceries", "amount": 850, "category": "Food"},
    {"name": "Rent", "amount": 9000, "category": "Home"},
    {"name": "Movie", "amount": 350, "category": "Fun"},
    {"name": "Metro", "amount": 60, "category": "Travel"},
]
salary = 25000
LIMIT = 500

total = sum([e["amount"] for e in expenses])
over_limit = [e["name"] for e in expenses if e["amount"] > LIMIT]
categories = sorted({e["category"] for e in expenses})
category_totals = {
    c: sum([e["amount"] for e in expenses if e["category"] == c])
    for c in categories
}
print("Total:", total)
print("Over limit:", over_limit)
print("Categories:", categories)
print("Category totals:", category_totals)
````

### Stage 3: the finished report (loop for printing, comprehensions for thinking)

````python
print("=" * 40)
print("PaisaTracker: Priya's month")
print("=" * 40)

for c in categories:
    print(c, "=> Rs", category_totals[c])

print("-" * 40)
print("Total spent :", total)
print("Salary left :", salary - total)
print("Spent % of salary :", round(total / salary * 100, 1))

if over_limit:
    print("Warning! Above Rs", LIMIT, ":", over_limit)
````

### Which concept is used where

| Line | Concept |
|------|---------|
| `[int(t) for t in typed]` | Concept 1: transform every item |
| `sum([e["amount"] for e in expenses])` | Concept 7: comprehension + `sum` |
| `[e["name"] for e in expenses if e["amount"] > LIMIT]` | Concept 3 and 6: filter on a list of dicts |
| `{e["category"] for e in expenses}` | Concept 9: set comprehension |
| `{c: sum([...]) for c in categories}` | Concept 8 + 7: dict comprehension with a total inside |
| `for c in categories: print(...)` | Concept 12: printing stays in a normal loop |

## Summary

- A comprehension is a **short way to build a new list, dict or set** from a loop
- `[expr for x in data]` transforms, `... if test` filters, `a if test else b` chooses
- `{k: v for ...}` builds a dict, `{expr for ...}` builds a set
- Use `sum`, `max`, `min`, `len` on the result for totals and counts
- Printing and multi-step work stay in a normal `for` loop
