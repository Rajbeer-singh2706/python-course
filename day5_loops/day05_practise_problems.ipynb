### Problem E2  🔴 Challenge: PaisaTracker: budget guard

Priya's monthly budget is 5000. Add each new expense dict to the list  
and keep a running total. If the total goes ABOVE the budget, print a  
warning with the expense name and break. If the loop finishes without  
break, print: Within budget.

**Expected output**

```text
Warning: phone bill pushed the total to ₹5200 (budget ₹5000)
Expenses recorded: 3
```

````python
budget = 5000
incoming = [
    {"name": "rent", "amount": 3000},
    {"name": "groceries", "amount": 1500},
    {"name": "phone bill", "amount": 700},
    {"name": "chai", "amount": 20},
]
expenses = []
total = 0
index = 0
# YOUR CODE HERE
````

<details>
<summary><b>Show solution</b> (try first!)</summary>

```python
while index < len(incoming):
    expense = incoming[index]
    expenses.append(expense)
    total += expense["amount"]
    if total > budget:
        print(f"Warning: {expense['name']} pushed the total to ₹{total} (budget ₹{budget})")
        break
    index += 1
else:
    print("Within budget")
print(f"Expenses recorded: {len(expenses)}")
```

</details>

### Problem E3  🔴 Challenge: PaisaTracker: category summary

Use a while loop (with an index) to build a dict of totals per  
category from the list of tuples (name, amount, category). Then use a  
for loop to print each category total, and finally print the category  
with the highest total.

**Expected output**

```text
food: ₹660
travel: ₹115
housing: ₹12000
Biggest category: housing
```

````python
entries = [
    ("chai", 20, "food"),
    ("auto-rickshaw", 85, "travel"),
    ("groceries", 640, "food"),
    ("metro", 30, "travel"),
    ("rent", 12000, "housing"),
]
totals = {}
index = 0
# YOUR CODE HERE
````

<details>
<summary><b>Show solution</b> (try first!)</summary>

```python
while index < len(entries):
    name, amount, category = entries[index]
    if category in totals:
        totals[category] += amount
    else:
        totals[category] = amount
    index += 1

biggest = None
for category in totals:
    print(f"{category}: ₹{totals[category]}")
    if biggest is None or totals[category] > totals[biggest]:
        biggest = category
print(f"Biggest category: {biggest}")
```
