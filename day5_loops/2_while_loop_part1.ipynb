## 1️⃣ The Story — Arjun and the Tatkal Ticket


It is 9:58 AM. Arjun wants to book a train ticket. The booking page opens at 10:00, but the website is jammed.

So Arjun **keeps pressing refresh** until the page opens.

> ❓ How many times will he refresh?
>
> He does not know! Maybe 3 times, maybe 40.

He just thinks: **"WHILE the page is not open, keep refreshing."**

That sentence is a `while` loop.

````python
# The server's replies, in order (we prepared them so the notebook can run)
replies = ["busy", "busy", "busy", "OPEN"]

page_status = "busy"       # at the start, the page is busy
refresh_count = 0          # how many times Arjun has refreshed

while page_status != "OPEN":                 # keep going WHILE the page is not open
    page_status = replies[refresh_count]     # refresh and see the reply
    refresh_count += 1                       # count this refresh
    print("Refresh number", refresh_count, "→", page_status)

print("🎉 The booking page opened!")
````

We never wrote "4 times". The **condition** (`page_status != "OPEN"`) decided when to stop.

## 2️⃣ The 3 Things Every `while` Loop Needs


```python
count = 1                  # ① START: a variable to begin with

while count <= 5:          # ② CONDITION: keep going while this is True
    print(count)
    count += 1             # ③ UPDATE: change the variable, so the loop can end
```

| Ingredient | Job | If you forget it… |
|-----------|-----|-------------------|
| ① Start | gives the loop something to check | 💥 `NameError` |
| ② Condition | decides: run again or stop | the loop may not behave as planned |
| ③ Update | moves the loop towards the end | ♾️ **infinite loop** |

### How Python reads a `while` loop

```
        ┌──────────────────────────┐
        ▼                          │
  Is the condition True?           │
     │            │                │
    YES           NO               │
     │            │                │
  run the body    skip the loop,   │
     │            carry on below   │
     └────────────────────────────┘
```

The condition is checked **before every round**.

````python
# Your first while loop: count from 1 to 5
count = 1                         # ① start

while count <= 5:                 # ② condition
    print("Count is", count)
    count += 1                    # ③ update  (same as: count = count + 1)

print("Done! The loop stopped because count is now", count)
````

### 🔍 What happened inside? (a trace table)

A **trace table** shows the variable after every round. Use it whenever a loop confuses you.

````python
# The same idea, printing a trace table
count = 1
print("round | count (before) | count (after)")
round_number = 1
while count <= 3:
    before = count
    count += 1
    print(f"{round_number:>5} | {before:>14} | {count:>13}")
    round_number += 1
print("count is now 4, so 4 <= 3 is False → the loop stops")
````

## 3️⃣ Counting Up and Counting Down

* **Counting up** → start small, **add** in the update, use `<` or `<=`
* **Counting down** → start big, **subtract** in the update, use `>` or `>=`
* You can jump by more than 1: `count += 2`

````python
# Counting up
n = 1
while n <= 5:
    print(n)
    n += 1

print("-" * 20)

# Counting down  (like a rocket launch)
n = 5
while n >= 1:
    print(n)
    n -= 1
print("🚀 Lift off!")
````

````python
# Jumping by 2: the even numbers up to 20
n = 2
while n <= 20:
    print(n, end=" ")        # end=" " keeps everything on one line
    n += 2
print()                      # moves to a new line at the end
````

## 4️⃣ Adding Things Up (Totals)

Very often a loop **collects a total**. The recipe is always:

1. Make `total = 0` **before** the loop
2. Add to it **inside** the loop
3. Show it **after** the loop

````python
# Add the numbers 1 to 10
number = 1
total = 0                    # ① before the loop

while number <= 10:
    total += number          # ② inside the loop
    number += 1

print("Sum of 1 to 10 =", total)    # ③ after the loop
````

````python
# Real life: how many days will my pocket money last?
money = 500
daily_spend = 120
days = 0

while money >= daily_spend:          # can I still afford one more day?
    money -= daily_spend
    days += 1
    print(f"Day {days}: spent Rs {daily_spend}, left Rs {money}")

print(f"\nMy money lasted {days} full days. Rs {money} is left over.")
````

## 5️⃣ Conditions with `and`, `or`, `not`

You can use the logical operators from Day 3 inside a `while` condition.

| Operator | Loop keeps going while… |
|----------|-------------------------|
| `and` | **both** parts are True |
| `or` | **at least one** part is True |
| `not` | the thing is **False** |

````python
# `and` → Riya buys sweets while she has money AND has bought fewer than 3
budget = 100
price = 30
bought = 0

while budget >= price and bought < 3:
    budget -= price
    bought += 1
    print(f"Bought sweet #{bought}. Money left: Rs {budget}")

print(f"\nRiya bought {bought} sweets and has Rs {budget} left.")
````

````python
# `or` → keep playing while the answer is "y" OR "Y"
answers = ["y", "Y", "n"]            # what the player types after each round
position = 0
answer = answers[position]

while answer == "y" or answer == "Y":
    print("Playing round", position + 1)
    position += 1
    answer = answers[position]       # the player answers again

print("Game over. The player typed:", answer)
````

````python
# `not` → keep looking while we have NOT found the lucky number
numbers = [4, 9, 15, 7, 22]
found = False
index = 0

while not found:                     # means: while found is False
    print("Checking", numbers[index])
    if numbers[index] == 7:
        found = True
    index += 1

print("Found 7! ✅")
````

## 6️⃣ A Loop That Runs Zero Times

The condition is checked **first**. If it is `False` straight away, the body **never runs**. This is normal, not an error.

````python
n = 10
while n < 5:                       # 10 < 5 is False from the very start
    print("This line will never print")
    n += 1

print("The loop body ran zero times. n is still", n)
````

## 7️⃣ Infinite Loops ♾️

An **infinite loop** never stops, because the condition **never becomes False**.

The #1 reason: **you forgot the update line** (ingredient ③).

```python
count = 0
while count < 5:
    print(count)
    # forgot count += 1   → count is always 0 → runs forever!
```

### 🛑 How to stop one
| Where you run Python | How to stop |
|----------------------|-------------|
| Jupyter notebook | click the **■ Interrupt** button |
| VS Code terminal | press **Ctrl + C** |

### 🛡️ A safety net
Add a **guard counter** so even a buggy loop cannot run forever. Below we *deliberately* forget the update, and the guard rescues us.

````python
count = 0
guard = 0                                # the safety net

while count < 5:
    print("count is", count)             # oops: count never changes!
    guard += 1
    if guard == 6:                       # emergency exit after 6 rounds
        print("🛑 Emergency stop! The loop was infinite.")
        break

print("\nFix: add 'count += 1' inside the loop.")
````

### Four causes of infinite loops

| Cause | Example | Fix |
|-------|---------|-----|
| Forgot the update | no `count += 1` | add it |
| Update goes the wrong way | `while n < 5:` with `n -= 1` | use `+=` |
| Condition always True | `while 1 < 2:` | write a real condition |
| Typo in the variable name | updating `cont` instead of `count` | read carefully |

## 8️⃣ `break`, `continue`, `pass`

| Keyword | What it does | Easy way to remember |
|---------|--------------|----------------------|
| `break` | **leaves the loop immediately** | "Stop! I am done." |
| `continue` | **skips the rest of this round**, goes to the next | "Skip this one." |
| `pass` | **does nothing** | "Placeholder." |

````python
# break → stop shopping when the budget would be crossed
prices = [120, 80, 250, 60, 300]
budget = 400
total = 0
i = 0

while i < len(prices):
    if total + prices[i] > budget:
        print(f"Item {i + 1} (Rs {prices[i]}) is too expensive. Stopping!")
        break                                  # leave the loop right now
    total += prices[i]
    print(f"Added item {i + 1}. Total so far: Rs {total}")
    i += 1

print("Final total:", total)
````

````python
# continue → print only the odd numbers from 1 to 9
n = 0
while n < 10:
    n += 1                  # ⚠️ update FIRST, before continue
    if n % 2 == 0:          # even number?
        continue            # skip the print and go back to the top
    print(n)
````

> ⚠️ **Careful with `continue` in a `while` loop!**
> `continue` jumps back to the top and **skips everything below it**, including the update line. If your update is *below* `continue`, the loop gets stuck forever.
>
> ```python
> n = 0
> while n < 10:
>     if n % 2 == 0:
>         continue        # jumps up, n never changes → infinite loop!
>     n += 1
> ```
> **Rule:** do the update *before* `continue`, as we did above.

````python
# pass → a placeholder for code you will write later
n = 0
while n < 3:
    if n == 1:
        pass                      # TODO: handle n == 1 later (does nothing for now)
    else:
        print("n is", n)
    n += 1
````

## 9️⃣ `while True` — "Keep going until I say stop"

`while True:` runs **forever**, so the only way out is a `break`. It is perfect for menus and programs that wait for the user to quit.

```python
while True:
    ...
    if time_to_stop:
        break
```

````python
# A chai stall takes orders until the customer says "stop"
orders = ["tea", "coffee", "tea", "stop", "coffee"]   # what the customer says
position = 0
cups = 0

while True:
    order = orders[position]
    position += 1

    if order == "stop":
        print("No more orders. Bill time!")
        break                              # the ONLY way out of this loop

    cups += 1
    print(f"Preparing {order} ☕  (cup #{cups})")

print("Total cups:", cups)
print("(The last 'coffee' came after 'stop', so it was never made.)")
````

## 🔟 `else` on a `while` Loop

An `else` after a loop runs **only if the loop ended normally**, meaning **without hitting `break`**.

> Think: **"I searched everywhere and never broke out, so it must be missing."**

````python
# Search for a number in a list
numbers = [3, 9, 12, 15]
target = 7
i = 0

while i < len(numbers):
    if numbers[i] == target:
        print(target, "found at position", i)
        break                              # found it → else will NOT run
    i += 1
else:
    print(target, "is not in the list")    # runs only because there was no break
````

````python
# Same code, but now the number IS there
numbers = [3, 9, 12, 15]
target = 12
i = 0

while i < len(numbers):
    if numbers[i] == target:
        print(target, "found at position", i)
        break
    i += 1
else:
    print(target, "is not in the list")
````

## 1️⃣1️⃣ Asking Again Until the Answer Is Valid

A very common use: **keep asking until the user gives a sensible answer**.

In a real program you would write `answer = input("Age: ")`. Here we use a ready-made list of typed answers.

````python
typed = ["abc", "-5", "200", "25"]      # what the user types, one try at a time
position = 0
age = None                               # None = "no valid age yet"

while age is None:
    answer = typed[position]             # real program: answer = input("Enter your age: ")
    position += 1

    if answer.isdigit() and 1 <= int(answer) <= 120:
        age = int(answer)
        print(f"'{answer}' ✅ accepted")
    else:
        print(f"'{answer}' ❌ not a valid age. Try again!")

print("Your age is", age)
````

## 1️⃣2️⃣ A Loop Inside a Loop

The **inner** loop runs *completely* for every round of the **outer** loop.

⚠️ Remember to **reset the inner counter** each time the outer loop goes round.

````python
# Multiplication grid: 3 rows × 5 columns
row = 1
while row <= 3:
    col = 1                              # reset the inner counter for every new row!
    while col <= 5:
        print(f"{row * col:>4}", end="")
        col += 1
    print()                              # new line after each row
    row += 1
````

````python
# A star triangle
row = 1
while row <= 4:
    star = 1
    while star <= row:
        print("*", end="")
        star += 1
    print()
    row += 1
````

## `for` or `while`? How to choose

| Use **`for`** when… | Use **`while`** when… |
|--------------------|----------------------|
| you have a list to go through | you are waiting for something to happen |
| you know how many times | you do **not** know how many times |
| "do this for **every** item" | "keep doing this **until**…" |

Both can often do the same job. Pick the one that reads more naturally.

## 📋 Summary — Your `while` Loop Cheat Sheet

```python
# ① start
count = 1

# ② condition
while count <= 5:
    print(count)
    count += 1          # ③ update
```

| Idea | Remember |
|------|----------|
| Needs | start · condition · update |
| Runs | while the condition is `True`; may run **zero** times |
| `break` | leave the loop now |
| `continue` | skip to the next round (**update before it!**) |
| `pass` | do nothing |
| `while True` | runs forever; needs a `break` |
| `else` | runs only if there was **no** `break` |
| Infinite loop | forgot the update → press **Ctrl + C** / **■ Interrupt** |
| Safety net | a guard counter or a maximum number of tries |
| `for` vs `while` | `for` = known count or list · `while` = until something happens ||
