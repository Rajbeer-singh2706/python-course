## 📖 The Story — Arjun and the Tatkal Ticket

It is 9:58 AM. Arjun must book a **Tatkal train ticket** to Lucknow for his sister's wedding. Booking opens at 10:00 AM sharp, and the website is jammed.

What does Arjun do? He hits **refresh… refresh… refresh…** until the booking page finally opens.

Ask yourself: *how many times will he refresh?*

He has **no idea**. Maybe 3, maybe 40. He just keeps going **while the page is not open yet**.

A `for` loop needs to know the count (or the collection) in advance. This situation is different — the loop must run **until something changes**. That is exactly what `while` is for.

````python
# Arjun's story in code (the server's replies are scripted so we can run it)
server_replies = ["busy", "busy", "busy", "busy", "busy", "busy", "OPEN"]

refresh_count = 0
page_status = "busy"

while page_status != "OPEN":                 # keep going WHILE the page is not open
    page_status = server_replies[refresh_count]
    refresh_count += 1
    print(f"Refresh #{refresh_count}: server says {page_status}")

print(f"\n🎉 Booking page opened after {refresh_count} refreshes!")
````

Notice: we never told Python "7 times". The **condition** decided when to stop.

## 1️⃣ Anatomy of a `while` loop

```
while  condition :          ← checked BEFORE every round
    <indented body>         ← runs if the condition is True
    <something that moves the condition towards False>
```

Python's thinking:
1. Is the condition `True`? **No** → skip the loop completely.
2. **Yes** → run the body once → go back to step 1.

### 🧠 The three things every `while` loop needs

| # | Ingredient | Arjun's version |
|---|------------|-----------------|
| 1 | **Starting state** (before the loop) | `page_status = "busy"` |
| 2 | **Condition** that can become `False` | `page_status != "OPEN"` |
| 3 | **Update** inside the body | `page_status = server_replies[...]` |

Forget ingredient 3 and you get the most famous beginner bug — an **infinite loop**. We will see (and tame) that later.

### `for` or `while`?

| Use `for` when… | Use `while` when… |
|-----------------|-------------------|
| you have a collection to walk through | you wait for something to happen |
| you know the count in advance | you do **not** know how many rounds |
| "do this for every order" | "keep trying until it works" |

## 2️⃣ Counting and goal-chasing loops

````python
# 🟢 SIMPLE — count up, then count down
count = 1
while count <= 5:
    print("Count:", count)
    count += 1                               # the UPDATE — never forget it

print("-" * 30)
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Lift off! 🚀")
````

````python
# 🟡 REAL-WORLD — how long until my SIP reaches Rs 10 lakh?
# We do not know the number of months in advance → while loop.

monthly_sip = 10000
annual_return = 12                           # % per year (assumed)
goal = 1000000

monthly_rate = annual_return / 12 / 100
corpus = 0
months = 0

while corpus < goal:
    corpus = (corpus + monthly_sip) * (1 + monthly_rate)
    months += 1

years, extra_months = months // 12, months % 12
print(f"Goal of Rs {goal:,} reached in {months} months ({years} years {extra_months} months)")
print(f"You invested Rs {months * monthly_sip:,} → corpus is Rs {corpus:,.0f}")
````

````python
# 🔴 INDUSTRY-READY — the same idea with a SAFETY GUARD
# Production code never trusts a loop to end "by luck". If the inputs are wrong
# (say, the monthly SIP amount is accidentally 0), the goal might never be reached. A maximum
# limit protects the system from running forever.

monthly_sip = 0                              # oops — the SIP amount was left empty!
annual_return = 12
goal = 1000000
MAX_MONTHS = 600                             # 50 years — give up after this

monthly_rate = annual_return / 12 / 100
corpus = 0
months = 0

while corpus < goal and months < MAX_MONTHS:
    corpus = (corpus + monthly_sip) * (1 + monthly_rate)
    months += 1

if corpus >= goal:
    print(f"Goal reached in {months} months")
else:
    print(f"⚠ Goal not reached within {MAX_MONTHS} months — check the inputs")

print(f"Months simulated: {months}, corpus: Rs {corpus:,.0f}")
````

## 3️⃣ Sentinel loops — "keep going until you see the stop signal"

A **sentinel** is a special value that means *"the data has ended"*, like the word `DONE` or an empty line. Data arrives, you process it, and you stop when the sentinel appears.

````python
# 🟢 SIMPLE — a billing counter: scan items until the cashier types DONE
scanned = ["Atta 5kg", "Dal 1kg", "Oil 1L", "DONE", "Rice 10kg"]

index = 0
items_billed = 0

while scanned[index] != "DONE":
    print("Scanned:", scanned[index])
    items_billed += 1
    index += 1

print(f"Total items billed: {items_billed}")
print("(Rice came after DONE, so it was never processed.)")
````

````python
# 🟡 REAL-WORLD — a billing counter with prices and a running total
cart = [("Atta 5kg", 245), ("Dal 1kg", 140), ("Oil 1L", 165), ("Milk", 32), ("STOP", 0)]

i = 0
bill = 0
while cart[i][0] != "STOP":
    item, price = cart[i]
    bill += price
    print(f"  {item:<10} Rs {price:>4}   running total: Rs {bill}")
    i += 1

print(f"\nPay Rs {bill}")
````

````python
# 🔴 INDUSTRY-READY — parsing a config file: read key=value lines until a blank line
# Config and protocol parsers (HTTP headers, .ini files, etc.) stop at a
# blank line. This is how real parsers are built.

lines = [
    "host=db.priyamithai.in",
    "port=5432",
    "debug=true",
    "timeout=30",
    "",                              # blank line = end of the header section
    "ignored=this-is-the-body",
]

config = {}
i = 0
while i < len(lines) and lines[i] != "":         # two stop conditions: end of data OR blank line
    key, value = lines[i].split("=")
    config[key] = value
    i += 1

print("Parsed config:")
for key in config:
    print(f"  {key:<8} → {config[key]}")
print(f"\nStopped at line {i + 1}; {len(lines) - i - 1} line(s) left unread.")
````

## 4️⃣ Validation loops — ask again until the answer is acceptable

Classic use of `while`: *"keep asking until the user gives something valid."*

Since notebooks cannot call `input()`, we feed pre-written answers from a list. In a real script you would replace `entered = attempts[...]` with `entered = input("...")`.

````python
# 🟢 SIMPLE — ask for age until it is a sensible number
typed_answers = ["abc", "-5", "250", "27"]          # simulating what the user types
position = 0

valid_age = None
while valid_age is None:
    entered = typed_answers[position]               # real code: entered = input("Age: ")
    position += 1
    if entered.isdigit() and 1 <= int(entered) <= 120:
        valid_age = int(entered)
        print(f"'{entered}' ✅ accepted")
    else:
        print(f"'{entered}' ❌ not a valid age, try again")

print(f"Age recorded: {valid_age}")
````

````python
# 🟡 REAL-WORLD — OTP verification with 3 attempts
correct_otp = "482916"
typed_otps = ["123456", "482961", "482916"]         # third try is correct

MAX_ATTEMPTS = 3
attempt = 0
verified = False

while attempt < MAX_ATTEMPTS and not verified:
    entered = typed_otps[attempt]
    attempt += 1
    if entered == correct_otp:
        verified = True
        print(f"Attempt {attempt}: ✅ OTP verified")
    else:
        print(f"Attempt {attempt}: ❌ wrong OTP — {MAX_ATTEMPTS - attempt} attempt(s) left")

if not verified:
    print("🔒 Account locked for 30 minutes")
````

````python
# 🔴 INDUSTRY-READY — login with lockout, using while…else
# while…else: the else block runs when the loop ends because the CONDITION
# became False (i.e. NOT through break). Here that means "all attempts used".

correct_password = "Priya@2026"
typed_passwords = ["priya123", "Priya2026", "PRIYA@2026"]    # all wrong
MAX_ATTEMPTS = 3
failed_log = []

attempt = 0
while attempt < MAX_ATTEMPTS:
    entered = typed_passwords[attempt]
    attempt += 1
    if entered == correct_password:
        print("✅ Login successful")
        break
    failed_log.append(f"failed attempt {attempt}")
    print(f"❌ Wrong password ({MAX_ATTEMPTS - attempt} attempt(s) left)")
else:
    print("🔒 Too many failures — account locked")
    print("Security log:", failed_log)
````

## 5️⃣ `while True` + `break` — the "run until told to stop" loop

Menus, servers, games and chat apps all run **forever** until the user quits. They use:

```python
while True:          # condition is always True…
    ...
    if quit_requested:
        break        # …so break is the ONLY way out
```

This is safe **only if** you can point to the `break` that will eventually run.

````python
# 🟢 SIMPLE — a tiny menu that exits on "quit"
menu_choices = ["hello", "hello", "quit", "hello"]
step = 0

while True:
    choice = menu_choices[step]
    step += 1
    if choice == "quit":
        print("Goodbye!")
        break
    print("You chose:", choice)
````

````python
# 🟡 REAL-WORLD — an ATM session (actions are scripted)
balance = 15000
actions = [
    ("balance", 0),
    ("withdraw", 5000),
    ("withdraw", 20000),        # more than the balance
    ("deposit", 2500),
    ("exit", 0),
]

step = 0
print("Welcome to Priya Bank ATM")
while True:
    action, amount = actions[step]
    step += 1

    if action == "exit":
        print("Thank you. Please collect your card.")
        break
    elif action == "balance":
        print(f"Balance: Rs {balance:,}")
    elif action == "withdraw":
        if amount > balance:
            print(f"❌ Withdraw Rs {amount:,} failed — insufficient funds")
        else:
            balance -= amount
            print(f"✅ Withdrew Rs {amount:,} → balance Rs {balance:,}")
    elif action == "deposit":
        balance += amount
        print(f"✅ Deposited Rs {amount:,} → balance Rs {balance:,}")
````

````python
# 🔴 INDUSTRY-READY — an EVENT LOOP (the heart of web servers, games and GUIs)
# The program sits in one big loop, picks up the next event, handles it, and
# keeps going until a "shutdown" event arrives.

incoming_events = [
    "login:anita", "view_product:kaju_katli", "add_to_cart:kaju_katli",
    "login:vikram", "checkout:anita", "bad_event!!", "logout:anita", "shutdown",
    "login:never_processed",
]

handled = {"login": 0, "view_product": 0, "add_to_cart": 0, "checkout": 0, "logout": 0}
unknown = []
position = 0

while True:
    event = incoming_events[position]
    position += 1

    if event == "shutdown":
        print("⏻ Shutdown signal received — server stopping")
        break

    if ":" not in event:                           # malformed event: log it and carry on
        unknown.append(event)
        continue

    kind, detail = event.split(":")
    if kind in handled:
        handled[kind] += 1
        print(f"  handled {kind:<13} → {detail}")
    else:
        unknown.append(event)

print("\nEvent summary:", handled)
print("Unknown events:", unknown)
print("Events left unread:", len(incoming_events) - position)
````

## 6️⃣ Retry with backoff — surviving flaky networks

Networks fail. A payment gateway times out for a second and then works. Good software **retries**, waiting a little longer each time (**exponential backoff**: 1s → 2s → 4s → 8s) so it does not hammer a struggling server.

This is one of the most common `while` patterns in industry. (We skip real waiting and only print the delay, so the notebook runs instantly.)

````python
# 🟢 SIMPLE — keep trying until success
results = ["fail", "fail", "success"]
tries = 0
status = "fail"

while status != "success":
    status = results[tries]
    tries += 1
    print(f"Try {tries}: {status}")

print(f"Succeeded after {tries} tries")
````

````python
# 🟡 REAL-WORLD — retry with a maximum and a growing delay
api_responses = ["timeout", "503 busy", "timeout", "200 OK"]       # scripted replies

MAX_RETRIES = 5
attempt = 0
delay_seconds = 1

while attempt < MAX_RETRIES:
    response = api_responses[attempt]
    attempt += 1
    if response == "200 OK":
        print(f"Attempt {attempt}: ✅ {response}")
        break
    print(f"Attempt {attempt}: ❌ {response} → waiting {delay_seconds}s before retrying")
    delay_seconds *= 2                              # 1, 2, 4, 8 …
else:
    print("Gave up — all retries used")
````

````python
# 🔴 INDUSTRY-READY — the same loop when the service is truly DOWN
# The else clause is the "alert the engineers" path.

api_responses = ["503 busy"] * 5                    # every reply fails
MAX_RETRIES = 5
attempt = 0
delay_seconds = 1
total_wait = 0
MAX_DELAY = 8                                       # never wait longer than 8s

while attempt < MAX_RETRIES:
    response = api_responses[attempt]
    attempt += 1
    if response == "200 OK":
        print("✅ success")
        break
    print(f"Attempt {attempt}: ❌ {response}  (wait {delay_seconds}s)")
    total_wait += delay_seconds
    delay_seconds = min(delay_seconds * 2, MAX_DELAY)   # exponential, but capped
else:
    print(f"🚨 ALERT: payment service down after {MAX_RETRIES} attempts, {total_wait}s of waiting")
````

## 7️⃣ Polling — "are we there yet?"

When you start a long job (video export, report generation, bank transfer), you cannot get the answer instantly. You **poll**: check the status every few seconds until it is `COMPLETED` — or until a **timeout** tells you to stop waiting.

````python
# 🟢 SIMPLE — check until the pizza is ready
pizza_status = ["ordered", "baking", "baking", "out for delivery", "delivered"]
check = 0
while pizza_status[check] != "delivered":
    print(f"Check {check + 1}: {pizza_status[check]}")
    check += 1
print("🍕 Pizza delivered!")
````

````python
# 🟡 REAL-WORLD + 🔴 INDUSTRY-READY — polling a report job with a timeout
# Two exits: the job finishes (break) OR we run out of patience (else).

job_states = ["QUEUED", "RUNNING", "RUNNING", "RUNNING", "COMPLETED"]
POLL_INTERVAL = 5            # seconds between checks
TIMEOUT = 60                 # give up after a minute

elapsed = 0
poll = 0

while elapsed <= TIMEOUT:
    state = job_states[min(poll, len(job_states) - 1)]
    print(f"t={elapsed:>2}s  poll #{poll + 1}: {state}")
    if state in ("COMPLETED", "FAILED"):
        print(f"→ Job finished with status {state}")
        break
    poll += 1
    elapsed += POLL_INTERVAL
else:
    print("⏰ Timed out — job still not finished")
````

````python
# Same loop, but this job is stuck forever → the timeout rescues us
job_states = ["QUEUED", "RUNNING"]                  # never completes
POLL_INTERVAL = 20
TIMEOUT = 60

elapsed = 0
poll = 0
while elapsed <= TIMEOUT:
    state = job_states[min(poll, len(job_states) - 1)]
    print(f"t={elapsed:>2}s  poll #{poll + 1}: {state}")
    if state in ("COMPLETED", "FAILED"):
        break
    poll += 1
    elapsed += POLL_INTERVAL
else:
    print("⏰ Timed out — job still not finished, raising an alert")
````

## 8️⃣ Queues — `while tasks:` keeps working until the pile is empty

An empty list is treated as `False`, a non-empty list as `True`. So **`while my_list:`** means *"while there is still work left"*.

This is the engine behind print queues, background job workers (Celery, RQ), and customer-support ticket systems.

````python
# 🟢 SIMPLE — serve customers in a bank queue, first come first served
queue = ["Anita", "Vikram", "Sneha"]

while queue:                                  # True while the list is not empty
    customer = queue.pop(0)                   # take the FIRST person out of the line
    print(f"Serving {customer}. People waiting: {len(queue)}")

print("Queue empty — counter closed.")
````

````python
# 🟡 REAL-WORLD — a worker that re-queues failed tasks
task_queue = ["resize_image_1", "send_email_2", "generate_report_3"]
fails_remaining = {"send_email_2": 2}              # this task fails twice, then works

while task_queue:
    task = task_queue.pop(0)
    if fails_remaining.get(task, 0) > 0:
        fails_remaining[task] -= 1
        print(f"❌ {task} failed — put back at the end of the queue")
        task_queue.append(task)
    else:
        print(f"✅ {task} completed")

print("All tasks done.")
````

````python
# 🔴 INDUSTRY-READY — a worker where tasks can CREATE more tasks
# A web crawler starts with one page, discovers links, and adds them to the
# queue. A `for` loop cannot do this (you must not grow a list while looping
# over it) — `while queue:` handles it naturally.

site_links = {
    "home":      ["products", "about"],
    "products":  ["kaju_katli", "laddu", "home"],     # "home" links back (a cycle!)
    "about":     ["contact"],
    "kaju_katli": [],
    "laddu":     [],
    "contact":   ["home"],
}

to_visit = ["home"]
visited = []

while to_visit:
    page = to_visit.pop(0)
    if page in visited:                               # avoid going round in circles forever
        continue
    visited.append(page)
    print(f"Crawled {page:<11} found {len(site_links[page])} link(s)")
    for link in site_links[page]:
        if link not in visited:
            to_visit.append(link)

print(f"\nSite map ({len(visited)} pages): {visited}")
````

## 9️⃣ Pagination — fetching data page by page

APIs never return a million records at once. They return **one page** plus a **cursor** pointing to the next. You loop until there is no next page. You will meet this in Day 23 (APIs).

````python
# 🔴 INDUSTRY-READY — pull ALL orders from a paginated API
# (the API is simulated with a dict: cursor → (items, next_cursor))

api_pages = {
    None: (["Order 1", "Order 2", "Order 3"], "page2"),
    "page2": (["Order 4", "Order 5", "Order 6"], "page3"),
    "page3": (["Order 7"], None),                      # None = no more pages
}

cursor = None
all_orders = []
page_number = 0

while True:                                            # we must fetch at least once
    items, next_cursor = api_pages[cursor]
    page_number += 1
    all_orders.extend(items)
    print(f"Fetched page {page_number}: {len(items)} item(s) → next cursor: {next_cursor}")
    if next_cursor is None:
        break
    cursor = next_cursor

print(f"\nTotal orders downloaded: {len(all_orders)}")
print(all_orders)
````

## 🔟 Number-crunching `while` loops (algorithms)

When you keep **shrinking** a problem until nothing is left — digits of a number, a search range — `while` is the natural tool.

````python
# 🟢 SIMPLE — sum the digits of a number using % and //
number = 94857
original = number
digit_sum = 0

while number > 0:
    last_digit = number % 10          # peel off the last digit
    digit_sum += last_digit
    number //= 10                     # drop the last digit

print(f"Digit sum of {original} = {digit_sum}")
````

````python
# 🟡 REAL-WORLD — GCD (Euclid's algorithm) to simplify a ratio
# Used to reduce fractions like 252/105 to lowest terms.
a, b = 252, 105
x, y = a, b
while y != 0:
    x, y = y, x % y
print(f"GCD({a}, {b}) = {x}")
print(f"{a}/{b} simplified = {a // x}/{b // x}")
````

````python
# 🔴 INDUSTRY-READY — BINARY SEARCH
# Searching 334 sorted product IDs by checking one by one could take 334
# steps. Binary search halves the range every round — about 9 steps at most.
# Search engines and databases are built on this idea.

sorted_ids = list(range(1000, 2000, 3))       # 1000, 1003, 1006, … 1999
target = 1990

low = 0
high = len(sorted_ids) - 1
steps = 0
found_at = -1

while low <= high:
    mid = (low + high) // 2
    steps += 1
    if sorted_ids[mid] == target:
        found_at = mid
        break
    elif sorted_ids[mid] < target:
        low = mid + 1                         # target is in the right half
    else:
        high = mid - 1                        # target is in the left half

print(f"List size : {len(sorted_ids)}")
print(f"Found {target} at index {found_at} in just {steps} steps")
````

## ⚠️ Pitfalls — the infinite loop and how to avoid it

An **infinite loop** happens when the condition can never become `False`.

In a real script press **Ctrl + C** to stop it. In Jupyter use the ■ **Interrupt** button.

| Cause | Example | Fix |
|-------|---------|-----|
| Forgot the update | `while i < 5:` but never `i += 1` | add the update |
| Update goes the wrong way | `while i < 5:` with `i -= 1` | check direction |
| Condition can be skipped over | `while x != 0:` with `x -= 3` starting at 10 | use `>` / `<` instead of `!=` |
| Misspelled variable | updating `totl` instead of `total` | read carefully |

````python
# ❌ This loop would run forever (kept commented out on purpose):
# i = 0
# while i < 5:
#     print(i)            # no i += 1  → condition never changes

# ❌ Dangerous: x goes 10, 7, 4, 1, -2, -5 … it jumps OVER 0 and never equals it
# ✅ Two defences — a safer comparison, and a safety valve:

x = 10
iterations = 0
MAX_ITERATIONS = 100

while x > 0:                                  # defence 1: use > instead of !=
    x -= 3
    iterations += 1
    if iterations >= MAX_ITERATIONS:          # defence 2: a safety valve
        print("Safety valve triggered!")
        break

print(f"Loop ended with x = {x} after {iterations} iterations")
````

````python
# Another classic: off-by-one with indexes
items = ["laddu", "barfi", "jalebi"]

i = 0
while i < len(items):          # ✅ correct: stops when i == 3
    print(i, items[i])
    i += 1

# ❌ while i <= len(items):  would try items[3] → IndexError: list index out of range
````

## 🚀 Mini-Project — Order Fulfilment Worker

Priya's shop now processes payments through a gateway that sometimes fails. Build a worker that:

1. Takes orders from a **queue** (`while order_queue:`)
2. For each order, **retries payment up to 3 times** (an inner `while`)
3. Orders that still fail go to a **dead-letter list** for a human to review
4. Prints a summary

This combines four patterns from this notebook: **queue worker + retry loop + while…else + safety limit**.

To keep it deterministic, the dict says how many times each order's payment will fail before it works.

````python
order_queue = [1001, 1002, 1003, 1004]
failures_before_success = {1001: 0, 1002: 2, 1003: 5, 1004: 1}   # 1003 will never succeed in 3 tries

MAX_ATTEMPTS = 3
completed = []
dead_letter = []

print("🚚 Fulfilment worker started\n")

while order_queue:
    order_id = order_queue.pop(0)
    print(f"Order #{order_id}")

    attempt = 0
    while attempt < MAX_ATTEMPTS:
        attempt += 1
        if attempt > failures_before_success[order_id]:
            print(f"   attempt {attempt}: ✅ payment successful")
            completed.append((order_id, attempt))
            break
        print(f"   attempt {attempt}: ❌ payment failed")
    else:
        print(f"   🚨 all {MAX_ATTEMPTS} attempts failed → moved to dead-letter list")
        dead_letter.append(order_id)

print("\n" + "=" * 45)
print("SUMMARY")
print("=" * 45)
for order_id, attempts_used in completed:
    note = "first try" if attempts_used == 1 else f"after {attempts_used} attempts"
    print(f"  #{order_id}  completed ({note})")
for order_id in dead_letter:
    print(f"  #{order_id}  FAILED — needs manual review")
print(f"\nCompleted: {len(completed)}   Dead-letter: {len(dead_letter)}")
````

### 🔴 Exercise 6 — Bank queue waiting time
Each customer needs the given number of minutes at the counter. Customers are served one after another. For every customer print how long they **waited** before being served, then the **average wait**.

Expected:
```
Anita waited 0 min
Vikram waited 5 min

```

````python
service_times = [("Anita", 5), ("Vikram", 3), ("Sneha", 12), ("Rohit", 4)]
# YOUR CODE HERE
````

````python
# SOLUTION
service_times = [("Anita", 5), ("Vikram", 3), ("Sneha", 12), ("Rohit", 4)]
queue = service_times[:]                  # copy so the original list is untouched
clock = 0
total_wait = 0
served = 0
while queue:
    name, minutes = queue.pop(0)
    print(f"{name} waited {clock} min")
    total_wait += clock
    clock += minutes
    served += 1
print(f"Average wait: {total_wait / served:.1f} min")
````

### 🔴 Exercise 7 — Retry until the app works
A server restarts and answers `"down"` several times before saying `"up"`. Retry with delays 1, 2, 4, 8… seconds, up to **6 attempts**. Print each attempt and the delay you would wait. If it comes up, print the attempt number; if not, print `Alert!`.

Expected:
```
Attempt 1: down — wait 1s
Attempt 2: down — wait 2s
Attempt 3: down — wait 4s

```

````python
replies = ["down", "down", "down", "up", "up", "up"]
MAX_ATTEMPTS = 6
# YOUR CODE HERE
````

````python
# SOLUTION
replies = ["down", "down", "down", "up", "up", "up"]
MAX_ATTEMPTS = 6
attempt = 0
delay = 1
while attempt < MAX_ATTEMPTS:
    reply = replies[attempt]
    attempt += 1
    if reply == "up":
        print(f"Attempt {attempt}: up ✅")
        break
    print(f"Attempt {attempt}: down — wait {delay}s")
    delay *= 2
else:
    print("Alert!")
````

### 🔴 Exercise 8 — Guess the number (binary search bot)
The secret number is **77** (between 1 and 100). Write a bot that always guesses the middle of the remaining range and is told "too low" or "too high". Print each guess and the total number of guesses.

Expected:
```
Guess 1: 50 → too low
Guess 2: 75 → too low
Guess 3: 88 → too high
Guess 4: 81 → too high
Guess 5: 78 → too high
Guess 6: 76 → too low
Guess 7: 77 → correct!
Solved in 7 guesses
```

````python
secret = 77
low, high = 1, 100
# YOUR CODE HERE
````

````python
# SOLUTION
secret = 77
low, high = 1, 100
guesses = 0
while True:
    guess = (low + high) // 2
    guesses += 1
    if guess == secret:
        print(f"Guess {guesses}: {guess} → correct!")
        break
    elif guess < secret:
        print(f"Guess {guesses}: {guess} → too low")
        low = guess + 1
    else:
        print(f"Guess {guesses}: {guess} → too high")
        high = guess - 1
print(f"Solved in {guesses} guesses")
````

## 📋 Summary

| Concept | Remember |
|---------|----------|
| `while condition:` | checked before every round; may run zero times |
| 3 ingredients | starting state · condition · update |
| Sentinel loop | stop when a special value (`DONE`, blank line) appears |
| Validation loop | repeat until the answer is acceptable |
| `while True` + `break` | menus, event loops, servers — needs a reachable `break` |
| `while …: else:` | `else` runs if the loop ended **without** `break` |
| `while queue:` | keep working until the pile is empty |
| Safety guard | `MAX_ATTEMPTS`, `TIMEOUT`, `MAX_ITERATIONS` protect against endless loops |

**Industry patterns you have now seen:** SIP goal calculator · config parser · OTP and login lockout · event loop · retry with exponential backoff · job polling with timeout · task queue worker · web crawler · cursor pagination · binary search · fulfilment worker.

### Choosing between the two loops

| Question | Answer → loop |
|----------|---------------|
| "Do something for **each** item" | `for` |
| "Do it **N times**" | `for … in range(N)` |
| "Repeat **until** something happens" | `while` |
| "Keep **retrying / polling / waiting**" | `while` |
| "Process a pile that **can grow**" | `while queue:` |
