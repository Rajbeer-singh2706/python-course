# 🚀 Mini-Project — "Priya Bank" Mini ATM

Time to combine **everything** you learned into one program.

### What the ATM does

| Part | What happens | `while` skills used |
|------|--------------|---------------------|
| **1. Login** | The user has **3 tries** to enter the correct PIN. Too many wrong tries → card blocked. | counting loop, `and`, `break`, `while…else` |
| **2. Menu** | The user keeps choosing: check balance, withdraw, deposit, or exit. | `while True`, `break` |
| **3. Checking** | Bad amounts (letters, zero, too much money) are rejected. | validation, `continue` |
| **4. Receipt** | At the end, all transactions are printed one by one. | `while` over a list |

### Build it in 4 steps
We build the program one part at a time, then show the full version.

### Step 1 — Login with 3 attempts

```python
correct_pin = "4321"
typed_pins = ["1111", "2222", "4321"]     # the user's three tries

attempt = 0
logged_in = False

while attempt < 3:
    pin = typed_pins[attempt]
    attempt += 1
    if pin == correct_pin:
        logged_in = True
        print("✅ PIN correct. Welcome to Priya Bank!")
        break                              # stop asking, the PIN is right
    print(f"❌ Wrong PIN. Attempts left: {3 - attempt}")
else:
    print("🔒 Too many wrong attempts. Card blocked!")

print("logged_in =", logged_in)
```

**What if all 3 tries are wrong?** Here is the "card blocked" path:

```python
correct_pin = "4321"
typed_pins = ["0000", "1234", "9999"]      # all wrong this time

attempt = 0
logged_in = False
while attempt < 3:
    pin = typed_pins[attempt]
    attempt += 1
    if pin == correct_pin:
        logged_in = True
        print("✅ PIN correct.")
        break
    print(f"❌ Wrong PIN. Attempts left: {3 - attempt}")
else:
    print("🔒 Too many wrong attempts. Card blocked!")
```

### Step 2 — The full ATM

Read it slowly. Every `while` idea from this notebook appears somewhere.

```python
# ═══════════════════════ PRIYA BANK MINI ATM ═══════════════════════

balance = 15000
correct_pin = "4321"
typed_pins = ["1111", "2222", "4321"]                 # try changing these!

# What the user does at the menu: (choice, amount typed)
menu_inputs = [
    ("balance",  ""),
    ("withdraw", "2000"),
    ("withdraw", "abc"),        # not a number
    ("withdraw", "50000"),      # more than the balance
    ("deposit",  "1500"),
    ("withdraw", "0"),          # zero is not allowed
    ("dance",    ""),           # not a menu option
    ("exit",     ""),
]

history = []                    # a list where we remember every successful transaction

# ───────── PART 1: LOGIN (3 attempts) ─────────
attempt = 0
logged_in = False
while attempt < 3:
    pin = typed_pins[attempt]
    attempt += 1
    if pin == correct_pin:
        logged_in = True
        print("✅ PIN correct. Welcome to Priya Bank!")
        break
    print(f"❌ Wrong PIN. Attempts left: {3 - attempt}")
else:
    print("🔒 Too many wrong attempts. Card blocked!")

# ───────── PART 2: MENU (keeps running until "exit") ─────────
if logged_in:
    position = 0
    while True:
        choice, amount_text = menu_inputs[position]
        position += 1
        print(f"\n> {choice} {amount_text}")

        if choice == "exit":
            print("Thank you for banking with us. Goodbye! 👋")
            break

        if choice == "balance":
            print(f"Your balance is Rs {balance:,}")
            continue                                   # back to the menu

        if choice == "withdraw" or choice == "deposit":
            # ───── PART 3: CHECK THE AMOUNT ─────
            if not amount_text.isdigit():
                print("❌ Amount must be a number.")
                continue
            amount = int(amount_text)
            if amount <= 0:
                print("❌ Amount must be more than 0.")
                continue

            if choice == "withdraw":
                if amount > balance:
                    print(f"❌ Not enough money. You have only Rs {balance:,}.")
                    continue
                balance -= amount
                history.append(f"Withdrew Rs {amount:,}")
                print(f"✅ Please collect Rs {amount:,}. New balance: Rs {balance:,}")
            else:
                balance += amount
                history.append(f"Deposited Rs {amount:,}")
                print(f"✅ Rs {amount:,} deposited. New balance: Rs {balance:,}")
        else:
            print("❌ Unknown option. Choose balance, withdraw, deposit or exit.")

    # ───────── PART 4: RECEIPT (a while loop over the history list) ─────────
    print("\n" + "=" * 32)
    print("   TRANSACTION RECEIPT")
    print("=" * 32)
    i = 0
    while i < len(history):
        print(f"{i + 1}. {history[i]}")
        i += 1
    print("-" * 32)
    print(f"Closing balance: Rs {balance:,}")
```

### 🎯 Where did each idea appear?

| Idea | Where in the ATM |
|------|------------------|
| Start / condition / update | `attempt`, `position`, `i` |
| `and`, `or`, `not` | `choice == "withdraw" or choice == "deposit"`, `not amount_text.isdigit()` |
| `while …: else:` | "Card blocked" message |
| `break` | correct PIN → stop asking; `exit` → leave the menu |
| `continue` | every rejected input goes back to the menu |
| `while True` | the menu loop |
| Validation | number, more than 0, enough balance |
| Loop over a list with an index | the receipt |
