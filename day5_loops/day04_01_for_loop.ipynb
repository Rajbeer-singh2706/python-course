## 📖 The Story — Priya's Diwali Problem

Priya runs **Priya Mithai Bhandar**, a small sweet shop in Meerut. Diwali is three days away and she wants to send a personal greeting to her loyal customers.

She opens Python and starts typing…

````python
# The "copy-paste" way — one print per customer
print("Happy Diwali, Anita! 🪔 Get 10% off on Kaju Katli.")
print("Happy Diwali, Vikram! 🪔 Get 10% off on Kaju Katli.")
print("Happy Diwali, Sneha! 🪔 Get 10% off on Kaju Katli.")
print("Happy Diwali, Rohit! 🪔 Get 10% off on Kaju Katli.")
print("Happy Diwali, Meena! 🪔 Get 10% off on Kaju Katli.")
````

That took 5 lines for 5 customers. But Priya actually has **500 customers**, and tomorrow she wants to change "10% off" to "15% off".

With copy-paste that means **500 lines to write and 500 lines to edit**. One typo and a customer gets the wrong offer.

A programmer's rule: *if you are repeating yourself, there should be a loop.* Here is the same job using a `for` loop:

````python
customers = ["Anita", "Vikram", "Sneha", "Rohit", "Meena"]
offer = "15% off on Kaju Katli"          # change ONE place, every message updates

for name in customers:
    print(f"Happy Diwali, {name}! 🪔 Get {offer}.")
````

Now watch what happens when the list grows to **500 customers**. The loop code does not change by even one character:

````python
customers = []
for n in range(1, 501):                  # build a pretend list of 500 customers
    customers.append(f"Customer {n}")

sent = 0
for name in customers:
    message = f"Happy Diwali, {name}! 🪔 Get {offer}."
    sent += 1

print(f"Messages sent: {sent}")
print("Last message :", message)
````

**That is the whole reason loops exist:** write the instruction *once*, let Python repeat it as many times as needed — 5 times or 5 million times.

## 1️⃣ Anatomy of a `for` loop



```
for  item  in  collection :
     ───┬──     ────┬────
        │           └── anything Python can walk through one-by-one
        └── a variable that holds the CURRENT item
    <indented body>     ← runs once for every item
```

* Python picks the **first** item, puts it in `item`, runs the body.
* Then it picks the **next** item, runs the body again.
* When the collection is **finished**, the loop ends by itself — no counter, no manual stop.
* The body is whatever is **indented** (4 spaces). No indent → not in the loop.

````python
# 🟢 SIMPLE — see the "current item" change on every round
for sweet in ["Laddu", "Jalebi", "Barfi"]:
    print("Packing:", sweet)

print("Loop finished — this line is NOT indented, so it runs only once.")
````

## 2️⃣ What can a `for` loop walk through?

Lists, strings, tuples, dictionaries and sets are all **iterable** — a `for` loop can visit what is inside them.

| Collection | `for x in ...` gives you |
|------------|--------------------------|
| list / tuple | each element, in order |
| string | each character |
| dict | each **key** (use `.items()` for key + value) |
| set | each element (order not guaranteed) |

````python
# 🟢 SIMPLE — one of each
cities = ["Delhi", "Mumbai", "Pune"]
for city in cities:
    print("City     :", city)

print("-" * 40)
for letter in "DIWALI":
    print("Letter   :", letter)

print("-" * 40)
marks = {"Priya": 92, "Arjun": 85, "Riya": 78}
for student in marks:                        # keys only
    print("Student  :", student)

print("-" * 40)
for student, score in marks.items():         # key AND value, unpacked
    print(f"{student} scored {score}")
````

````python
# 🟡 REAL-WORLD — a GST invoice for the sweet shop
items = [
    {"name": "Kaju Katli (1 kg)", "price": 800, "qty": 2, "gst": 5},
    {"name": "Soan Papdi (500 g)", "price": 220, "qty": 3, "gst": 5},
    {"name": "Gift Box",          "price": 150, "qty": 1, "gst": 12},
]

subtotal = 0
total_gst = 0
print(f"{'Item':<22}{'Qty':>4}{'Amount':>10}{'GST':>8}")
print("-" * 44)

for item in items:
    amount = item["price"] * item["qty"]
    gst_amount = amount * item["gst"] / 100
    subtotal += amount
    total_gst += gst_amount
    print(f"{item['name']:<22}{item['qty']:>4}{amount:>10,.2f}{gst_amount:>8,.2f}")

print("-" * 44)
print(f"{'Subtotal':<26}{subtotal:>10,.2f}")
print(f"{'Total GST':<26}{total_gst:>10,.2f}")
print(f"{'GRAND TOTAL':<26}{subtotal + total_gst:>10,.2f}")
````

````python
# 🔴 INDUSTRY-READY — analysing server logs (counting events by level)
# Real systems produce thousands of log lines per minute. Engineers loop
# over them to find out "how many errors happened?" in seconds.

log_lines = [
    "2026-11-10 09:15:02 INFO  Payment received order=1001",
    "2026-11-10 09:15:09 INFO  Order packed order=1001",
    "2026-11-10 09:16:41 ERROR Payment gateway timeout order=1002",
    "2026-11-10 09:17:03 WARN  Low stock item=KajuKatli",
    "2026-11-10 09:17:30 INFO  Payment received order=1003",
    "2026-11-10 09:18:12 ERROR Address not found order=1004",
    "2026-11-10 09:19:55 INFO  Order packed order=1003",
]

level_counts = {}
error_lines = []

for line in log_lines:
    parts = line.split()           # ['2026-11-10', '09:15:02', 'INFO', ...]
    level = parts[2]
    level_counts[level] = level_counts.get(level, 0) + 1
    if level == "ERROR":
        error_lines.append(line)

print("Log summary:")
for level, count in level_counts.items():
    print(f"  {level:<6} {count}")

print("\nLines needing attention:")
for line in error_lines:
    print("  ⚠", line)
````

## 3️⃣ `range()` — loops by *count*



Sometimes you do not have a collection — you just want to do something **N times**, or walk through numbers. `range()` produces those numbers.

| Form | Produces |
|------|----------|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 8)` | 2, 3, 4, 5, 6, 7 |
| `range(0, 20, 5)` | 0, 5, 10, 15 |
| `range(10, 0, -1)` | 10, 9, 8, … 1 |

> ⚠️ The **stop** value is never included. `range(1, 5)` stops at 4.

````python
# 🟢 SIMPLE — the four forms
print("range(5)         :", list(range(5)))
print("range(2, 8)      :", list(range(2, 8)))
print("range(0, 20, 5)  :", list(range(0, 20, 5)))
print("range(10, 0, -1) :", list(range(10, 0, -1)))

print("-" * 40)
print("Table of 7:")
for i in range(1, 11):
    print(f"7 x {i:>2} = {7 * i:>3}")
````

````python
# 🟡 REAL-WORLD — a home-loan EMI schedule (Rs 3,00,000 at 9% for 12 months)
principal = 300000
annual_rate = 9
months = 12

monthly_rate = annual_rate / 12 / 100
emi = principal * monthly_rate * (1 + monthly_rate) ** months / ((1 + monthly_rate) ** months - 1)

balance = principal
print(f"Monthly EMI: Rs {emi:,.2f}\n")
print(f"{'Month':>5}{'Interest':>12}{'Principal':>12}{'Balance':>14}")

for month in range(1, months + 1):
    interest = balance * monthly_rate
    principal_paid = emi - interest
    balance = max(balance - principal_paid, 0)
    print(f"{month:>5}{interest:>12,.2f}{principal_paid:>12,.2f}{balance:>14,.2f}")
````

````python
# 🔴 INDUSTRY-READY — processing records in BATCHES
# Databases and APIs reject huge requests. Production code sends data in
# chunks (e.g. 100 rows at a time). range(start, stop, STEP) does this neatly.

records = list(range(1, 1051))        # pretend we have 1,050 customer records
BATCH_SIZE = 100

total_batches = 0
for start in range(0, len(records), BATCH_SIZE):
    batch = records[start:start + BATCH_SIZE]       # slice out one chunk
    total_batches += 1
    print(f"Batch {total_batches:>2}: records {batch[0]:>4} to {batch[-1]:>4}  ({len(batch)} rows)")

print(f"\nDone. {len(records)} records sent in {total_batches} batches.")
````

## 4️⃣ `enumerate()` and `zip()` — when you need position or pairs

* `enumerate(items)` gives **(position, item)** — no manual counter needed.
* `zip(a, b)` walks **two lists together**, pairing item 1 with item 1, item 2 with item 2…

Beginners often write `for i in range(len(items))`. Prefer `enumerate` — it is shorter and harder to get wrong.

````python
# 🟢 SIMPLE — a leaderboard with ranks
toppers = ["Priya", "Arjun", "Riya"]

for rank, name in enumerate(toppers, start=1):      # start=1 → human numbering
    print(f"Rank {rank}: {name}")

print("-" * 40)
students = ["Priya", "Arjun", "Riya"]
scores = [92, 85, 78]
for student, score in zip(students, scores):
    print(f"{student} scored {score}")
````

````python
# 🟡 REAL-WORLD — report card with grade
names = ["Priya", "Arjun", "Riya", "Kabir"]
maths = [92, 58, 74, 39]
science = [88, 64, 81, 45]

for name, m, s in zip(names, maths, science):
    average = (m + s) / 2
    result = "Pass" if m >= 40 and s >= 40 else "Fail"
    print(f"{name:<7} Maths {m:>3}  Science {s:>3}  Avg {average:>5.1f}  → {result}")
````

````python
# 🔴 INDUSTRY-READY — validating an uploaded CSV and reporting the LINE NUMBER
# of every bad row. enumerate() is how real data-validation code points users
# to exactly where the problem is.

csv_rows = [
    "order_id,customer,amount",      # line 1 = header
    "1001,Anita,2500",
    "1002,,1800",                    # missing customer
    "1003,Vikram,abc",               # amount is not a number
    "1004,Sneha,-50",                # negative amount
    "1005,Rohit,4200",
]

errors = []
valid_rows = 0

for line_no, row in enumerate(csv_rows, start=1):
    if line_no == 1:                 # skip the header
        continue

    order_id, customer, amount = row.split(",")

    if not customer:
        errors.append(f"Line {line_no}: customer name is missing")
        continue
    if not amount.lstrip("-").isdigit():
        errors.append(f"Line {line_no}: amount '{amount}' is not a number")
        continue
    if int(amount) < 0:
        errors.append(f"Line {line_no}: amount cannot be negative")
        continue

    valid_rows += 1

print(f"Valid rows : {valid_rows}")
print(f"Bad rows   : {len(errors)}")
for message in errors:
    print("  ✗", message)
````

## 5️⃣ Accumulator patterns — the loops you will write 1,000 times


Almost every real loop is one of these four shapes:

| Pattern | Idea | Starts with |
|---------|------|-------------|
| **Total** | add each value | `total = 0` |
| **Count** | count items that match | `count = 0` |
| **Max / Min** | remember the best so far | `best = first item` |
| **Collect / Group** | build a new list or dict | `result = []` or `{}` |

The trick: **create the accumulator BEFORE the loop, update it INSIDE the loop, use it AFTER the loop.**

````python
# 🟢 SIMPLE — total, count, max in one pass
daily_sales = [12000, 8500, 15000, 9200, 21000, 18000, 7000]

total = 0
busy_days = 0                         # days with sales above 10,000
best_sale = daily_sales[0]

for sale in daily_sales:
    total += sale
    if sale > 10000:
        busy_days += 1
    if sale > best_sale:
        best_sale = sale

print(f"Weekly total : Rs {total:,}")
print(f"Daily average: Rs {total / len(daily_sales):,.0f}")
print(f"Busy days    : {busy_days}")
print(f"Best day     : Rs {best_sale:,}")
````

````python
# 🟡 REAL-WORLD — finding the best month and which months beat the target
months = ["Jul", "Aug", "Sep", "Oct", "Nov"]
revenue = [310000, 275000, 340000, 520000, 480000]     # Diwali in Oct/Nov!
TARGET = 400000

best_month = months[0]
best_revenue = revenue[0]
months_above_target = []

for month, amount in zip(months, revenue):
    if amount > best_revenue:
        best_revenue = amount
        best_month = month
    if amount >= TARGET:
        months_above_target.append(month)

print(f"Best month        : {best_month} (Rs {best_revenue:,})")
print(f"Months above Rs {TARGET:,}: {months_above_target}")
````

````python
# 🔴 INDUSTRY-READY — GROUP BY (the loop behind every dashboard)
# "Revenue per city" is the first report any e-commerce team builds.
# This dict-accumulator pattern is exactly what SQL's GROUP BY does.

transactions = [
    ("Delhi", 1200), ("Mumbai", 3400), ("Delhi", 800), ("Pune", 1500),
    ("Mumbai", 2100), ("Delhi", 2500), ("Pune", 900), ("Mumbai", 1750),
]

revenue_by_city = {}
orders_by_city = {}

for city, amount in transactions:
    revenue_by_city[city] = revenue_by_city.get(city, 0) + amount
    orders_by_city[city] = orders_by_city.get(city, 0) + 1

top_city = None
top_revenue = 0

print(f"{'City':<8}{'Orders':>8}{'Revenue':>10}{'Avg order':>11}")
for city in revenue_by_city:
    revenue = revenue_by_city[city]
    orders = orders_by_city[city]
    print(f"{city:<8}{orders:>8}{revenue:>10,}{revenue / orders:>11,.0f}")
    if revenue > top_revenue:
        top_revenue = revenue
        top_city = city

print(f"\nTop city: {top_city} with Rs {top_revenue:,}")
````

## 6️⃣ Controlling the loop — `break`, `continue`, `pass`, `else`



| Keyword | What it does | Think of it as |
|---------|--------------|----------------|
| `break` | leave the loop **right now** | "Found it, stop searching." |
| `continue` | skip the rest of **this round**, go to the next item | "Not this one, next please." |
| `pass` | do nothing (a placeholder) | "I'll fill this in later." |
| `else` | runs only if the loop finished **without** `break` | "Searched everything — not found." |

````python
# 🟢 SIMPLE — break and continue side by side
numbers = [4, 9, -2, 15, 7, -8, 30]

print("continue → print only positive numbers:")
for n in numbers:
    if n < 0:
        continue
    print("  ", n)

print("\nbreak → stop at the first number above 10:")
for n in numbers:
    if n > 10:
        print("   Found", n, "— stopping")
        break
    print("   checked", n)
````

````python
# 🟡 REAL-WORLD — a customer search using for…else
registered = ["anita@mail.com", "vikram@mail.com", "sneha@mail.com"]

for email_to_find in ["sneha@mail.com", "karan@mail.com"]:
    for email in registered:
        if email == email_to_find:
            print(f"✅ {email_to_find} is already registered")
            break
    else:
        # runs ONLY when the inner loop ended without hitting break
        print(f"🆕 {email_to_find} is new — creating account")
````

````python
# 🔴 INDUSTRY-READY — assigning a delivery partner (first match wins)
# Real dispatch systems scan partners and stop at the first suitable one.
# for…else handles the "nobody available" case without any flag variable.

partners = [
    {"name": "Ramesh", "available": False, "distance_km": 1.2},
    {"name": "Sunita", "available": True,  "distance_km": 7.5},   # too far
    {"name": "Imran",  "available": True,  "distance_km": 3.1},   # ← first good one
    {"name": "Kavita", "available": True,  "distance_km": 0.8},
]
MAX_DISTANCE_KM = 5

for partner in partners:
    if not partner["available"]:
        continue                                   # busy → skip
    if partner["distance_km"] > MAX_DISTANCE_KM:
        continue                                   # too far → skip
    print(f"Order assigned to {partner['name']} ({partner['distance_km']} km away)")
    break
else:
    print("No partner free nearby — order added to waiting queue")
````

## 7️⃣ Nested loops — a loop inside a loop

The **inner** loop runs completely for **every single round** of the outer loop.
Think of a clock: the minute hand (inner) goes all the way round for each step of the hour hand (outer).

If the outer loop runs 3 times and the inner runs 4 times, the inner body runs **3 × 4 = 12** times.

> ⚠️ `break` inside the inner loop only stops the **inner** loop.

````python
# 🟢 SIMPLE — multiplication grid and a star triangle
for row in range(1, 5):
    for col in range(1, 6):
        print(f"{row * col:>4}", end="")      # end="" keeps the row on one line
    print()                                    # new line after each row

print("-" * 30)
for row in range(1, 6):
    print("*" * row)
````

````python
# 🟡 REAL-WORLD — a cinema seating chart
rows = "ABCDE"
seats_per_row = 6
booked = {"A3", "B1", "B2", "D5", "E6"}

print("        SCREEN THIS WAY")
print()
for row in rows:
    line = f"Row {row}  "
    for seat in range(1, seats_per_row + 1):
        code_ = f"{row}{seat}"
        line += "XX  " if code_ in booked else f"{code_:<4}"
    print(line)

print("\nXX = already booked")
````

````python
# 🔴 INDUSTRY-READY — a store x month sales matrix with row and column totals
# This is the classic "pivot table" report finance teams ask for every month.

months = ["Aug", "Sep", "Oct"]
sales_lakhs = {
    "Delhi":  [12.5, 13.8, 21.0],
    "Mumbai": [18.2, 17.5, 29.4],
    "Pune":   [ 8.1,  9.4, 14.2],
}

print(f"{'Store':<9}" + "".join(f"{m:>8}" for m in months) + f"{'Total':>9}")
print("-" * 36)

column_totals = [0] * len(months)

for store in sales_lakhs:
    row_total = 0
    print(f"{store:<9}", end="")
    for i in range(len(months)):
        value = sales_lakhs[store][i]
        print(f"{value:>8.1f}", end="")
        row_total += value
        column_totals[i] += value
    print(f"{row_total:>9.1f}")

print("-" * 36)
print(f"{'All':<9}" + "".join(f"{t:>8.1f}" for t in column_totals) + f"{sum(column_totals):>9.1f}")
print("\n(figures in Rs lakhs)")
````

## 8️⃣ Classic mistakes (everyone makes these once)

````python
# ❌ Mistake 1 — removing items from a list WHILE looping over it
numbers = [2, 4, 6, 8, 1]
for n in numbers:
    if n % 2 == 0:
        numbers.remove(n)
print("Wrong result :", numbers, "← 4 and 8 are still there! Python skipped them.")

# ✅ Fix — build a NEW list instead of changing the one you are walking through
numbers = [2, 4, 6, 8, 1]
odd_only = []
for n in numbers:
    if n % 2 != 0:
        odd_only.append(n)
print("Right result :", odd_only)
````

````python
# ❌ Mistake 2 — off-by-one with range (stop is NOT included)
print("range(1, 5)  :", list(range(1, 5)), "← missing 5")
print("range(1, 6)  :", list(range(1, 6)), "← correct for 1 to 5")

# ❌ Mistake 3 — using a loop variable after the loop and expecting it to reset
for day in range(1, 8):
    pass
print("After the loop, 'day' is still", day, "— it keeps the LAST value.")

# ❌ Mistake 4 — forgetting that the accumulator must start BEFORE the loop
# for price in prices:
#     total = 0        # resets to 0 every round!  Put it above the loop.
#     total += price
````

## 🚀 Mini-Project — Diwali Order Processing Pipeline

Priya's shop now receives orders online. Each night a script must:

1. **Reject** orders that have no items or an invalid line (zero/negative quantity or price)
2. Add up each valid order, give **10% discount** if the subtotal is ₹2,000 or more, then add **5% GST**
3. Collect **revenue per city**
4. Print a clean report, including why each rejected order failed

This single loop uses almost everything you learned: `for` over a list of dicts, tuple unpacking, an inner `for…else`, `continue`, accumulators and a group-by dict.

````python
orders = [
    {"id": 1001, "customer": "Anita Sharma", "city": "Delhi",
     "items": [("Kaju Katli", 2, 650), ("Soan Papdi", 1, 320)]},
    {"id": 1002, "customer": "Vikram Singh", "city": "Mumbai",
     "items": []},                                             # empty order
    {"id": 1003, "customer": "Sneha Iyer", "city": "Pune",
     "items": [("Laddu Box", 1, 450), ("Gift Hamper", 1, 900)]},
    {"id": 1004, "customer": "Rohit Verma", "city": "Delhi",
     "items": [("Barfi", 0, 400)]},                            # quantity 0
    {"id": 1005, "customer": "Meena Joshi", "city": "Mumbai",
     "items": [("Kaju Katli", 3, 650), ("Gift Box", 2, 150), ("Jalebi", 1, 280)]},
]

GST_RATE = 0.05
DISCOUNT_LIMIT = 2000
DISCOUNT_RATE = 0.10

processed = []
rejected = []
revenue_by_city = {}

for order in orders:
    # Rule 1: an empty order is rejected immediately
    if not order["items"]:
        rejected.append((order["id"], "no items in order"))
        continue

    subtotal = 0
    for name, qty, price in order["items"]:
        if qty <= 0 or price <= 0:
            rejected.append((order["id"], f"invalid line for '{name}'"))
            break                                  # stop checking this order
        subtotal += qty * price
    else:
        # for…else: reached only if NO line caused a break → order is valid
        discount = subtotal * DISCOUNT_RATE if subtotal >= DISCOUNT_LIMIT else 0
        taxable = subtotal - discount
        gst = taxable * GST_RATE
        total = taxable + gst

        processed.append((order["id"], order["customer"], order["city"], total))
        revenue_by_city[order["city"]] = revenue_by_city.get(order["city"], 0) + total

print("=" * 52)
print("  NIGHTLY ORDER REPORT — Priya Mithai Bhandar")
print("=" * 52)

print("\nPROCESSED ORDERS")
for order_id, customer, city, total in processed:
    print(f"  #{order_id}  {customer:<14}{city:<8}Rs {total:>9,.2f}")

print("\nREJECTED ORDERS")
for order_id, reason in rejected:
    print(f"  #{order_id}  ✗ {reason}")

print("\nREVENUE BY CITY")
grand_total = 0
for city, amount in revenue_by_city.items():
    print(f"  {city:<8} Rs {amount:>10,.2f}")
    grand_total += amount

print("-" * 52)
print(f"  Valid orders: {len(processed)}   Rejected: {len(rejected)}   Total: Rs {grand_total:,.2f}")
````

### 🟡 Exercise 5 — Password character counter
Count how many uppercase letters, lowercase letters, digits and symbols the password has.

Expected: `Upper: 2  Lower: 5  Digits: 4  Symbols: 2`

````python
password = "Priya@2026#Ok"
# YOUR CODE HERE
````

````python
# SOLUTION
password = "Priya@2026#Ok"
upper = lower = digits = symbols = 0
for ch in password:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digits += 1
    else:
        symbols += 1
print(f"Upper: {upper}  Lower: {lower}  Digits: {digits}  Symbols: {symbols}")
````

### 🟡 Exercise 6 — Find duplicate employee IDs
Print every ID that appears **more than once** (print each duplicate only once).

Expected: `Duplicate IDs: [102, 105]`

````python
employee_ids = [101, 102, 103, 102, 104, 105, 105, 106, 102]
# YOUR CODE HERE
````

````python
# SOLUTION
employee_ids = [101, 102, 103, 102, 104, 105, 105, 106, 102]
seen = []
duplicates = []
for emp_id in employee_ids:
    if emp_id in seen and emp_id not in duplicates:
        duplicates.append(emp_id)
    seen.append(emp_id)
print("Duplicate IDs:", duplicates)
````


````python
total_emails = 23
BATCH_SIZE = 5
# YOUR CODE HERE
````

````python
# SOLUTION
total_emails = 23
BATCH_SIZE = 5
batch_no = 0
for start in range(1, total_emails + 1, BATCH_SIZE):
    end = min(start + BATCH_SIZE - 1, total_emails)
    batch_no += 1
    print(f"Batch {batch_no}: sending {end - start + 1} emails ({start}-{end})")
````

### 🔴 Exercise 8 — Inventory restock report
For every product whose `stock` is **at or below** its `reorder_level`, print a restock line. Finish with the **total value** (`stock × price`) of everything currently in the warehouse.

Expected:
```
RESTOCK: Soan Papdi (stock 8, reorder at 10)
RESTOCK: Gift Box (stock 5, reorder at 5)
Total warehouse value: Rs 42,010
```

````python
inventory = [
    {"name": "Kaju Katli",  "stock": 40, "reorder_level": 15, "price": 650},
    {"name": "Soan Papdi",  "stock": 8,  "reorder_level": 10, "price": 220},
    {"name": "Laddu Box",   "stock": 30, "reorder_level": 12, "price": 450},
    {"name": "Gift Box",    "stock": 5,  "reorder_level": 5,  "price": 150},
]
# YOUR CODE HERE
````

````python
# SOLUTION
inventory = [
    {"name": "Kaju Katli",  "stock": 40, "reorder_level": 15, "price": 650},
    {"name": "Soan Papdi",  "stock": 8,  "reorder_level": 10, "price": 220},
    {"name": "Laddu Box",   "stock": 30, "reorder_level": 12, "price": 450},
    {"name": "Gift Box",    "stock": 5,  "reorder_level": 5,  "price": 150},
]
warehouse_value = 0
for product in inventory:
    if product["stock"] <= product["reorder_level"]:
        print(f"RESTOCK: {product['name']} (stock {product['stock']}, reorder at {product['reorder_level']})")
    warehouse_value += product["stock"] * product["price"]
print(f"Total warehouse value: Rs {warehouse_value:,}")
````

## 📋 Summary

| Concept | Remember |
|---------|----------|
| `for x in collection` | visit every item once; loop ends by itself |
| `range(start, stop, step)` | numbers on demand; **stop is excluded** |
| `enumerate()` | item + position, no manual counter |
| `zip()` | walk two lists side by side |
| Accumulators | create **before**, update **inside**, use **after** |
| `break` / `continue` | stop the loop / skip this round |
| `for … else` | `else` runs only if there was no `break` |
| Nested loops | inner loop completes for every outer round |
