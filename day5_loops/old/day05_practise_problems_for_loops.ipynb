### Problem D4  🔴 Challenge: Weekly totals (nested loops)

Print the total of each week, then the total of all weeks.

**Expected output**

```text
Week 1: ₹650
Week 2: ₹750
Week 3: ₹140
All weeks: ₹1540
```

````python
weeks = [[200, 150, 300], [100, 400, 250], [80, 60]]
# YOUR CODE HERE
````

<details>
<summary><b>Show solution</b> (try first!)</summary>

```python

for week_number in range(len(weeks)):
    week_total = 0
    for amount in weeks[week_number]:
        week_total += amount
    print(f"Week {week_number + 1}: ₹{week_total}")
    grand_total += week_total
print(f"All weeks: ₹{grand_total}")
```

### Problem E3  🔴 Challenge: PaisaTracker: biggest category

Build a dict of totals per category with a for loop, print each total,  
then print the category with the highest total.

**Expected output**

```text
food: ₹660
travel: ₹115
housing: ₹12000
Biggest category: housing
```

````python
expenses = [
    {"name": "chai", "amount": 20, "category": "food"},
    {"name": "auto-rickshaw", "amount": 85, "category": "travel"},
    {"name": "groceries", "amount": 640, "category": "food"},
    {"name": "metro", "amount": 30, "category": "travel"},
    {"name": "rent", "amount": 12000, "category": "housing"},
]
totals = {}
# YOUR CODE HERE
````

<details>
<summary><b>Show solution</b> (try first!)</summary>

```python
for expense in expenses:
    category = expense["category"]
    if category in totals:
        totals[category] += expense["amount"]
    else:
        totals[category] = expense["amount"]

biggest = None
for category, amount in totals.items():
    print(f"{category}: ₹{amount}")
    if biggest is None or amount > totals[biggest]:
        biggest = category
print(f"Biggest category: {biggest}")
```

</details>
