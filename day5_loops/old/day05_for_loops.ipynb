## SECTION 1: What is a for loop?

### Concept: do the same job for EACH item
Imagine Priya has a shopping list. She walks through the shop and picks up **each** item one by one. A `for` loop does exactly this.

```text
for item in collection:
    do something with item
```
* `item` is the **loop variable**. Python puts the next value in it on every round. You choose the name.
* The line ends with a colon `:`.
* The indented lines (the **loop body**) repeat once for every item.
* When the list is finished, the loop stops by itself. No update line is needed (unlike `while`).

````python
groceries = ["atta", "dal", "rice", "milk"]

for item in groceries:
    print(f"Priya buys: {item}")

print("Shopping list finished!")
````

### Concept: you can loop over more than lists
A string is a row of characters, and a tuple is a fixed list. Both work in a `for` loop.

````python
for letter in "CHAI":
    print(letter)

print()
days = ("Mon", "Tue", "Wed")
for day in days:
    print(f"Arjun works on {day}")
````

## SECTION 2: range() - looping a number of times

### Concept: range() makes a sequence of numbers
`range()` is the tool when you want to repeat something a certain number of times, or count.

| Call | Numbers it gives |
|------|------------------|
| `range(5)` | 0, 1, 2, 3, 4 (starts at 0, stops BEFORE 5) |
| `range(1, 6)` | 1, 2, 3, 4, 5 (start, stop before) |
| `range(0, 11, 2)` | 0, 2, 4, 6, 8, 10 (the third number is the step) |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 (a negative step counts down) |

The stop number is **never** included. `list(range(...))` lets us see the numbers.

````python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 11, 2)))
print(list(range(5, 0, -1)))
````

````python
# Priya's mother makes 3 cups of chai every morning
for cup in range(1, 4):
    print(f"Making chai cup number {cup}")

print()
print("Table of 7")
for multiplier in range(1, 6):
    print(f"7 x {multiplier} = {7 * multiplier}")
````

## SECTION 3: The accumulator pattern

### Concept: build up an answer while looping
Make a variable **before** the loop, then change it **inside** the loop. After the loop, the variable holds your answer. This is how totals, counts, the largest and the smallest are found.

````python
expenses = [120, 45, 300, 80, 250]

total = 0                      # start at zero BEFORE the loop
count = 0
for amount in expenses:
    total += amount            # add this expense to the total
    count += 1                 # count it

average = total / count
print(f"Total: ₹{total}")
print(f"Number of expenses: {count}")
print(f"Average: ₹{round(average, 2)}")
````

### Example: largest and smallest
Start by assuming the first item is both the largest and smallest. Then challenge that guess with every item.

````python
largest = expenses[0]
smallest = expenses[0]
for amount in expenses:
    if amount > largest:
        largest = amount
    if amount < smallest:
        smallest = amount

print(f"Largest expense : ₹{largest}")
print(f"Smallest expense: ₹{smallest}")
````

### Example: counting only some items
Add an `if` inside the loop to count or add only the items you care about.

````python
limit = 100
big_count = 0
big_total = 0
for amount in expenses:
    if amount > limit:
        big_count += 1
        big_total += amount

print(f"{big_count} expenses were above ₹{limit}, adding up to ₹{big_total}")
````

## SECTION 4: Looping over a dictionary

### Concept: keys, values or both
Looping over a dict gives you the **keys** by default. Use `.values()` for the values and `.items()` for key and value together (two loop variables).

````python
budget = {"food": 3000, "travel": 1200, "rent": 12000}

print("Keys:")
for category in budget:
    print(" ", category)

print("Values:")
for limit in budget.values():
    print(" ", limit)

print("Both:")
for category, limit in budget.items():
    print(f"  {category} budget is ₹{limit}")
````

````python
# Adding up all the values of a dict
monthly_total = 0
for limit in budget.values():
    monthly_total += limit
print(f"Total monthly budget: ₹{monthly_total}")
````

## SECTION 5: Looping with an index

### Concept: range(len(...))
Sometimes you need the **position** of each item, for example to print a numbered list or to read two lists side by side. `len(list)` gives the size, and `range(len(list))` gives every valid position.

````python
names = ["chai", "auto-rickshaw", "lunch"]
amounts = [20, 85, 150]

for i in range(len(names)):
    print(f"{i + 1}. {names[i]} - ₹{amounts[i]}")
````

## SECTION 6: break, continue, pass and loop else

* `break` leaves the loop immediately.
* `continue` skips the rest of **this round** and goes to the next item.
* `pass` does nothing. It is a placeholder.
* A loop `else` block runs **only if the loop finished without hitting `break`**.

Unlike `while` loops, a `for` loop moves to the next item by itself, so `continue` is always safe.

````python
# continue: skip zero entries
entries = [200, 0, 150, 0, 400]
total = 0
for amount in entries:
    if amount == 0:
        print("  Skipping a zero entry")
        continue
    total += amount
print(f"Total: ₹{total}")
````

````python
# break: stop at the first suspicious expense
daily_expenses = [40, 120, 90, 2500, 60]
for amount in daily_expenses:
    if amount > 1000:
        print(f"Suspicious expense of ₹{amount}! Stop checking.")
        break
    print(f"  ₹{amount} looks fine")
````

````python
# pass: a placeholder for code we will write later
for amount in daily_expenses:
    if amount < 0:
        pass          # TODO: handle refunds later
print("Loop with pass finished safely")
````

````python
# loop else: runs only when break never happened
categories = ["food", "travel", "rent"]

for wanted in ["rent", "shopping"]:
    for category in categories:
        if category == wanted:
            print(f"Found '{wanted}'")
            break
    else:
        print(f"'{wanted}' is not in the list")
````

## SECTION 7: Nested loops

### Concept: a loop inside a loop
For every round of the **outer** loop, the **inner** loop runs completely. Think of a clock: for every hour (outer), the minutes go all the way around (inner).

````python
for table in range(2, 5):
    for multiplier in range(1, 4):
        print(f"{table} x {multiplier} = {table * multiplier}")
    print("-" * 12)
````

### Example: weekly expenses
Riya keeps her expenses as a list of weeks, and each week is a list of amounts.

````python
weeks = [[200, 150, 300], [100, 400, 250], [80, 60]]

grand_total = 0
for week_number in range(len(weeks)):
    week_total = 0
    for amount in weeks[week_number]:
        week_total += amount
    print(f"Week {week_number + 1}: ₹{week_total}")
    grand_total += week_total
print(f"All weeks: ₹{grand_total}")
````

## SECTION 8: Building a new list in a loop

### Concept: start with an empty list and append
This is the long form of what a list comprehension (Day 4) does in one line. Both give the same result.

````python
expenses = [120, 45, 300, 80, 250]

big_expenses = []
for amount in expenses:
    if amount > 100:
        big_expenses.append(amount)
print("With a loop        :", big_expenses)

print("With a comprehension:", [amount for amount in expenses if amount > 100])
````

## PAISATRACKER STEP: the expense report (v0 "Calculator")

Today PaisaTracker learns to read **all** its expenses. We loop over the list of expense dictionaries from Day 4 and produce a report: a numbered list, the total, the biggest expense, a warning for expenses above a limit, and totals per category.

````python
expenses = [
    {"name": "chai", "amount": 20, "category": "food"},
    {"name": "auto-rickshaw", "amount": 85, "category": "travel"},
    {"name": "groceries", "amount": 640, "category": "food"},
    {"name": "metro", "amount": 30, "category": "travel"},
    {"name": "movie", "amount": 450, "category": "fun"},
    {"name": "rent", "amount": 12000, "category": "housing"},
]
limit = 500

print("PAISATRACKER EXPENSE REPORT")
print("=" * 40)

total = 0
biggest = expenses[0]
for number in range(len(expenses)):
    expense = expenses[number]
    line = f"{number + 1}. {expense['name']} - ₹{expense['amount']} ({expense['category']})"
    if expense["amount"] > limit:
        line = line + "   <-- above limit!"
    print(line)
    total += expense["amount"]
    if expense["amount"] > biggest["amount"]:
        biggest = expense

print("-" * 40)
print(f"Total spent    : ₹{total}")
print(f"Biggest expense: {biggest['name']} (₹{biggest['amount']})")
````

````python
# Totals per category
category_totals = {}
for expense in expenses:
    category = expense["category"]
    if category in category_totals:
        category_totals[category] += expense["amount"]
    else:
        category_totals[category] = expense["amount"]

print("Spending by category:")
for category, amount in category_totals.items():
    print(f"  {category}: ₹{amount}")
````

## SUMMARY

What you learned today:
- `for item in collection:` runs the body once per item and stops by itself
- `range(stop)`, `range(start, stop)` and `range(start, stop, step)` make number sequences (stop is never included)
- Accumulators (total, count, largest, smallest) are created BEFORE the loop and updated INSIDE it
