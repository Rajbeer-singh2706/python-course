## **7: Ask again until the answer is valid (retry)**


**Idea:** people make typing mistakes. A while loop can keep asking **while the answer is not good**.

Helper: `text.isdigit()` is `True` only when the text is made of digits, like `"250"`.
Here we use a **flag** (a True / False variable). `valid = False` means "we do not have a good answer yet".

```python
entries = [200, 0, 150, 0, 400]
index = 0
total = 0

while index < len(entries):
    amount = entries[index]
    index += 1                      # update FIRST, so continue is safe
    if amount == 0:
        print("  Skipping a zero entry")
        continue                    # go straight back to the condition
    total += amount

print(f"Total: ₹{total}")
```

```python
entries = [200, 0, 150, 0, 400]
index = 0
total = 0
```

```python
while index < len(entries):
    amount = entries[index]
    print(f"AMount : {amount}")
    index +=1

    if amount <= 0:
        continue

    total += amount  # total = 200+0

print(f"Final total is : {total}")
```
