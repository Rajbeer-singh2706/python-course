

**Idea:** this is where `while` shines. Arjun saves ₹500 every month. He wants ₹2,000. How many months? We do not calculate, we just keep going until the goal is reached.

```python
savings = 0 
goal = 2000
months = 0 

while savings < goal:
    savings += 500
    months = months + 1
    
print(f"Arjun reaches ₹{savings} in {months} months")
```

```python
expenses = [120, 45, 300]
index = 0 

while index <= len(expenses):  # 0, 1, 2, 3
    print(f"Expense {index + 1}: ₹{expenses[index]}")
    index += 1
```

```python
typed = ["priya"]

name = typed.pop(0)

print("Hello,", name)
print("Items left in typed:", typed)
```

```python
#typed = ["120", "45", "300", "done"]

total = 0 
answer = typed.pop(0)

while answer != 'done':
    amount = int(answer)
    total += amount
    print(f"Added ₹{amount}. Total so far: ₹{total}")
    
    answer = typed.pop(0)

print(f"Final total: ₹{total}")
```

```python
total = 0 
answer = input("Enter your expenses, or done to exit")

while answer != 'done':
    amount = int(answer)
    total += amount
    print(f"Added ₹{amount}. Total so far: ₹{total}")
    
    answer = input("Enter your expenses, or done to exit")

print(f"Final total: ₹{total}")
```
