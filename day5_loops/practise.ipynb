```python
limit = 500
typed = ["1", "chai", "20", "food", 
    "1", "grocercies", "abc","640", "food",
    "2", "3", "9", "4"
]
expenses = []

print("Welcome to PaisaTracker V0")
running = True
while running:
    print()
    print("1) Add Expense => 2) View Expenses => 3) Show Total ==> 4) Exit ")
    choice  = typed.pop(0)
    print(f"you chose: {choice}")

    if choice == "1":
        print("(adding an expsnsed - coming in stage 2)")
        name = typed.pop(0)
        amount_text = typed.pop(0)
        while not (amount_text.isdigit() and int(amount_text) > 0): # retry Loop
            print(f" '{amount_text}' is not valud amount , Try again")
            amount_text = typed.pop(0)
        amount = int(amount_text)
        category = typed.pop(0)
        expenses.append({"name": name,"amount": amount , "category": category})
    elif choice == "2":
        print("showing expsense - coing ins tage 2")
        if len(expenses) == 0:
            print("now expsnes yet")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - R{expense['amount']} ({expense['category']})")
                number+=1
    elif choice == "3":
        print("(showing total - coming in stage 2)")
        total = 0 
        for expense in expenses:
            total += expense['amount']
        print(f"total Spend : {total}")
    elif choice == "4":
        running = False 
        print("Goodbye! (Your data is lost for now. On Day 8 we save it to a file.)")
    else:
        print("INvalid Choice : , please choice of 1 or 2 or 3 of 4")
```
