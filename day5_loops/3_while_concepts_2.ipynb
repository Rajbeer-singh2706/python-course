## **7: Ask again until the answer is valid (retry)**



**Idea:** people make typing mistakes. A while loop can keep asking **while the answer is not good**.

Helper: `text.isdigit()` is `True` only when the text is made of digits, like `"250"`.
Here we use a **flag** (a True / False variable). `valid = False` means "we do not have a good answer yet".

````python
typed = ["abc", "-50", "250"]       # two mistakes, then a good answer
valid = False

while not valid:                    # keep asking while NOT valid
    text = typed.pop(0)
    if text.isdigit() and int(text) > 0:
        valid = True                # good answer: the loop will stop
    else:
        print(f"'{text}' is not a valid amount. Try again.")

amount = int(text)
print(f"Accepted ₹{amount}")
````

## **8: `break` leaves the loop now**


**Idea:** `break` is the **emergency exit door**. The moment Python reaches it, the loop ends, even if the condition is still `True`.

````python
expenses = [40, 120, 90, 2500, 60]
index = 0

while index < len(expenses):
    if expenses[index] > 1000:
        print(f"Suspicious expense: ₹{expenses[index]} at position {index}")
        break                              # leave the loop immediately
    index += 1

print("Checking finished")
````

**What to notice:** the ₹60 at the end was never checked, because `break` stopped the loop at ₹2500.

## **9: `continue` skips this round**



**Idea:** `continue` means "forget the rest of **this** round, go back and check the condition again".

> **Important rule for while loops:** do the **update first**, then use `continue`. If `continue` jumps back before the update runs, the loop gets stuck forever.

````python
entries = [200, 0, 150, 0, 400]
index = 0
total = 0

while index < len(entries):
    amount = entries[index]
    index += 1                      # update FIRST, so continue is safe
    if amount == 0:
        print("  Skipping a zero entry")
        continue                    # go straight back to the condition
    total += amount

print(f"Total: ₹{total}")
````

## **10: `pass` does nothing**



**Idea:** `pass` is a placeholder that means "I will write this part later". Python does not allow an empty block, so `pass` keeps the code valid.

````python
count = 1
while count <= 3:
    if count == 2:
        pass                        # TODO: decide what to do for 2 later
    print(count)
    count += 1
````

**What to notice:** `pass` did nothing at all, so all of 1, 2, 3 were printed.

## **11: Loop `else`**



**Idea:** a loop can have an `else`. It runs **only if the loop ended normally**, without a `break`. Think: *"I searched everywhere and never hit break, so it was not found."*

````python
categories = ["food", "travel", "rent"]
wanted = "shopping"
index = 0

while index < len(categories):
    if categories[index] == wanted:
        print(f"Found '{wanted}' at position {index}")
        break
    index += 1
else:
    print(f"'{wanted}' is not in the list")
````

**Try it:** change `wanted` to `"rent"`. Now `break` happens, so the `else` message does **not** print.

## **12: `while True`**



**Idea:** `True` is always true, so `while True:` never stops **by itself**. We stop it ourselves with `break`. This is used when we want the loop to run **at least once** and we decide to stop somewhere in the middle.

````python
typed = ["chai", "samosa", "stop", "jalebi"]

while True:
    item = typed.pop(0)
    if item == "stop":
        break                       # the ONLY way out
    print("Order taken:", item)

print("Canteen closed. Not served:", typed)
````

**Safety rule:** a `while True` loop **must** have a reachable `break`, or it runs forever.

## **13: `while` or `for`?**



The same job, both ways: add the numbers 1 to 5.

````python
total_for = 0
for number in range(1, 6):
    total_for += number

total_while = 0
number = 1
while number <= 5:
    total_while += number
    number += 1

print("for  :", total_for)
print("while:", total_while)
````

| Situation | Use |
|-----------|-----|
| I have a list or I know the count | `for` |
| Keep going until something happens | `while` |
| Keep asking until the input is valid | `while` |
| A menu that runs until Exit | `while` |

# Summary



### You learned
- A `while` loop repeats **while a condition is True**
- Every while loop has a **start value, a condition and an update**
- No update means an **infinite loop** (stop it with Ctrl + C)
- **Sentinel** loops stop at a special word, **retry** loops ask until valid
- `break` exits, `continue` skips a round (update first!), `pass` does nothing
- Loop `else` runs only when there was **no** `break`
- `while True` needs a `break`
- The **PaisaTracker menu** is a while loop with a `running` flag
