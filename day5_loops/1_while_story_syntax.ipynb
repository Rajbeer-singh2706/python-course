# **PART 1: A story**


### Priya and the never-ending expenses
Priya lives in Mumbai. She wrote a small PaisaTracker program last week. It works, but there is a problem.

She types her expenses into the code, runs the program, and **it ends**. To add another expense, she must edit the code and run it again. Chai in the morning. Run again. Auto-rickshaw. Run again. Lunch. Run again.

*"Why can't my program just keep asking me for expenses until I say stop?"* Priya asks her teacher.

The teacher smiles and shows her a chai stall.

> The chai-wala does not pour **exactly 10** cups every day.
> He keeps pouring **while there are customers waiting**.
> Monday 6 customers. Tuesday 25 customers. He does not know the number in advance. He just checks, again and again: *"Is anyone still waiting? Yes? Pour another cup."*

That is a **while loop**: *keep repeating WHILE something is true.*

### Where else do we meet this in real life?
* An ATM keeps showing the menu **while** you still want more transactions.
* A lift keeps going **while** there are floors left to reach.
* You keep scrolling Instagram **while** you are bored (we have all been there).

### for vs while: the one-minute difference
| `for` loop | `while` loop |
|-----------|--------------|
| "Do this once for **each** item in the list" | "Keep doing this **as long as** a condition is True" |
| You know the items or the count | You do **not** know how many rounds |
| Like handing a sweet to each child in the line | Like pouring chai while customers wait |

# PART 2: The syntax


```python
while condition:
    # lines that repeat (indented by 4 spaces)
    # ...
    # something that changes, so the loop can stop
```

### The three parts of EVERY while loop
| # | Part | Question it answers | Example |
|---|------|---------------------|---------|
| 1 | **Start value** | Where do we begin? | `count = 1` |
| 2 | **Condition** | Should we go around again? | `while count <= 5:` |
| 3 | **Update** | What moves us toward stopping? | `count = count + 1` |

### How Python reads a while loop
```text
        +-------------------+
        |   Check condition |<--------------+
        +---------+---------+               |
          True    |     False               |
   +--------------+---------------+         |
   v                              v         |
 Run the lines inside      Leave the loop   |
 (including the update)    and continue     |
   |                       with the code    |
   +----------------------- below ----------+
```
Python checks the condition **before every round**. If it is `True`, the indented lines run, then Python checks again. The moment it is `False`, the loop ends.

### Three small rules
1. End the `while` line with a colon `:`
2. Indent the lines inside the loop (4 spaces)
3. Always make sure something inside the loop can make the condition `False`

# PART 3: Simple examples


### Example 1: Count from 1 to 5

````python
count = 1                    # 1. start value
while count <= 5:            # 2. condition
    print(count)
    count = count + 1        # 3. update
print("Done!")
````

**Follow the loop round by round:**

| Round | `count` before | Is `count <= 5`? | What happens |
|-------|---------------|------------------|--------------|
| 1 | 1 | True | print 1, count becomes 2 |
| 2 | 2 | True | print 2, count becomes 3 |
| 3 | 3 | True | print 3, count becomes 4 |
| 4 | 4 | True | print 4, count becomes 5 |
| 5 | 5 | True | print 5, count becomes 6 |
| 6 | 6 | **False** | loop ends, "Done!" prints |

**Try it:** change `5` to `10`, run again. Then change the start value to `3`.

### Example 2: Countdown

````python
seconds = 5
while seconds > 0:
    print(seconds)
    seconds = seconds - 1
print("Happy Diwali!")
````

### Example 3: The chai kettle

````python
cups_left = 3
while cups_left > 0:
    print("Pouring a cup of chai")
    cups_left = cups_left - 1
print("The kettle is empty")
````

**Shortcut you will see everywhere:** `count += 1` means `count = count + 1`, and `count -= 1` means `count = count - 1`.
