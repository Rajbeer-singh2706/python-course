## **Concept 1: The infinite loop**



**Idea:** if nothing inside the loop ever makes the condition `False`, the loop runs **forever**. This is the most common beginner bug, and every programmer has done it.

```python
# BROKEN - do NOT run this
cups = 1
while cups <= 3:
    print("Pouring chai...")     # cups never changes!
```

**How to stop a runaway loop:** press the **stop (square) button** in Jupyter, or **Ctrl + C** in a terminal.

**The fix:** add the update line.

````python
### infinite loop program
cups = 1
while cups <= 3:
    print(f"Pouring chai cup {cups}")
````

````python
### valid one 
cups = 1
while cups <= 3:
    print(f"Pouring chai cup {cups}")
    cups += 1                    # the update saves us
````

**Before you run any while loop, ask yourself 3 questions:**
1. What is the start value?
2. What will make the condition `False`?
3. Which line moves us toward `False`?

## **concept 2: Adding up (the accumulator)**



**Idea:** make a variable **before** the loop, and add to it **inside** the loop. After the loop it holds your answer.

````python
number = 1
total = 0                        # starts at zero BEFORE the loop
while number <= 5:
    total = total + number       # add to the total
    number = number + 1
print("Sum of 1 to 5 is", total)
````

**What to notice:** `total` changes 0, 1, 3, 6, 10, 15. We never reset it inside the loop.

## **concept 3: When you do not know the number of rounds**



**Idea:** this is where `while` shines. Arjun saves ₹500 every month. He wants ₹2,000. How many months? We do not calculate, we just keep going until the goal is reached.

````python
savings = 0
goal = 2000
months = 0

while savings < goal:
    savings += 500
    months += 1

print(f"Arjun reaches ₹{savings} in {months} months")
````

## **concept 4: Walking through a list**



**Idea:** a list has numbered positions that **start at 0**. `len(list)` is how many items it holds. We use a counter called `index` to visit each position.

````python
expenses = [120, 45, 300]
index = 0                                    # first position is 0

while index < len(expenses):                 # len(expenses) is 3, so positions 0, 1, 2
    print(f"Expense {index + 1}: ₹{expenses[index]}")
    index += 1
````

**What to notice:** the condition is `index < len(expenses)`, **not** `<=`. If we used `<=`, index would reach 3, and `expenses[3]` does not exist (an `IndexError`).

*(A `for` loop does this more easily. We practise it with `while` because it teaches how loops really work.)*

## **concept 5: Simulating typing**

## **concept 6: Stop when the user says "done" (sentinel)**



**Idea:** a **sentinel** is a special word that means "stop", like a security guard at a gate. Priya will type amounts, and the word `done` ends the loop.

**The pattern:**
1. Ask **once before** the loop.
2. `while` the answer is **not** the sentinel: use the answer.
3. Ask **again** as the **last line** of the loop (this is the update!).

````python
typed = ["120", "45", "300", "done"]       # pretend the user types these

total = 0
answer = typed.pop(0)                      # 1. ask once before the loop
while answer != "done":                    # 2. not the sentinel yet
    amount = int(answer)                   # int() turns the text "120" into the number 120
    total += amount
    print(f"Added ₹{amount}. Total so far: ₹{total}")
    answer = typed.pop(0)                  # 3. ask again (the update)

print(f"Final total: ₹{total}")
````

````python
total = 0 
answer = input("Enter your expenses, or done to exit")

while answer != 'done':
    amount = int(answer)
    total += amount
    print(f"Added ₹{amount}. Total so far: ₹{total}")
    
    answer = input("Enter your expenses, or done to exit")

print(f"Final total: ₹{total}")
````
