## SECTION 1: What is a while loop?

### Concept: repeat WHILE something is true  
A for loop says:  "do this once for EACH item in a collection."  
A while loop says: "keep doing this AS LONG AS a condition is True."  

Real-life analogy: a chai-wala keeps pouring cups WHILE there is  
still chai in the kettle. He does not know in advance how many cups  
it will be. He just checks the kettle again and again.  

Every while loop needs three things:  
  1. a START value        (cups_poured = 1)  
  2. a CONDITION          (while cups_poured <= 3:)  
  3. an UPDATE inside     (cups_poured = cups_poured + 1)  

````python
cups_poured = 1                       # 1. start value
while cups_poured <= 3:               # 2. condition (checked before every round)
    print(f"Pouring chai cup number {cups_poured}")
    cups_poured = cups_poured + 1     # 3. update (moves us toward stopping)
print("Kettle is empty. All cups served!")
````

### Concept: the infinite loop (the number one while-loop bug)  
If you forget the UPDATE, the condition never becomes False and  
the loop runs FOREVER. Here is the broken version. DO NOT run it:  

```python
cups_poured = 1
while cups_poured <= 3:
    print("Pouring chai...")      # cups_poured never changes!

```
If this ever happens to you, press Ctrl + C in the terminal to stop it.  

Checklist before running any while loop:  
  [ ] Did I set a start value?  
  [ ] Can the condition ever become False?  
  [ ] Does something INSIDE the loop move us toward False?  

### Example: counting down  

````python
print("Countdown for Priya's Mumbai local train:")
minutes_left = 3
while minutes_left > 0:
    print(f"  Train arrives in {minutes_left} minute(s)")
    minutes_left -= 1                 # same as: minutes_left = minutes_left - 1
print("  Train is here! Hurry up!")
````

## SECTION 2: Loops when you do NOT know how many rounds

### Concept: the right tool for "keep going until..."  
With a for loop you usually know the collection up front.  
With a while loop the number of rounds depends on what happens  
DURING the loop. Example: Arjun saves ₹500 every month and wants  
to reach ₹5000. How many months? We do not know, so we check each month.  

````python
savings = 0
monthly_saving = 500
goal = 5000
months = 0

while savings < goal:
    savings += monthly_saving         # += means "add to the existing value"
    months += 1
print(f"Arjun reaches ₹{savings} after {months} months")
````

### Example: stay inside a budget  
Priya has a weekly budget of ₹2000. We add her expenses one by one  
and stop BEFORE the budget would be crossed.  
The condition has two parts joined with "and":  

- index < len(expenses): we have not run out of expenses  
- the next expense still fits inside the budget  
Python checks the left part first. If it is False, it never looks  
at the right part, so we never read a position that does not exist.  

````python
expenses = [450, 300, 800, 250, 600]
budget = 2000
spent = 0
index = 0                             # which expense we are looking at

while index < len(expenses) and spent + expenses[index] <= budget:
    spent += expenses[index]
    index += 1

print(f"Priya fitted {index} expenses inside her budget")
print(f"Spent: ₹{spent}  |  Left: ₹{budget - spent}")
if index < len(expenses):
    print(f"Next expense of ₹{expenses[index]} would cross the budget")
````

## SECTION 3: Sentinel loops ("keep going until the user says done")

### Concept: the sentinel value  
A sentinel is a special value that means "stop". Like a security  
guard at a gate, it tells the loop when to end. Here the sentinel  
is the word "done".  

Pattern:  
  1. Ask once BEFORE the loop  
  2. while answer is not the sentinel: use it, then ask again  
  3. The last line inside the loop asks the next question (the update!)  

````python
replies = ["120", "45", "300", "done"]    # simulating what the user types
total = 0

user_input = replies.pop(0)               # simulating input("Amount or 'done': ")
while user_input != "done":
    amount = int(user_input)              # int() turns the text "120" into the number 120
    total += amount
    print(f"Added ₹{amount}. Running total: ₹{total}")
    user_input = replies.pop(0)           # ask again (this is the update step)

print(f"Final total: ₹{total}")
````

## SECTION 4: Retry loops (input validation)

### Concept: loop until you get a good answer  
Users make typing mistakes. A while loop lets us say:  
"Ask again WHILE the answer is not good."  
We use two helpers:  
  text.isdigit()  -> True only if the text is made of digits (like "250")  
  None            -> Python's way of saying "no value yet"  

````python
replies = ["abc", "-50", "0", "250"]      # the user fumbles three times!
amount = None                             # no valid amount yet
attempts = 0

while amount is None:
    text = replies.pop(0)                 # simulating input("Enter amount: ")
    attempts += 1
    if text.isdigit() and int(text) > 0:
        amount = int(text)                # valid! the condition becomes False
    else:
        print(f"  '{text}' is not a valid amount. Try again.")

print(f"Accepted ₹{amount} after {attempts} attempts")
````

## SECTION 5: break, continue and pass

### break: leave the loop immediately  
Think of break as an emergency exit door. The moment Python sees it,  
the loop ends, even if the condition is still True.  

````python
daily_expenses = [40, 120, 90, 2500, 60]
limit = 1000
position = 0

while position < len(daily_expenses):
    amount = daily_expenses[position]
    if amount > limit:
        print(f"Suspicious expense of ₹{amount} at position {position}. Stop checking!")
        break
    print(f"  ₹{amount} looks fine")
    position += 1
````

### continue: skip the rest of THIS round only  
continue means "forget the rest of this round, go back and check  
the condition again."  

WARNING (very important for while loops): if you use continue BEFORE  
the update line, the update is skipped and you get an infinite loop.  
The safe habit: do the update (index += 1) FIRST, then use continue.  

````python
entries = [200, 0, 150, -30, 400]
index = 0
valid_total = 0

while index < len(entries):
    amount = entries[index]
    index += 1                            # update FIRST, so continue is safe
    if amount <= 0:
        print(f"  Skipping invalid entry: {amount}")
        continue                          # jump straight back to the condition
    valid_total += amount

print(f"Total of valid entries: ₹{valid_total}")
````

### pass: do nothing (a placeholder)  
pass is Python's "I will fill this in later". Python does not allow  
an empty block, so pass keeps the code valid until you are ready.  

````python
index = 0
while index < len(entries):
    if entries[index] < 0:
        pass                              # TODO: later we will log negative entries
    index += 1
print("Loop with a pass placeholder finished safely")
````

## SECTION 6: The loop else clause

### Concept: else after a loop  
The else block of a loop runs ONLY IF the loop ended normally  
(the condition became False). If the loop was stopped by break,  
the else block is skipped.  

Think of it as: "if I searched everywhere and never hit break,  
then the thing was not found."  

````python
categories = ["food", "travel", "rent"]

wanted = "rent"
index = 0
while index < len(categories):
    if categories[index] == wanted:
        print(f"Found '{wanted}' at position {index}")
        break
    index += 1
else:
    print(f"'{wanted}' is not in the list")

wanted = "shopping"
index = 0
while index < len(categories):
    if categories[index] == wanted:
        print(f"Found '{wanted}' at position {index}")
        break
    index += 1
else:
    print(f"'{wanted}' is not in the list")
````

## SECTION 7: while True with break

### Concept: an infinite loop ON PURPOSE  
"while True:" never stops by itself, because True is always True.  
We stop it ourselves with break. This is the standard pattern when  
the loop must run at least once and the stop decision happens  
in the MIDDLE of the loop (menus, games, ATMs).  

````python
orders = ["samosa", "chai", "vada pav", "stop", "jalebi"]
order_number = 0

while True:
    item = orders.pop(0)                  # simulating input("Order: ")
    if item == "stop":
        print("Canteen closed for Raj's order. Bye!")
        break                             # the only way out
    order_number += 1
    print(f"Order {order_number}: {item}")

print(f"Items left unserved: {orders}")
````

## SECTION 8: while or for? How to choose

### Rule of thumb  
  Know the items or the count in advance  -> for loop  
  Keep going until a condition changes    -> while loop  

The SAME job (add numbers 1 to 5) both ways:  

````python
total_for = 0
for number in range(1, 6):
    total_for += number

total_while = 0
number = 1
while number <= 5:
    total_while += number
    number += 1

print(f"for loop total  : {total_for}")
print(f"while loop total: {total_while}")
````

Both give the same answer, but the for loop is shorter and safer here.  
Use while only when for does not fit (unknown number of rounds).  

## PAISATRACKER STEP: the menu loop (v0 "Calculator")

Today PaisaTracker gets its heart: a menu that keeps running until  
the user chooses Exit. We use a "running" flag (a True/False variable  
that controls the loop). Every pass shows the menu, takes ONE choice,  
does the job, and comes back to the menu.  

````python
expenses = []                                         # our list of expense dicts
choices = ["1", "1", "2", "3", "9", "4"]              # simulating menu choices
new_entries = [                                       # simulating what the user types
    ("chai", 20, "food"),
    ("auto-rickshaw", 85, "travel"),
]

print("Welcome to PaisaTracker v0")
running = True
while running:
    print()
    print("1) Add expense   2) View expenses   3) Show total   4) Exit")
    choice = choices.pop(0)                           # simulating input("Choose: ")
    print(f"Choose: {choice}")

    if choice == "1":
        name, amount, category = new_entries.pop(0)   # simulating three input() calls
        expenses.append({"name": name, "amount": amount, "category": category})
        print(f"Added: {name} - ₹{amount} ({category})")
    elif choice == "2":
        if len(expenses) == 0:
            print("No expenses yet.")
        else:
            number = 1
            for expense in expenses:
                print(f"{number}. {expense['name']} - ₹{expense['amount']} ({expense['category']})")
                number += 1
    elif choice == "3":
        grand_total = 0
        for expense in expenses:
            grand_total += expense["amount"]
        print(f"Total spent: ₹{grand_total}")
    elif choice == "4":
        running = False                               # the loop will stop at the next check
        print("Goodbye! Your expenses are safe (until Day 8, when we save to a file).")
    else:
        print("Invalid choice. Please type 1, 2, 3 or 4.")
````
