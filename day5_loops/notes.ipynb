## 🧠 Concepts

### 1. The `for` Loop

A `for` loop **iterates over every item in a sequence** — a list, string, range, dictionary, etc.

**Syntax:**
```python
for variable in iterable:
    # body — runs once per item
```

- `variable` takes the value of each item in turn
- The body must be **indented** (4 spaces)
- Python loops over *any iterable* — lists, strings, ranges, dicts, sets, tuples
