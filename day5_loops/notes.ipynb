```python
items = [
    {"name": "Kaju Katli (1 kg)", "price": 800, "qty": 2, "gst": 5},
    {"name": "Soan Papdi (500 g)", "price": 220, "qty": 3, "gst": 5},
    {"name": "Gift Box",          "price": 150, "qty": 1, "gst": 12},
]
```

```python
items
```

```python
subtotal = 0
total_gst = 0
print(f"{'Item':<22}{'Qty':>4}{'Amount':>10}{'GST':>8}")
print("-" * 44)
```

```python
for item in items:
    amount = item['price'] * item['qty']
    gst_amount = amount * item['gst'] / 100 
    subtotal += amount
    total_gst += gst_amount
    print(f"{item['name']:<22}{item['qty']:>4}{amount:>10,.2f}{gst_amount:>8,.2f}")
```

```python
names = ["Priya", "Arjun", "Riya", "Kabir"]
maths = [92, 58, 74, 39]
science = [88, 64, 81, 45]
```

```python
for name,m, s in zip(names, maths, science):
    avg = (m+s) / 2 
    result = "Pass" if m >= 40 and s>= 40 else "Fail" 
    print(f"{name:<7} Maths {m:>3}  Science {s:>3}  Avg {avg:>5.1f}  → {result}")
```

```python
numbers = [4, 9, -2, 15, 7, -8, 30]

print("continue → print only positive numbers:")
for n in numbers:
    if n < 0:
        continue
    print("  ", n)
```

```python
print("\nbreak → stop at the first number above 10:")
for n in numbers:
    if n > 10:
        print("   Found", n, "— stopping")
        break
    print("   checked", n)
```

```python
# 🟡 REAL-WORLD — a customer search using for…else
registered = ["anita@mail.com", "vikram@mail.com", "sneha@mail.com"]

for email_to_find in ["sneha@mail.com", "karan@mail.com"]:
    for email in registered:
        if email == email_to_find:
            print(f"✅ {email_to_find} is already registered")
            break
    else:
        # runs ONLY when the inner loop ended without hitting break
        print(f"🆕 {email_to_find} is new — creating account")
```
